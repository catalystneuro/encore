# Reproducing Watson et al. (2016) on DANDI:000041 and CRCNS:fcx-1

**Paper.** Watson BO, Levenstein D, Greene JP, Gelinas JN, Buzsáki G (2016). Network homeostasis and state dynamics of
neocortical sleep. *Neuron* 90, 839-852. doi:10.1016/j.neuron.2016.03.036.

**Claims.** `claims/watson-2016.yaml`, written from the accepted author manuscript in PubMed Central (PMC4873379) with
its supplement 1 (Table S1, Figures S1 to S5 legends, Supplemental Experimental Procedures) and supplement 2. Every claim
lists both datasets, and each is tested on each, except `fig5c-wake-movement`, tested on CRCNS only as DANDI holds no
motion or EMG signal.

**Datasets.**
- DANDI:000041, version **0.261104.0300** (doi:10.48324/dandi.000041/0.261104.0300), 22 NWB files, streamed with
  remfile and h5py. The paper cites no version. This is the most recent published version; its release notes say that
  all files were re-converted from the lab's source files, correcting the unit table and its cell-type labels, the spike
  times of Dino-072114 and the LFP gain, all of which the analysis uses, so no earlier version describes the paper's data
  better. The draft (modified 2026-11-06) is not used, as it can change.
- CRCNS:fcx-1 (doi:10.6080/K02N506Q), not versioned: the 27 session tar.gz files under `fcx-1/live/data/` of the
  public bucket, each streamed once over parallel range requests and decompressed as it arrives; the MD5 sums from the
  data set's `checksums.md5` of the files read are printed below.

**How the two datasets differ.** CRCNS:fcx-1 holds the lab's own state intervals (WAKE, SLEEP, WAKE-SLEEP cycles,
packets, MAs, REM, WAKE interruptions), its UP/DOWN and spindle detections, and motion; those are used as deposited
(reading a of the claims). DANDI:000041 holds Awake, Non-REM and REM epochs only, so the paper's intervals are built
from them by the rules of Table S1, and UP/DOWN states and spindles are detected here (reading b). Both hold the LFP;
band powers are computed here for both from the lab's spindle/good channel.

**Outcome per claim.** One outcome is given per claim, judged on CRCNS:fcx-1, which holds all 27 sessions, the lab's own cell classification (`_CellIDs`, which gives the paper's 995 pE and 126 pI) and the lab's own state intervals, UP/DOWN states and spindles, so it holds the data the paper's tests used. DANDI:000041 holds 22 of the sessions, cell-type labels that differ from the lab's in 7 of them, and Awake, Non-REM and REM epochs only, from which the paper's intervals are rebuilt here; its result is stated beside each CRCNS result, and where it differs the section says so. This is a choice of this attempt (the protocol gives one outcome per claim and the claims file sets no rule for two datasets). In round 1 a claim was reproduced only when it held on both datasets; the verifier pointed out that the claims file has no such rule and that DANDI's differences come from what it lacks.

## Data check

Counts by subject and session, units, state intervals, events, ranges of times, the channel count of each CRCNS LFP stream against its `.xml`, and a check that spikes, states and UP states share a clock: population spikes rise sharply at UP onsets and not when the onsets are shifted by 30 minutes.

```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
from scipy import stats
import watson as W
DS = ["000041", "crcns-fcx-1"]
STATES = ["wake", "nrem", "rem"]
def pr(*a):
    print(*a)
def fmt(x, n=3):
    return "nan" if x is None or not np.isfinite(x) else f"{x:.{n}g}"
def rp(res):
    return "; ".join(f"{k}: R={fmt(v[0])}, p={fmt(v[1])}, n={v[2]}" for k, v in res.items())
def nancorr(a, b):
    ok = np.isfinite(a) & np.isfinite(b)
    return np.corrcoef(a[ok], b[ok])[0, 1] if ok.sum() > 2 else np.nan
def cells_of(s, kind):
    return s["E"] if kind == "E" else s["I"]

for ds in DS:
    ss = D[ds]
    pr(f"== {ds}: {len(ss)} sessions, rats {sorted(set(s['rat'] for s in ss))}")
    for s in ss:
        I = s["ints"]
        sp = s.get("spindles")
        pr(f"{s['name']:18s} rat={s['rat']:9s} units={len(s['spikes']):3d} pE={len(s['E']):3d} pI={len(s['I']):2d} "
           f"cycles={len(s['cycles'])} packets={len(I['SWSPacket'])} MA={len(I['MA'])} REM={len(I['REM'])} "
           f"UP={len(s.get('up', []))} DOWN={len(s.get('dn', []))} spindles={0 if sp is None else len(sp['start'])} "
           f"LFP={'yes' if 'bp_delta' in s['G'] else 'no'} last spike={max(x.max() for x in s['spikes'] if len(x)):.0f} s "
           f"LFP length={s.get('lfp_seconds', 0):.0f} s")
    allsp = np.concatenate([np.concatenate(s['spikes']) for s in ss])
    pr("spike times: min", allsp.min(), "max", allsp.max(), "s; fraction of times that are whole seconds",
       np.mean(allsp == np.round(allsp)))
    lab = [s["ctype"] for s in ss]
    pr("cell types:", {k: int(sum((l == k).sum() for l in lab)) for k in ["E", "I", "?"]})
    if ds == "crcns-fcx-1":
        pr("channels: .xml against the count found from the .eeg stream where the .xml came later:",
           [(s["name"], s["nch_xml"], s["nch_guess"]) for s in ss if s["nch_guess"]])
        pr("LFP sampling rate in _BasicMetaData:", sorted(set(s["lfp_rate_xml"] for s in ss)))
        pr("LFP length against last state interval end (s):",
           [(s["name"], round(s["lfp_seconds"]), round(max(np.max(np.asarray(c['sleep'])) for c in s['cycles']))) for s in ss][:5], "...")
        pr("spindle amplitude range (uV):", np.percentile(np.concatenate([s['spindles']['amp_uV'] for s in ss]), [0, 50, 100]))
    else:
        pr("LFP rate:", sorted(set(s.get("lfp_rate", np.nan) for s in ss)), "sessions without LFP:", [s["name"] for s in ss if not s.get("have_lfp")])
        pr("regions of units:", {k: int(sum((s['region'] == k).sum() for s in ss)) for k in sorted(set(np.concatenate([s['region'] for s in ss])))})
        pr("electrode locations:", sorted(set(np.concatenate([s['electrode_location'] for s in ss]))))
        pr("LFP starting_time (s), distinct values over files with LFP:", sorted(set(s.get("lfp_starting_time", np.nan) for s in ss if s.get("have_lfp"))),
           "(added to the LFP window times and LFP detections)")
        pr("acquisition series per file (name, shape, rate):", sorted(set(str(s.get("acquisition")) for s in ss)),
           "; the only ElectricalSeries with data is the 1250 Hz LFP (rate above), so no wideband signal for an EMG")
    # clock check: rate of pE cells is higher in WAKE than nonREM? and UP-state onsets align with
    # population spiking, labels shifted by 30 min break the alignment
    on, off = [], []
    for s in ss:
        up = np.asarray(s.get('up', np.zeros((0, 2))))
        if len(up) < 50:
            continue
        mua = np.sort(np.concatenate([s['spikes'][i] for i in s['E']]))
        for shift, lst in [(0.0, on), (1800.0, off)]:
            st = up[:, 0] + shift
            c1 = np.searchsorted(mua, st + 0.1) - np.searchsorted(mua, st)
            c0 = np.searchsorted(mua, st) - np.searchsorted(mua, st - 0.1)
            lst.append((c1.sum() + 1) / (c0.sum() + 1))
    pr("population spikes 0-100 ms after over 100-0 ms before UP onsets, median over sessions:",
       fmt(np.median(on)), "; with UP onsets shifted by 1800 s:", fmt(np.median(off)))
    # LFP clock: delta power in nonREM packets against WAKE, with the state intervals as scored and
    # shifted by 1800 s (a contrast a clock offset between the LFP and the state table would remove)
    on, off = [], []
    for s in ss:
        if "bp_delta" not in s["G"] or not s["cycles"]: continue
        lp = np.log(s["G"]["bp_delta"])
        for shift, lst in [(0, on), (1800, off)]:
            kn = W.secs(W.state_pieces(s, "nrem") + shift); kw = W.secs(W.state_pieces(s, "wake") + shift)
            kn, kw = kn[kn < len(lp)], kw[kw < len(lp)]
            lst.append(np.nanmean(lp[kn]) - np.nanmean(lp[kw]))
    pr("log delta power, nonREM packets minus WAKE, median over sessions:", fmt(np.median(on)), "; with the intervals shifted by 1800 s:",
       fmt(np.median(off)), f"; sessions {len(on)}, with a positive difference {int(np.sum(np.array(on) > 0))}")
```

Recordings deposited more than once, within and across the datasets.

```python
# Recordings deposited more than once: sessions of the same rat and date, within each dataset, and the
# same session across the two datasets (identical spike trains, LFP length).
def same_trains(a, b):
    ka = {(len(x), round(float(x[0]), 5), round(float(x[-1]), 5)) for x in a["spikes"] if len(x)}
    return sum(1 for x in b["spikes"] if len(x) and (len(x), round(float(x[0]), 5), round(float(x[-1]), 5)) in ka)
def key(s):
    n = s["name"].replace("-", "_")
    return "_".join(n.split("_")[:2])
for ds in DS:
    ss = D[ds]
    for i in range(len(ss)):
        for j in range(i + 1, len(ss)):
            if key(ss[i]) == key(ss[j]):
                a, b = ss[i], ss[j]
                c = np.nan
                if "bp_delta" in a["G"] and "bp_delta" in b["G"] and a["G"]["T"] == b["G"]["T"]:
                    c = nancorr(np.log(a["G"]["bp_delta"]), np.log(b["G"]["bp_delta"]))
                pr(f"{ds}: {a['name']} and {b['name']}: identical spike trains {same_trains(a, b)} of {len(a['spikes'])}/{len(b['spikes'])}; "
                   f"LFP bytes {a.get('eeg_bytes')} / {b.get('eeg_bytes')}; good channels {a.get('goodeegchannel')} / {b.get('goodeegchannel')}; "
                   f"correlation of their delta power {fmt(c)}")
    pr(ds, "pairs of sessions with any identical spike train across different dates:",
       sum(1 for i in range(len(ss)) for j in range(i + 1, len(ss)) if key(ss[i]) != key(ss[j]) and same_trains(ss[i], ss[j]) > 0))
cr = {key(s) + ("_ACC" if s["name"].endswith("ACC") else ""): s for s in D["crcns-fcx-1"]}
rows = []
for s in D["000041"]:
    c = cr.get(key(s))
    if c is None:
        rows.append(f"{s['name']}: no CRCNS session"); continue
    st = same_trains(c, s)
    tt = [x for x in cr if x.startswith(key(s)) and x != key(s)]
    extra = "; " + ", ".join(f"{x}: {same_trains(cr[x], s)} identical" for x in tt) if tt else ""
    rows.append(f"{s['name']}: units {len(s['spikes'])} against {len(c['spikes'])} ({c['name']}), identical trains {st}, pE {len(s['E'])}/{len(c['E'])}, pI {len(s['I'])}/{len(c['I'])}{extra}")
pr("DANDI file against the CRCNS session of the same name:\n  " + "\n  ".join(rows))
pr("CRCNS sessions with no DANDI file:", sorted(set(cr) - {key(s) for s in D["000041"]}))
```

**Data check, stated.** The CRCNS files match the paper's description: 11 rats, 27 sessions, 1121 stable units (995 pE, 126 pI) of 1360 clustered, LFP at 1250 Hz, and, as printed below under the claims, Table S1's state durations, UP/DOWN counts and spindle values exactly. Each channel count found from a CRCNS `.eeg` stream equals its `.xml`. DANDI:000041 holds 22 of the 27 sessions: rat B22 (the three 2014052x sessions) and the ACC probes of Dino 061914 and 062014 are not in it, and Dino-061914 has no LFP. Its spike trains are those of CRCNS, train for train, except in Dino-072114 (13 of 16 identical), Dino-072314 and Dino-072414 (10 and 15 units more than CRCNS) and Dino-061814 (one more); in 7 sessions whose spike trains are all identical to CRCNS the pE and pI counts still differ (for example BWRat20-101513, 63 pE and 3 pI against 60 and 6), so its cell-type labels do not all follow `_CellIDs`. The Dino 061914 and 062014 ACC and mPFC folders of CRCNS hold LFP files of the same size with different good channels and no shared spike trains: two probes of one recording, counted as two sessions as Table S1 does. UP onsets are followed by a 50- to 70-fold rise of population spiking, which vanishes when the onsets are shifted by 1800 s, so spikes and the UP/DOWN intervals share a clock; state intervals and rates are consistent with that (WAKE and nonREM rates differ by state as below).

## `wake-sleep-cycles`

> The data hold 54 WAKE-SLEEP episodes (WAKE of at least 7 min followed by SLEEP of at least 20 min), an average of 2.0 per session, and sessions without one were discarded.

Source: Results, Identification of brain states, paragraph 1; Results, Firing pattern and LFP changes across the course of sleep, sentence 1; Table S1. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
for ds in DS:
    ss = D[ds]
    n = [len(s["cycles"]) for s in ss]
    pr(f"== {ds} reading {'a (deposited WakeSleep intervals)' if ds=='crcns-fcx-1' else 'b (built from sleep_states)'}: "
       f"{sum(n)} cycles, {sum(n)/len(ss):.2f} per session, sessions without a cycle: {[s['name'] for s,k in zip(ss,n) if k==0]}")
    if ds == "crcns-fcx-1":
        nb = []
        for s in ss:
            I = s["ints"]
            ep = [("Non-REM", a, b) for a, b in I["SWSPacket"]] + [("REM", a, b) for a, b in I["REM"]]
            ep.sort(key=lambda e: e[1])
            seq, t = [], 0.0
            for l, a, b in ep:
                if a > t + 0.5:
                    seq.append(("Awake", t, a))
                seq.append((l, a, b)); t = b
            seq.append(("Awake", t, max(x.max() for x in s["spikes"] if len(x))))
            lab, a, b = zip(*seq)
            nb.append(len(W.derive_states(np.array(lab), np.array(a), np.array(b))["WakeSleep"]))
        pr(f"   reading b (built from the deposited packets and REM, all else awake): {sum(nb)} cycles, {sum(nb)/len(ss):.2f} per session, "
           f"sessions without: {sum(k==0 for k in nb)}")
```

**Result: reproduced.** CRCNS, reading a (deposited cycles): 51 cycles, 1.89 per session, every session has one; both within tolerance (54 +- 5.4, 2.0 +- 0.2). DANDI, reading b (cycles built from `sleep_states`): 47 cycles, 2.14 per session, and Dino-072114 has none. On CRCNS, cycles built by the same rules from the deposited packets and REM give 47, so the DANDI shortfall comes from rebuilding the intervals.

## `tables1-state-durations`

## `burst-index-by-state`

> The burst index (fraction of spikes with ISIs below 15 ms) is highest in nonREM (median 11.8%), lowest in WAKE (4.0%), with REM (5.5%) slightly above WAKE.

Source: Results, Assessment of firing patterns in different brain states, paragraph 3. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def burst_cell(s, pieces, cells, reading):
    """Burst index per cell over `pieces`, from spike times; an ISI counts only when both spikes lie
    in one interval of the state (round 1 also dropped every ISI that crossed a second of the grid)."""
    P = np.asarray(pieces, float).reshape(-1, 2)
    if reading == "a":
        n, sh = W.isi_counts(s, cells, P)
        return np.where(n > 0, sh / np.maximum(n, 1), np.nan)
    if reading == "b":
        bins = np.array([[t0, t0 + 10] for a, b in P for t0 in np.arange(np.ceil(a), np.floor(b) - 9, 10)]).reshape(-1, 2)
        if len(bins) == 0: return np.full(len(cells), np.nan)
        n, sh = W.isi_counts(s, cells, bins, per_piece=True)
        with np.errstate(invalid="ignore"):
            return np.nanmean(np.where(n > 0, sh / np.maximum(n, 1), np.nan), 1)
    if len(P) == 0: return np.full(len(cells), np.nan)
    out = []
    for i in cells:
        t = s["sorted_spikes"][i]
        iv = np.searchsorted(P[:, 0], t, "right") - 1
        ok = (iv >= 0) & (t < P[np.maximum(iv, 0), 1]); t, iv = t[ok], iv[ok]
        if len(t) < 2: out.append(np.nan); continue
        sh = (np.diff(t) < 0.015) & (iv[1:] == iv[:-1])
        short = np.zeros(len(t), bool); short[1:] |= sh; short[:-1] |= sh
        out.append(short.mean())
    return np.array(out)
def burst_grid_round1(s, pieces, cells):
    """Round 1's reading a, for comparison: ISIs kept only when both spikes fall in the same second."""
    k = W.secs(pieces); out = []
    for i in cells:
        t = s["sorted_spikes"][i]; ki = t.astype(int)
        same = (ki[1:] == ki[:-1]) & np.isin(ki[1:], k)
        isi = np.diff(t)[same]
        out.append((isi <= 0.015).mean() if len(isi) else np.nan)
    return np.array(out)
BI = {}
for ds in DS:
    for r in "abc":
        X = []
        for s in D[ds]:
            if not s["cycles"]: continue
            X.append(np.stack([burst_cell(s, W.state_pieces(s, st), s["E"], r) for st in STATES], 1))
        X = np.concatenate(X) * 100; BI[ds, r] = X
        m = np.nanmedian(X, 0)
        cols = [X[np.isfinite(X[:, i]), i] for i in range(3)]
        kw = stats.kruskal(*cols)
        rs = stats.mannwhitneyu(cols[2], cols[0])
        allv = np.concatenate(cols); rk = stats.rankdata(allv); N = len(allv)
        n_ = [len(c) for c in cols]; mr = [rk[sum(n_[:i]):sum(n_[:i+1])].mean() for i in range(3)]
        z = (mr[2] - mr[0]) / np.sqrt(N * (N + 1) / 12 * (1 / n_[2] + 1 / n_[0])); pd = 2 * stats.norm.sf(abs(z))
        pr(f"== {ds} reading {r}: median burst index WAKE {m[0]:.1f}% nonREM {m[1]:.1f}% REM {m[2]:.1f}%; Kruskal-Wallis p={fmt(kw.pvalue)}; "
           f"REM vs WAKE rank-sum p={fmt(rs.pvalue)}, Dunn z={z:.2f} p={fmt(pd)}; cells with a value {n_}")
    X = np.concatenate([np.stack([burst_grid_round1(s, W.state_pieces(s, st), s["E"]) for st in STATES], 1) for s in D[ds] if s["cycles"]]) * 100
    m = np.nanmedian(X, 0)
    pr(f"   {ds} round 1 grid reading a (ISIs within one second only), for comparison: WAKE {m[0]:.1f}% nonREM {m[1]:.1f}% REM {m[2]:.1f}%")
```

**Result: partially reproduced.** ISIs are now counted from spike times. On both datasets and every reading nonREM > REM > WAKE, the Kruskal-Wallis p is below 1e-37 and REM exceeds WAKE (CRCNS Dunn p = 1.3e-6, 7.7e-6 and 4.9e-6 under readings a, b and c). No reading puts the three medians within tolerance: CRCNS (a) 2.4/6.3/3.6%, (b) 2.2/6.0/3.0%, (c) 4.6/11.5/6.5% (WAKE/nonREM/REM, against 4.0/11.8/5.5); reading c has nonREM within tolerance and WAKE and REM above it. Round 1's grid, which kept only ISIs within one second, gave 5.2/14.3/6.7% on CRCNS, so it inflated the index.

## `fig2c-acg-peaks`

> The average spike autocorrelogram of pE cells peaks near 8 ms in nonREM, with smaller peaks near 125 ms (theta) in REM and WAKE.

Source: Results, Assessment of firing patterns in different brain states, paragraph 3; Figure 2C and legend. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
MAXLAG = 0.5
def acg_cells(s, pieces, cells, maxlag=MAXLAG):
    """Autocorrelogram counts in 1 ms bins to maxlag, spike pairs within the same interval: bins
    [k, k+1) ms, and bins centred on whole ms, [k-0.5, k+0.5) (the convention of CCG-type code)."""
    pieces = np.asarray(pieces).reshape(-1, 2)
    nb = int(round(maxlag * 1000))
    out = np.zeros((len(cells), nb)); outc = np.zeros((len(cells), nb))
    if len(pieces) == 0:
        return out, outc
    for j, i in enumerate(cells):
        t = s["spikes"][i]
        iv = np.searchsorted(pieces[:, 0], t, "right") - 1
        ok = (iv >= 0) & (t < pieces[np.maximum(iv, 0), 1])
        t, iv = t[ok], iv[ok]
        for k in range(1, 400):
            if k >= len(t): break
            d = t[k:] - t[:-k]; same = iv[k:] == iv[:-k]
            m = same & (d < maxlag)
            if not m.any() and (d >= maxlag).all(): break
            np.add.at(out[j], np.minimum((d[m] * 1000).astype(int), nb - 1), 1)
            np.add.at(outc[j], np.minimum(np.rint(d[m] * 1000).astype(int), nb - 1), 1)
    return out, outc
ACG, ACGC = {}, {}
for ds in DS:
    for st in STATES:
        R_ = [acg_cells(s, W.state_pieces(s, st), s["E"]) for s in D[ds] if s["cycles"]]
        ACG[ds, st] = np.concatenate([a for a, _ in R_]); ACGC[ds, st] = np.concatenate([b for _, b in R_])
    for st in STATES:
        A = ACG[ds, st][:, :300]
        ok = A.max(1) > 0
        mean = (A[ok] / A[ok].max(1, keepdims=True)).mean(0)
        lag = np.arange(300) + 0.5
        pk = int(np.argmax(mean))
        loc = [i for i in range(60, 250) if mean[i] >= mean[i - 1] and mean[i] >= mean[i + 1]]
        loc_best = max(loc, key=lambda i: mean[i]) if loc else None
        sm = np.convolve(mean, np.ones(9) / 9, "same")
        loc_s = [i for i in range(60, 250) if sm[i] >= sm[i - 1] and sm[i] >= sm[i + 1]]
        lb = max(loc_s, key=lambda i: sm[i]) if loc_s else None
        pr(f"== {ds} {st}: cells {ok.sum()}, peak of mean ACG at {lag[pk]:.1f} ms; largest local max 60-250 ms at "
           f"{'none' if loc_best is None else lag[loc_best]} ms (9 ms smoothed: {'none' if lb is None else lag[lb]} ms, "
           f"height {sm[lb] if lb else np.nan:.3f} against {sm[60:250].min():.3f} minimum in the window)")
    for st in STATES:
        A = ACGC[ds, st][:, :300]; ok = A.max(1) > 0
        mean = (A[ok] / A[ok].max(1, keepdims=True)).mean(0); mean[0] = 0  # bin 0 holds only half a bin
        pk = int(np.argmax(mean))
        loc = [i for i in range(60, 250) if mean[i] >= mean[i - 1] and mean[i] >= mean[i + 1]]
        lb = max(loc, key=lambda i: mean[i]) if loc else None
        pr(f"   {ds} {st}, bins centred on whole ms: peak of mean ACG at {pk} ms (values at 7-11 ms: "
           f"{', '.join(f'{mean[i]:.3f}' for i in range(7, 12))}); largest local max 60-250 ms at {lb} ms")
```

Only labelling the bin by its start (9 ms) puts it inside. The largest local maximum between 60 and 250 ms is at 113.5 ms in REM and 134.5 ms in WAKE on both, within 125 +- 12.5 ms; with bins centred on whole ms it is at 114 ms in REM and 99 ms in WAKE on CRCNS (113 and 99 ms on DANDI). ## `fig2c-acg-decay`

> The autocorrelogram of pE cells decays faster in nonREM than in WAKE and REM (decay fit exponent medians -1.35, -0.83 and -0.80).

Source: Results, Assessment of firing patterns in different brain states, paragraph 3; Figure 2C. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def decay(A, reading):
    out = []
    for a in A:
        if a.max() <= 0: out.append(np.nan); continue
        a = a / a.max(); pk = int(np.argmax(a))
        end = 100 if reading == "a" else 500
        lag = (np.arange(len(a)) + 0.5)
        sl = slice(pk, end)
        y, x = a[sl], lag[sl]; ok = y > 0
        if ok.sum() < 5: out.append(np.nan); continue
        if reading in ("a", "b"):
            out.append(np.polyfit(np.log(x[ok]), np.log(y[ok]), 1)[0])
        else:
            out.append(np.polyfit(x[ok] / 1000, np.log(y[ok]), 1)[0])
    return np.array(out)
for ds in DS:
    for r in "abc":
        ex = [decay(ACG[ds, st], r) for st in STATES]
        cols = [e[np.isfinite(e)] for e in ex]
        kw = stats.kruskal(*cols)
        pr(f"== {ds} reading {r}: median exponent WAKE {np.median(cols[0]):.2f} nonREM {np.median(cols[1]):.2f} REM {np.median(cols[2]):.2f} "
           f"(n {[len(c) for c in cols]}); Kruskal-Wallis p={fmt(kw.pvalue)}")
```

## `synchrony-skew`

> Population synchrony (fraction of pE cells firing in 50 ms windows) has a skewed distribution; more than 15% of pE cells rarely fire synchronously with others (in 4% of windows) while fewer than 4.2% fire synchronously in at least half of the windows.

Source: Results, Assessment of firing patterns in different brain states, paragraph 4; Figure S2F. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def sync_session(s, pieces):
    E = s["E"]
    if len(E) < 2 or len(np.asarray(pieces).reshape(-1, 2)) == 0: return None
    k25 = []
    for a, b in np.asarray(pieces).reshape(-1, 2):
        k25.append(np.arange(int(np.ceil(a / 0.025)), int(b / 0.025) - 1))
    k25 = np.concatenate(k25)
    T = int(k25.max()) + 3
    act = np.zeros((len(E), T), bool)
    for j, i in enumerate(E):
        b = (s["spikes"][i] / 0.025).astype(int); act[j, b[b < T]] = True
    win = act[:, k25] | act[:, k25 + 1]  # 50 ms windows every 25 ms
    n = win.sum(0)
    fa = (win & (n >= 2)[None, :]).mean(1)  # fires with at least one other pE cell
    fb = win.mean(1)
    return fa, fb
for ds in DS:
    for nm, sel in [("all cycle time", None), ("wake", "wake"), ("nrem", "nrem"), ("rem", "rem")]:
        FA, FB = [], []
        for s in D[ds]:
            if not s["cycles"]: continue
            p = np.array([c["wake"][0:1].tolist() + [c["sleep"][1]] for c in s["cycles"]]) if sel is None else W.state_pieces(s, sel)
            r = sync_session(s, p)
            if r is None: continue
            FA.append(r[0]); FB.append(r[1])
        FA, FB = np.concatenate(FA), np.concatenate(FB)
        pr(f"== {ds} {nm}: n={len(FA)}; (a) with another pE cell: <=4% of windows {100*np.mean(FA<=0.04):.1f}%, >=50% {100*np.mean(FA>=0.5):.1f}%; "
           f"(b) any spike: <=4% {100*np.mean(FB<=0.04):.1f}%, >=50% {100*np.mean(FB>=0.5):.1f}%; skew (a) {stats.skew(FA):.2f}")
```

**Result: reproduced.** Over all cycle time, 62.9% (CRCNS) and 64.7% (DANDI) of pE cells fire with another pE cell in at most 4% of the 50 ms windows, and 0.0% in at least half; the per-cell measure is right-skewed (2.31 and 2.25). Both are beyond the paper's bounds in the stated direction, as the criterion asks, under readings a and b and in each state; the share of rarely synchronous cells is four times the paper's 15%, so the measure is unlikely to be the paper's (a question to a reader is filed).

## `fig1e-up-state-dynamics`

> Within UP states, firing rates of pyramidal cells and interneurons decrease over the course of the UP state, with decreasing E/(E+I) ratio and decreasing pE population synchrony; UP states begin with a surge of gamma power followed by increased sigma power.

Source: Results, Assessment of firing patterns in different brain states, paragraph 4; Figure 1E and legend; Figure S4C. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
NB = 10
def up_bins(s, ups):
    """Per UP state and normalized-time bin: pE rate, pI rate, fraction of pE and of pI cells spiking."""
    E, I = s["E"], s["I"]
    edges = ups[:, :1] + (ups[:, 1:] - ups[:, :1]) * np.linspace(0, 1, NB + 1)[None, :]
    dur = (ups[:, 1] - ups[:, 0]) / NB
    def cnts(cells):
        C = np.zeros((len(cells), len(ups), NB))
        for j, i in enumerate(cells):
            t = s["spikes"][i]
            for b in range(NB):
                C[j, :, b] = np.searchsorted(t, edges[:, b + 1]) - np.searchsorted(t, edges[:, b])
        return C
    CE = cnts(E); CI = cnts(I) if len(I) else None
    rE = CE.sum(0) / len(E) / dur[:, None]
    fE = (CE > 0).mean(0)
    rI = CI.sum(0) / len(I) / dur[:, None] if CI is not None else None
    fI = (CI > 0).mean(0) if CI is not None else None
    return rE, rI, fE, fI
for ds in DS:
    RE, RI, FE, EI = [], [], [], []
    for s in D[ds]:
        if not s["cycles"] or len(s["E"]) < 2: continue
        up = np.asarray(s.get("up", np.zeros((0, 2)))).reshape(-1, 2)
        sl = np.array([c["sleep"] for c in s["cycles"]])
        up = up[((up[:, 0][:, None] >= sl[:, 0]) & (up[:, 1][:, None] <= sl[:, 1])).any(1)]
        if len(up) == 0: continue
        rE, rI, fE, fI = up_bins(s, up)
        RE.append(rE); FE.append(fE)
        if rI is not None:
            RI.append(rI); EI.append(fE / np.where(fE + fI > 0, fE + fI, np.nan))
    out = {}
    for nm, X in [("pE rate", RE), ("pI rate", RI), ("pE synchrony", FE), ("E/(E+I)", EI)]:
        m = np.nanmean(np.concatenate(X), 0)
        R, p = stats.pearsonr(np.arange(NB), m)
        out[nm] = (R, p, m[0], m[-1])
        pr(f"== {ds} {nm}: R={R:.3f} p={fmt(p)}, first bin {m[0]:.3f}, last bin {m[-1]:.3f}, bins {np.round(m, 3).tolist()}")
    # UP-onset-triggered sigma and gamma power (short windows, 100 ms steps), UP states >= 1 s
    trig = {"sigma": [], "gamma": []}
    lags = np.arange(-5, 16)
    for s in D[ds]:
        if "short" not in s or not s["cycles"]: continue
        ts = s["short_t"]; pk = np.concatenate([c["packets"] for c in s["cycles"]])
        inn = np.zeros(len(ts), bool)
        for a, b in pk: inn |= (ts >= a) & (ts < b)
        up = np.asarray(s.get("up", np.zeros((0, 2)))).reshape(-1, 2)
        up = up[(up[:, 1] - up[:, 0]) >= 1.0]
        for band in trig:
            v = np.log(s["short"][band] + 1e-12); v = (v - v[inn].mean()) / v[inn].std()
            j = np.round((up[:, 0] - 0.1) / 0.1).astype(int)
            ok = (j + lags.min() >= 0) & (j + lags.max() < len(v))
            if ok.any(): trig[band].append(v[j[ok][:, None] + lags[None, :]])
    if trig["sigma"]:
        mg = np.concatenate(trig["gamma"]).mean(0); ms = np.concatenate(trig["sigma"]).mean(0)
        post = lags >= 0
        pg = lags[post][np.argmax(mg[post])] * 0.1; ps = lags[post][np.argmax(ms[post])] * 0.1
        pr(f"   UP-onset-triggered (UP >= 1 s, n={len(np.concatenate(trig['sigma']))}): gamma peak at {pg:.1f} s, sigma peak at {ps:.1f} s after onset; "
           f"gamma z {np.round(mg, 2).tolist()}; sigma z {np.round(ms, 2).tolist()} (lags -0.5 to 1.5 s)")
```

**Result: partially reproduced.** On CRCNS, across 10 bins of normalized UP time, the pI rate (R = -0.983), the pE rate (R = -0.293) and pE synchrony (R = -0.447) have negative R and are lower in the last bin than in the first; E/(E+I) has positive R (0.470). In the UP-onset-triggered power (UP states of at least 1 s), sigma peaks in the 0.0 s window and gamma in the 0.1 s window, so gamma does not precede sigma at this resolution. Three of four measures decrease. DANDI gives the same pattern (R = -0.971, -0.719, -0.794 and +0.375).

## `fig2de-fast-before-slow`

> In the early UP state pE cells fire in sequences stable across the first and second halves of SLEEP, and a cell's onset latency is negatively correlated with its firing rate.

Source: Results, Assessment of firing patterns in different brain states, paragraph 5; Figure 2D and 2E and legends. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def first_latency(s, ups, cells, win=0.2):
    L = np.full((len(cells), len(ups)), np.nan)
    for j, i in enumerate(cells):
        t = s["spikes"][i]
        k = np.searchsorted(t, ups[:, 0])
        ok = k < len(t)
        lat = np.full(len(ups), np.nan); lat[ok] = t[k[ok]] - ups[ok, 0]
        lat[(lat > win) | (lat > (ups[:, 1] - ups[:, 0]))] = np.nan
        L[j] = lat
    return L
for ds in DS:
    for half in ["a", "b"]:
        X1, X2, RW, RN = [], [], [], []
        for s in D[ds]:
            if not s["cycles"] or len(s["E"]) < 3: continue
            up = np.asarray(s.get("up", np.zeros((0, 2)))).reshape(-1, 2)
            if len(up) < 20: continue
            if half == "a":
                m1 = np.zeros(len(up), bool); m2 = m1.copy()
                for c in s["cycles"]:
                    mid = (c["sleep"][0] + c["sleep"][1]) / 2
                    m1 |= (up[:, 0] >= c["sleep"][0]) & (up[:, 0] < mid); m2 |= (up[:, 0] >= mid) & (up[:, 0] < c["sleep"][1])
            else:
                mid = (up[0, 0] + up[-1, 1]) / 2; m1 = up[:, 0] < mid; m2 = ~m1
            L = first_latency(s, up, s["E"])
            X1.append(np.nanmean(L[:, m1], 1)); X2.append(np.nanmean(L[:, m2], 1))
            RW.append(W.wake_rate(s, s["E"])); RN.append(W.cell_rates(s, W.state_pieces(s, "nrem"), s["E"])[0])
        X1, X2, RW, RN = map(np.concatenate, (X1, X2, RW, RN))
        ok = np.isfinite(X1) & np.isfinite(X2)
        r12 = stats.pearsonr(X1[ok], X2[ok])
        lat = np.nanmean(np.c_[X1, X2], 1)
        res = []
        for nm, rr in [("WAKE", RW), ("nonREM", RN)]:
            k = np.isfinite(lat) & (rr > 0)
            res.append((nm, stats.pearsonr(np.log10(rr[k]), lat[k])))
        pr(f"== {ds} halves {half}: first vs second half latency R={r12[0]:.3f} p={fmt(r12[1])} n={ok.sum()}; latency vs log10 rate: "
           + "; ".join(f"{nm} R={r[0]:.3f} p={fmt(r[1])}" for nm, r in res))
```

**Result: reproduced.** Per-cell first-spike latencies in the first and second halves of SLEEP correlate (CRCNS R = 0.616, reading a; 0.559 with session halves), and latency correlates negatively with log10 rate (WAKE rate R = -0.129, nonREM rate R = -0.226, both p < 1e-4). DANDI agrees (R = 0.597, -0.081, -0.199).

## `fig2f-rate-group-correlations`

```python
def pair_corrs(s, cells, pieces, bin_=0.1):
    pieces = np.asarray(pieces).reshape(-1, 2)
    edges = np.concatenate([np.arange(a, b - bin_ / 2, bin_) for a, b in pieces]) if len(pieces) else np.zeros(0)
    if len(edges) < 10: return None
    C = np.zeros((len(cells), len(edges)))
    for j, i in enumerate(cells):
        t = s["spikes"][i]
        C[j] = np.searchsorted(t, edges + bin_) - np.searchsorted(t, edges)
    with np.errstate(invalid="ignore", divide="ignore"):
        R = np.corrcoef(C)
    return R
for ds in DS:
    for rd in ["wake sextiles", "nonREM-rate sextiles"]:
        if rd == "wake sextiles":
            W.sextiles(D[ds]); grp = [s["sextile"] for s in D[ds]]
        else:
            rn = [W.cell_rates(s, W.state_pieces(s, "nrem"), s["E"])[0] if s["cycles"] else np.full(len(s["E"]), np.nan) for s in D[ds]]
            flat = np.concatenate(rn); ok = np.isfinite(flat); rk = np.full(len(flat), -1)
            o = np.argsort(flat[ok], kind="stable"); q = np.zeros(ok.sum(), int); q[o] = (np.arange(ok.sum()) * 6) // ok.sum(); rk[ok] = q
            grp, i0 = [], 0
            for r in rn: grp.append(rk[i0:i0 + len(r)]); i0 += len(r)
        M = {(a, b): [] for a in range(6) for b in range(a, 6)}
        for s, g in zip(D[ds], grp):
            if not s["cycles"] or len(s["E"]) < 2: continue
            R = pair_corrs(s, s["E"], W.state_pieces(s, "nrem"))
            if R is None: continue
            n = len(s["E"])
            for i in range(n):
                for j in range(i + 1, n):
                    if g[i] < 0 or g[j] < 0 or not np.isfinite(R[i, j]): continue
                    a, b = sorted((g[i], g[j])); M[a, b].append(R[i, j])
        mean = {k: np.mean(v) if v else np.nan for k, v in M.items()}
        pv = {k: stats.wilcoxon(v).pvalue if len(v) > 5 else np.nan for k, v in M.items()}
        top = max(mean, key=lambda k: mean[k])
        pr(f"== {ds} {rd}: fastest-fastest {mean[5,5]:.4f} (p={fmt(pv[5,5])}, n={len(M[5,5])}), fastest-slowest {mean[0,5]:.4f} (p={fmt(pv[0,5])}, n={len(M[0,5])}), "
           f"slowest-slowest {mean[0,0]:.4f} (p={fmt(pv[0,0])}, n={len(M[0,0])}); largest mean at groups {top}")
        pr("   matrix (rows/cols lowest to highest):", np.array([[mean[tuple(sorted((a, b)))] for b in range(6)] for a in range(6)]).round(4).tolist())
```

## `post-sleep-wake-lower`

> pE rates in the 5 min of WAKE after SLEEP are lower than in the 5 min of WAKE before.

Source: Results, Firing pattern and LFP changes across the course of sleep, paragraph 1. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def wake_windows(s, c, reading):
    """Last 5 min of the cycle's WAKE, and the first 5 min of WAKE after its SLEEP: (i) the next scored
    WAKE interval; (ii) the arousal of more than 120 s that ends the SLEEP."""
    pre = np.array([[max(c["wake"][0], c["wake"][1] - 300), c["wake"][1]]])
    Wk = np.asarray(s["ints"]["Wake"]).reshape(-1, 2)
    if reading == "i":
        nxt = Wk[Wk[:, 0] >= c["sleep"][1] - 2]
        if not len(nxt): return pre, None
        a = nxt[0, 0]
        return pre, np.array([[a, min(a + 300, nxt[0, 1])]])
    a = c["sleep"][1]
    # arousal after SLEEP: time until the next sleep state
    nxt_sleep = np.concatenate([np.asarray(s["ints"][k]).reshape(-1, 2) for k in ["SWSPacket", "REM"]])
    nxt_sleep = nxt_sleep[nxt_sleep[:, 0] > a + 1]
    end = nxt_sleep[:, 0].min() if len(nxt_sleep) else max(x.max() for x in s["spikes"] if len(x))
    if end - a <= 120: return pre, None
    return pre, np.array([[a, min(a + 300, end)]])
PW = {}
for ds in DS:
    for rr in ["i", "ii"]:
        for m1 in "ab":
            X, Y = [], []
            for s in D[ds]:
                xs, ys = [], []
                for c in s["cycles"]:
                    pre, post = wake_windows(s, c, rr)
                    if post is None: continue
                    xs.append(W.cell_rates(s, pre, s["E"])[0]); ys.append(W.cell_rates(s, post, s["E"])[0])
                if not xs: continue
                if m1 == "a": X += xs; Y += ys
                else: X.append(np.mean(xs, 0)); Y.append(np.mean(ys, 0))
            X, Y = np.concatenate(X), np.concatenate(Y)
            PW[ds, rr, m1] = (X, Y)
            pr(f"== {ds} WAKE-after {rr}, M1 {m1}: n={len(X)} pre mean {X.mean():.3f} median {np.median(X):.3f}, post mean {Y.mean():.3f} median {np.median(Y):.3f} Hz; "
               f"signed-rank p two {fmt(stats.wilcoxon(X, Y).pvalue)} one {fmt(stats.wilcoxon(X, Y, alternative='greater').pvalue)}; "
               f"rank-sum two {fmt(stats.mannwhitneyu(X, Y).pvalue)} one {fmt(stats.mannwhitneyu(X, Y, alternative='greater').pvalue)}")
```

**Result: reproduced.** CRCNS, with the arousal that ends SLEEP as the WAKE after (reading ii), pE rates fall from 1.261 to 1.135 Hz (signed-rank p = 3.1e-11 by cell and cycle; 1.7e-6 two-sided, 8.3e-7 one-sided by cell), below the 1e-6 bound by cell and cycle; with the next scored WAKE (reading i) p = 0.018. DANDI: the decrease holds under every reading with two-sided signed-rank p between 2.4e-5 and 1.1e-3, not below 1e-6. Rank-sum tests are weaker: the one-sided rank-sum p is 0.027 on CRCNS under reading ii with one entry per cell and cycle, and 0.055 to 0.51 under the other readings on both datasets.

## `fig4e-late-packet-sigma`

> UP states late in nonREM packets have more sigma power than UP states early in packets.

Source: Results, Firing rate and LFP changes within nonREM episodes and packets, paragraphs 3 and 6; Figure 4E; Figure S4B and S4D and legends. The criterion and its readings are in the claims file; the code below runs each reading it names.

```python
def up_sigma(s, up, reading):
    ts = s["short_t"]; v = np.log(s["short"]["sigma"] + 1e-12)
    k = W.secs(W.state_pieces(s, "nrem")); inn = np.isin(ts.astype(int), k)
    v = (v - v[inn].mean()) / v[inn].std()
    out = []
    for a, b in up:
        if reading == "a":
            m = (ts >= a) & (ts < b)
        else:
            m = (ts >= a) & (ts < a + 1.0)
        out.append(v[m].mean() if m.any() else np.nan)
    return np.array(out)
for ds in DS:
    for rd in ["a", "b"]:
        E_, L_ = [], []
        for s in D[ds]:
            if "short" not in s or not s["cycles"]: continue
            up = np.asarray(s.get("up", np.zeros((0, 2)))).reshape(-1, 2)
            up = up[(up[:, 1] - up[:, 0]) > (0.5 if rd == "a" else 0.999)]
            for p in W.state_pieces(s, "nrem"):
                L = (p[1] - p[0]) / 3
                e = up[(up[:, 0] >= p[0]) & (up[:, 1] <= p[0] + L)]; l = up[(up[:, 0] >= p[1] - L) & (up[:, 1] <= p[1])]
                if len(e): E_.append(up_sigma(s, e, rd))
                if len(l): L_.append(up_sigma(s, l, rd))
        E_, L_ = np.concatenate(E_), np.concatenate(L_)
        E_, L_ = E_[np.isfinite(E_)], L_[np.isfinite(L_)]
        r = stats.mannwhitneyu(L_, E_)
        pr(f"== {ds} reading {rd}: UP-state sigma (z) first third {np.mean(E_):.3f} (n={len(E_)}), last third {np.mean(L_):.3f} (n={len(L_)}); rank-sum p={fmt(r.pvalue)}")
# S4B per-pixel map, coarse: UP states over 500 ms, 5 bins of normalized UP time x the two short-window
# bands (z-scored log power over nonREM); Welch t-test of last-third against first-third UP states
# per pixel. The 200 ms windows give 5 Hz resolution, so the map has two bands, not a spectrum.
def up_map(s, up, nb=5):
    ts = s["short_t"]; k = W.secs(W.state_pieces(s, "nrem")); inn = np.isin(ts.astype(int), k)
    Z = []
    for b_ in ["sigma", "gamma"]:
        v = np.log(s["short"][b_] + 1e-12); Z.append((v - v[inn].mean()) / v[inn].std())
    out = np.full((len(up), 2, nb), np.nan)
    i0 = np.searchsorted(ts, up[:, 0]); i1 = np.searchsorted(ts, up[:, 1])
    for u, (a, b) in enumerate(up):
        idx = np.arange(i0[u], i1[u])
        if len(idx) == 0: continue
        bi = np.minimum(((ts[idx] - a) / (b - a) * nb).astype(int), nb - 1)
        for j in range(2):
            out[u, j] = [Z[j][idx[bi == q]].mean() if (bi == q).any() else np.nan for q in range(nb)]
    return out
for ds in DS:
    E_, L_ = [], []
    for s in D[ds]:
        if "short" not in s or not s["cycles"]: continue
        up = np.asarray(s.get("up", np.zeros((0, 2)))).reshape(-1, 2); up = up[(up[:, 1] - up[:, 0]) > 0.5]
        for p in W.state_pieces(s, "nrem"):
            L = (p[1] - p[0]) / 3
            e = up[(up[:, 0] >= p[0]) & (up[:, 1] <= p[0] + L)]; l = up[(up[:, 0] >= p[1] - L) & (up[:, 1] <= p[1])]
            if len(e): E_.append(up_map(s, e))
            if len(l): L_.append(up_map(s, l))
    E_, L_ = np.concatenate(E_), np.concatenate(L_)
    for j, b_ in enumerate(["sigma 9-25 Hz", "gamma 40-100 Hz"]):
        cells_ = []
        for q in range(5):
            x, y = L_[:, j, q], E_[:, j, q]; x, y = x[np.isfinite(x)], y[np.isfinite(y)]
            t = stats.ttest_ind(x, y, equal_var=False)
            cells_.append(f"t={t.statistic:.1f} p={fmt(t.pvalue)}")
        pr(f"== {ds} S4B map, {b_}, late minus early by fifth of UP time: " + " | ".join(cells_))
```

**Result: reproduced.** UP states in the last third of packets have more sigma power than in the first third: CRCNS mean z 0.086 against -0.095 (reading a, rank-sum p below 1e-300), and 0.161 against -0.066 under reading b (p = 3.2e-33). In the coarse S4B map (5 bins of UP time, two bands), sigma is higher late in every bin (t = 9.3 to 14.2) and gamma in four of five. DANDI agrees (p = 1.3e-33 and 1.9e-15).

```python
# [claim] M6; [paper] Supplemental Experimental Procedures, Spiking metrics: triplets of nonREM packets,
# REM episodes, MAs and WAKE; nonREM-REM-nonREM triplets kept unless a WAKE (> 120 s) follows the REM
# before the second packet. Sequences are built within each SLEEP of a cycle from the packets (N),
# MAs (M), REM (R) and WAKE interruptions of 40-120 s (W).
def sequence(c):
    ev = [("N", p) for p in c["packets"]] + [("M", m) for m in c["ma"]] + [("R", r) for r in c["rem"]] + [("W", w) for w in c["wi"]]
    ev.sort(key=lambda e: e[1][0])
    return ev
def triplets(s, kind, wake_reading="b"):
    out = []
    for ci, c in enumerate(s["cycles"]):
        ev = sequence(c)
        L = [e[0] for e in ev]
        if kind == "MNM":
            for i in range(len(ev) - 2):
                if L[i:i + 3] == ["M", "N", "M"]: out.append((ev[i][1], ev[i + 1][1], ev[i + 2][1]))
        elif kind == "RNR":
            idx = [i for i, l in enumerate(L) if l == "R"]
            for a, b in zip(idx[:-1], idx[1:]):
                mid = L[a + 1:b]
                if mid and all(x in "NM" for x in mid) and "N" in mid:
                    nn = [ev[k][1] for k in range(a + 1, b) if L[k] == "N"]
                    out.append((ev[a][1], np.array([nn[0][0], nn[-1][1]]), ev[b][1]))
        elif kind in ("NMN", "NWN"):
            for i in range(len(ev) - 2):
                if L[i:i + 3] == ["N", kind[1], "N"]: out.append((ev[i][1], ev[i + 1][1], ev[i + 2][1]))
        elif kind == "NRN":
            for i in range(len(ev)):
                if L[i] != "N": continue
                j = i + 1
                if j < len(L) and L[j] == "R":
                    k = j + 1
                    while k < len(L) and L[k] in "MWR": k += 1
                    if k < len(L) and L[k] == "N":
                        r = [ev[q][1] for q in range(j, k) if L[q] == "R"]
                        out.append((ev[i][1], np.array([r[0][0], r[-1][1]]), ev[k][1]))
    if kind == "NWN" and wake_reading == "a":
        out = []
        for c0, c1 in zip(s["cycles"][:-1], s["cycles"][1:]):
            if len(c0["packets"]) and len(c1["packets"]):
                out.append((c0["packets"][-1], c1["wake"], c1["packets"][0]))
    return out
def triplet_changes(ds, kind, **kw):
    """Per (cell, triplet): sextile, rate n, rate n+1, WAKE rate, session, triplet index."""
    rows = []
    for si, s in enumerate(D[ds]):
        for ti, (a, b, c) in enumerate(triplets(s, kind, **kw)):
            r0 = W.cell_rates(s, [a], s["E"])[0]; r1 = W.cell_rates(s, [c], s["E"])[0]
            for j in range(len(s["E"])):
                rows.append([s["sextile"][j], r0[j], r1[j], s["wake_rate_E"][j], si, ti, j])
    return np.array(rows).reshape(-1, 7)
def sextile_tests(X, label):
    out = []
    for q in range(5, -1, -1):
        m = X[:, 0] == q; d = X[m, 2] - X[m, 1]
        # M1 b: one value per cell, its changes averaged over triplets
        keys = X[m][:, [4, 6]]
        if len(d) < 3: out.append(f"S{q+1}: n={len(d)}"); continue
        _, inv = np.unique(keys, axis=0, return_inverse=True)
        db = np.bincount(inv.ravel(), d) / np.bincount(inv.ravel())
        out.append(f"S{q+1} (1 lowest): mean {d.mean():+.3f} Hz n={len(d)} signed-rank p={fmt(stats.wilcoxon(d).pvalue)} t p={fmt(stats.ttest_1samp(d, 0).pvalue)}; "
                   f"per cell n={len(db)} mean {db.mean():+.3f} p={fmt(stats.wilcoxon(db).pvalue)}")
    pr(f"   {label}:\n     " + "\n     ".join(out))
def s5_corr(X):
    ok = (X[:, 1] > 0) & (X[:, 2] > 0) & (X[:, 3] > 0)
    Y = X[ok]; d = np.log10(Y[:, 2]) - np.log10(Y[:, 1])
    _, inv = np.unique(Y[:, [4, 6]], axis=0, return_inverse=True); inv = inv.ravel()
    db = np.bincount(inv, d) / np.bincount(inv); wr = np.bincount(inv, Y[:, 3]) / np.bincount(inv)
    ra = stats.pearsonr(np.log10(wr), db); rb = stats.pearsonr(wr, db)
    return f"log change vs log10 WAKE rate R={ra[0]:.3f} p={fmt(ra[1])}; vs linear WAKE rate R={rb[0]:.3f} p={fmt(rb[1])}; cells {len(db)} (dropped {int((~ok).sum())} cell-triplets with a zero rate)"
TR = {}
for ds in DS:
    for kind in ["MNM", "RNR", "NMN", "NRN"]:
        TR[ds, kind] = triplet_changes(ds, kind)
    for wr in "ab":
        TR[ds, "NWN" + wr] = triplet_changes(ds, "NWN", wake_reading=wr)
    pr(ds, "triplets:", {k: int(sum(len(triplets(s, k)) for s in D[ds])) for k in ["MNM", "RNR", "NMN", "NRN"]},
       "NWN (a scored WAKE):", sum(len(triplets(s, "NWN", "a")) for s in D[ds]), "NWN (b arousal 40-120 s):", sum(len(triplets(s, "NWN", "b")) for s in D[ds]))
```
