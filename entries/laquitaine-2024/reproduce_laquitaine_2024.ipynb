### Recordings deposited more than once

The protocol asks for a check of duplicate deposits. Each sorting's spike trains are hashed and
compared with every other; the dense probe sortings of depths 1 and 2 are compared explicitly.
The ground truth spike trains of the raw and fitted files of one simulation are expected to be the
same (one simulation, fitted traces), and the comparison checks that too.

```python
import hashlib

def train_hash(trains):
    h = hashlib.sha1()
    for t in trains:
        h.update(np.round(np.asarray(t), 6).tobytes())
    return h.hexdigest()

hashes = collections.defaultdict(list)
for (e, s), (df, tr) in SORT.items():
    hashes[train_hash(tr)].append(f"{e}/{s}")
dups = [v for v in hashes.values() if len(v) > 1]
print("identical sortings (10 min):", dups if dups else "none")
for s in SORTERS:
    a, b = SORT[("DS1", s)][1], SORT[("DS2", s)][1]
    same = len(a) == len(b) and all(len(x) == len(y) and np.array_equal(x, y) for x, y in zip(a, b))
    print(f"  {s}: DS1 {len(a)} units, DS2 {len(b)} units, identical spike trains: {same}")

# ground truth of the raw and fitted files
for raw, fit in [("sub-001/sub-001_ecephys.nwb", "sub-001-fitted/sub-001-fitted_ecephys.nwb"),
                 ("sub-004/sub-004_ecephys.nwb", "sub-004-fitted/sub-004-fitted_ecephys.nwb")]:
    _, t1 = sb.read_units(raw, columns=["unit_name"])
    _, t2 = sb.read_units(fit, columns=["unit_name"])
    print(raw.split("/")[0], "and", fit.split("/")[0], "ground truth identical:", train_hash(t1) == train_hash(t2))
# ground truth of DS1 and DS2 differ (different neurons near the probe)
print("DS1 and DS2 ground truth: ", len(GT["DS1"][0]), "and", len(GT["DS2"][0]), "units")
```

### Units, layers and clocks

Checks of the metadata against the values: the firing rates the sorting tables store against
the spike counts, the sorted units' layers against the electrode tables, and the clock of the
sortings against the ground truth. The spike times of a sorting and of its ground truth share a
clock when the agreement scores are far above those obtained after shifting the sorted spike times
(a shift of 50 ms breaks every true match). There is no trial table; the evoked stimulus schedule
is not in the files (see the header).

```python
# stored firing rates against counts over 600 s
for e, s in [("NS", "KS4"), ("E", "KS4"), ("M", "KS4"), ("H1", "KS4")]:
    df, tr = SORT[(e, s)]
    if "firing_rates" in df:
        d = np.abs(df["firing_rates"].values - df["rate"].values)
        print(f"{e}/{s}: stored firing_rates vs spike count / 600 s, max abs difference {d.max():.4f} spikes/s")
# a sorted unit's layer is the layer of its contact (the extremum channel)
for e, s in [("NS", "KS4"), ("E", "KS4"), ("M", "KS4"), ("H1", "KS4"), ("DS1", "KS4")]:
    df, tr = SORT[(e, s)]
    rec = {"H1": "H1", "DS1": "DS1"}.get(e, e)
    el = ELEC[rec]
    c = df["contact"].values
    if c.max() < len(el):
        how, lay = "row index", el["layers"].values[c]
    else:  # contacts numbered as the channel names
        how, lay = "channel name", el.set_index("channel_name")["layers"].reindex(c.astype(str)).values
    agree = (sb.std_layer(lay) == df["layer"].values).mean()
    print(f"{e}/{s}: contacts {c.min()} to {c.max()} read as {how}; layer of the unit equals the "
          f"layer of its contact for {agree:.3f} of units")

# clock check: agreement with true and shifted sorted spike times
def best_scores(exp_gt, sort_key, shift=0.0, singles_only=False, gt=None):
    gdf, gtr = gt if gt is not None else GT[exp_gt]
    sdf, st = SORT[sort_key]
    if singles_only:
        keep = sdf.single.values
        sdf, st = sdf[keep], [t for t, k in zip(st, keep) if k]
    st = [t + shift for t in st]
    fs = sb.SAMPLING_HZ[exp_gt]
    return sb.agreement_scores(gtr, gdf.index, st, sdf.index, fs)

def coincident_fraction(gt_trains, s_trains, shift=0.0, window=0.00065):
    """Fraction of sorted spikes within +-window s of any ground truth spike."""
    allg = np.sort(np.concatenate(gt_trains))
    t = np.concatenate(s_trains) + shift
    i = np.clip(np.searchsorted(allg, t), 1, len(allg) - 1)
    d = np.minimum(np.abs(allg[i] - t), np.abs(allg[i - 1] - t))
    return (d < window).mean()

CLOCK = []
for e, g in [("NS", "NS"), ("E", "E"), ("DS1", "DS1"), ("DS1", "DS2"), ("DS2", "DS2"), ("DS3", "DS3")]:
    gdf, gtr = GT[g]
    sdf, st = SORT[(e, "KS4")]
    s0 = best_scores(g, (e, "KS4")).max(axis=1)
    s1 = best_scores(g, (e, "KS4"), shift=0.05).max(axis=1)
    CLOCK.append({"sorting": f"{e}/KS4", "ground truth": g,
                  "units best >= 0.5": int((s0 >= 0.5).sum()), "same, shifted 50 ms": int((s1 >= 0.5).sum()),
                  "coincident spikes": round(coincident_fraction(gtr, st), 3),
                  "same, shifted 50 ms ": round(coincident_fraction(gtr, st, 0.05), 3)})
CLOCK = pd.DataFrame(CLOCK)
print(CLOCK.to_string(index=False))

# the same coincidence check for every sorter (verifier round 1 asked for the clock of every sorting)
rows = []
for e, g in [("NS", "NS"), ("E", "E"), ("DS2", "DS2"), ("DS3", "DS3")]:
    for s in SORTERS:
        _, gtr = gt_for(g, s)
        st = SORT[(e, s)][1]
        rows.append({"sorting": f"{e}/{s}", "coincident": round(coincident_fraction(gtr, st), 3),
                     "shifted 50 ms": round(coincident_fraction(gtr, st, 0.05), 3)})
CLOCK_ALL = pd.DataFrame(rows)
CLOCK_ALL["excess"] = CLOCK_ALL["coincident"] - CLOCK_ALL["shifted 50 ms"]
print(CLOCK_ALL.to_string(index=False))

# DS3: coincidence of the Kilosort4 spikes with each dense probe ground truth over lags of -100 to
# +100 ms (0.5 ms steps), to tell a time offset from a mismatched file
# (verifier round 2: the scan now covers every sorter, not Kilosort4 alone)
lags = np.round(np.arange(-0.1, 0.1001, 0.0005), 4)
rows = []
for s in SORTERS:
    for g in ["DS1", "DS2", "DS3"]:
        cf = np.array([coincident_fraction(GT[g][1], SORT[("DS3", s)][1], l) for l in lags])
        rows.append({"sorting": f"DS3/{s}", "ground truth": g, "at lag 0": round(cf[lags == 0][0], 3),
                     "largest": round(cf.max(), 3), "at (ms)": round(1000 * lags[cf.argmax()], 1),
                     "median over lags": round(np.median(cf), 3),
                     "largest - median": round(cf.max() - np.median(cf), 3)})
LAGSCAN = pd.DataFrame(rows)
print(LAGSCAN.to_string(index=False))

# which evoked simulation the E sortings belong to: the E Kilosort4 sorting against the ground truth
# of the 1-hour (E) and of the 10-minute 40 kHz (E40) recordings
for g in ["E", "E40"]:
    s0 = best_scores(g, ("E", "KS4"), gt=GT[g]).max(axis=1)
    s1 = best_scores(g, ("E", "KS4"), shift=0.05, gt=GT[g]).max(axis=1)
    print(f"E/KS4 against {g} ground truth: units with a best score >= 0.5: {int((s0 >= 0.5).sum())} "
          f"(shifted 50 ms: {int((s1 >= 0.5).sum())}); coincident spikes {coincident_fraction(GT[g][1], SORT[('E', 'KS4')][1]):.3f}")
```

The first table above is the clock check. For NS, E and DS2 (and the DS1 folder's sortings against
the DS2 ground truth) many sorted units reach an agreement of 0.5 and none does after the 50 ms
shift, so their sortings share the ground truth's clock. The DS1 folder's sortings match the DS1
ground truth no better than chance, as they are DS2's. The second table extends the coincidence
check to every sorter: a sorting shares the clock when its coincidence rate drops after the shift.
For DS3 no Kilosort4 unit reaches 0.5 and the coincidence rate barely exceeds the shifted control for
any sorter. The third table scans lags from -100 to +100 ms for every DS3 sorting against each of the
three dense probe ground truths: the largest coincidence rate is at most 0.011 above the median over
lags (for DS3 against its own ground truth, at 0 or -0.5 ms), against excesses of 0.18 to 0.53 for
the DS2 sortings at lag 0 over the shifted control, so the DS3 sortings could not be paired with the DS3 ground truth (neither an offset nor a swap of
depths explains them). DS3 scores below enter only claims that are blocked or reported, with this
caveat. The Kilosort2 sorting of E ends at 240 s (table of last spikes), as the Methods say it sorted
4 minutes, so its rates and scores use 240 s. The E Kilosort4 sorting matches the ground truth of the
1-hour recording (E) far better than that of the 10-minute 40 kHz one (E40), so it is a sorting of E.
There are no trials, so no stimulus contrast is possible; for the traces, the layer profiles below
are the only check.


## Traces: shared statistics

Six blocks of 10 s spread over the first 10 minutes of each recording (onsets printed below) are
streamed, high-pass filtered at 300 Hz and re-referenced to the common median, and the cortical
sites kept (Outside and WM dropped). Keeping the full 10 minutes of every recording in memory does
not fit the run, so the medians, noise, spectra and templates below rest on 60 s per recording, a
deviation from the claims' 10 minutes. The maxima and ANR extremes, which grow with duration, are
also taken over the whole first 600 s, streamed in blocks (next section). The dense probe and Horvath depths are
pooled by layer. Units: the simulated and Marques-Smith files store microvolts (conversion 1e-6 to
volts); the Horvath files store int16 counts with `conversion` 0.195 (which NWB reads as volts per
count) and an electrode column `gain_to_uV` of 195000, which cannot both be right and give
implausible values either way. Two scales are therefore carried: H in counts (`H`), and H at 0.195
uV per count (`H195`, the Intan step). The dense probe model was fitted to Horvath's noise (Methods,
Fitting amplitude and background noise), and the check below shows the fitted model's noise matches
H in counts, not at 0.195 uV per count; the comparisons of DS with H use counts, as the fit implies
(an assumption), and show H195 as a variant.

```python
print("block onsets (s):", sb.block_starts())
UV = {}
TR = {}
TEMPL = {}

def block_templates(e, pre, fs, el_cortical, sorter="KS3", half_ms=6.0):
    """Mean waveform (+-6 ms, all cortical sites) of each cortical single unit, from its spikes
    inside the six blocks (for fig3fg-wavemap-six-profiles)."""
    df, tr = SORT[(e, sorter)]
    keep = (df.single & df.layer.isin(sb.LAYERS)).values
    blen = int(sb.BLOCK_S * fs)
    h = int(round(half_ms * fs / 1000))
    out = {}
    for uid, t, k in zip(df.index, tr, keep):
        if not k:
            continue
        idx = []
        for b, s0 in enumerate(sb.block_starts()):
            m = (t >= s0) & (t < s0 + sb.BLOCK_S)
            i = b * blen + np.round((t[m] - s0) * fs).astype(int)
            idx.append(i[(i - h >= b * blen) & (i + h < (b + 1) * blen)])
        idx = np.concatenate(idx)
        if len(idx) == 0:
            continue
        w = pre[idx[:, None] + np.arange(-h, h)[None, :]].mean(axis=0)  # samples x sites
        out[uid] = (w, len(idx))
    pos = el_cortical[[c for c in ["rel_x", "rel_y", "rel_z"] if c in el_cortical]].to_numpy(float)  # 3-D for the models
    return {"templates": out, "pos": pos, "fs": fs}
for e in ["NS", "E", "E40", "M", "DS1", "DS2", "DS3", "H1", "H2", "H3"]:
    el = ELEC[e]
    ch = np.where(el.layer.isin(sb.LAYERS).values)[0]
    pre, rmax, rpsd, f_raw = sb.trace_blocks(e, ch, UV.get(e, 1.0))
    fs = sb.SAMPLING_HZ[e]
    f, psd = sb.site_psd(pre, fs)
    mad_min = sb.site_noise_mean_ad(pre, fs)
    medad = sb.site_noise_median_ad(pre)
    anr = pre / medad[None, :]
    hist_lo, hist_hi = np.percentile(anr, [0, 100])
    TR[e] = {"layer": el.layer.values[ch], "peak": np.abs(pre).max(axis=0), "raw_peak": rmax,
             "noise": mad_min, "noise_med": medad, "f": f, "psd": psd, "f_raw": f_raw, "raw_psd": rpsd,
             "anr_min": anr.min(axis=0), "anr_max": anr.max(axis=0),
             "anr_le_m10": (anr <= -10).any(axis=0), "anr_ge_10": (anr >= 10).any(axis=0),
             "anr_hist": np.histogram(anr, bins=np.linspace(-60, 60, 101))[0]}
    if e in ("NS", "E", "M"):
        TEMPL[e] = block_templates(e, pre, fs, el.iloc[ch])
    del pre, anr
    print(f"{e}: {len(ch)} cortical sites, {fs} Hz; peak |trace| {TR[e]['peak'].max():.1f} uV, "
          f"median noise {np.median(mad_min):.2f} uV")

def pooled(group):
    parts = {"DS": ["DS1", "DS2", "DS3"], "H": ["H1", "H2", "H3"]}.get(group, [group])
    out = {}
    for k in ["layer", "peak", "raw_peak", "noise", "noise_med", "anr_min", "anr_max", "anr_le_m10", "anr_ge_10"]:
        out[k] = np.concatenate([TR[p][k] for p in parts])
    out["psd"] = np.concatenate([TR[p]["psd"] for p in parts], axis=1)
    out["raw_psd"] = np.concatenate([TR[p]["raw_psd"] for p in parts], axis=1)
    out["f"], out["f_raw"] = TR[parts[0]]["f"], TR[parts[0]]["f_raw"]
    return out

P = {g: pooled(g) for g in ["NS", "E", "E40", "M", "DS", "H"]}
P["H195"] = dict(P["H"])
for k in ["peak", "raw_peak", "noise", "noise_med"]:
    P["H195"][k] = P["H"][k] * 0.195
for k in ["psd", "raw_psd"]:
    P["H195"][k] = P["H"][k] * 0.195 ** 2
for g in ["DS", "H", "H195"]:
    print(f"{g}: median noise (mean AD) over sites {np.median(P[g]['noise']):.2f}, median peak {np.median(P[g]['peak']):.1f}")
for g, d in P.items():
    print(g, "sites by layer", pd.Series(d["layer"]).value_counts().reindex(sb.LAYERS).to_dict())
```

### Ten-minute extremes

The maxima and the ANR extremes depend on duration, so the first 600 s of each recording are also

only each site's raw maximum |trace| and the minimum and maximum of the preprocessed trace. The ANR
extremes over 600 s divide these by each site's median absolute deviation from the 60 s blocks (a
median, which 60 s estimates well). The cumulative maxima after 60, 300 and 600 s show how much the
extremes still grow. A check of the scipy filter against SpikeInterface's on one block comes first.


```python
# scipy high-pass and common median reference against SpikeInterface's, on 2 s of NS
from spikeinterface.core import NumpyRecording
import spikeinterface.preprocessing as spre
from scipy import signal as ssig
with sb.open_nwb(sb.RECORDINGS["NS"]) as f:
    x = f["acquisition/ElectricalSeriesRaw/data"][int(100 * 40000):int(102 * 40000), :].astype(np.float32)
sos = ssig.butter(5, sb.FREQ_MIN, btype="highpass", fs=40000, output="sos")
y = ssig.sosfiltfilt(sos, x, axis=0)
y = y - np.median(y, axis=1, keepdims=True)
rec = NumpyRecording([x], sampling_frequency=40000)
z = spre.common_reference(spre.highpass_filter(rec, freq_min=sb.FREQ_MIN), reference="global", operator="median").get_traces()
inner = slice(4000, -4000)  # away from the edges of the 2 s excerpt
d = np.abs(y[inner] - z[inner])
print(f"scipy against SpikeInterface on NS, 100 to 102 s (inner 1.8 s): largest difference {d.max():.2g} uV, "
      f"median {np.median(d):.2g} uV; trace sd {y[inner].std():.2f} uV")
del x, y, z

EXT = {}
for e in ["NS", "E", "E40", "M", "DS1", "DS2", "DS3", "H1", "H2", "H3"]:
    ch = np.where(ELEC[e].layer.isin(sb.LAYERS).values)[0]
    r = sb.trace_extremes(e)
    EXT[e] = {k: v[:, ch] for k, v in r.items() if k != "starts"}
    EXT[e]["starts"] = r["starts"]
    row = []
    for t in [60, 300, 600]:
        b = r["starts"] < t
        row.append(f"{t} s: raw {EXT[e]['raw_absmax'][b].max():.1f}, preprocessed {np.abs(np.r_[EXT[e]['pre_min'][b].min(), EXT[e]['pre_max'][b].max()]).max():.1f}")
    print(f"{e}: largest |trace| over cortical sites after " + "; ".join(row))
for g, parts in {"NS": ["NS"], "E": ["E"], "E40": ["E40"], "M": ["M"], "DS": ["DS1", "DS2", "DS3"], "H": ["H1", "H2", "H3"]}.items():
    P[g]["peak10"] = np.concatenate([np.maximum(-EXT[p]["pre_min"].min(axis=0), EXT[p]["pre_max"].max(axis=0)) for p in parts])
    P[g]["raw_peak10"] = np.concatenate([EXT[p]["raw_absmax"].max(axis=0) for p in parts])
    P[g]["anr_min10"] = np.concatenate([EXT[p]["pre_min"].min(axis=0) / TR[p]["noise_med"] for p in parts])
    P[g]["anr_max10"] = np.concatenate([EXT[p]["pre_max"].max(axis=0) / TR[p]["noise_med"] for p in parts])
```

## fig2m-peak-amplitude-increases-with-depth

> In both in vivo datasets the median peak amplitude of the traces increased with cortical depth,
> and the biophysical models reproduced this trend, although their medians differed from in vivo.

```python
def by_layer_test(key, groups, label):
    rows = []
    for g in groups:
        d = P[g]
        vals = {l: d[key][d["layer"] == l] for l in sb.LAYERS if (d["layer"] == l).sum() > 0}
        h, p = stats.kruskal(*vals.values())
        med = {l: float(np.median(v)) for l, v in vals.items()}
        ok, inv = sb.layer_order_ok(med)
        rows.append({"exp": g, "layers": len(vals), "H": round(h, 1), "p": p, "inversions": inv,
                     "rising (<= 1 inversion)": ok, **{l: round(m, 3) for l, m in med.items()}})
    t = pd.DataFrame(rows)
    print(label)
    print(t.to_string(index=False))
    return t

def vs_vivo(key, pairs):
    rows = []
    for g, ref in pairs:
        for l in sb.LAYERS:
            x, y = P[g][key][P[g]["layer"] == l], P[ref][key][P[ref]["layer"] == l]
            if len(x) and len(y):
                rows.append({"model": g, "vs": ref, "layer": l, "median model": round(np.median(x), 2),
                             "median vivo": round(np.median(y), 2), "p": stats.mannwhitneyu(x, y).pvalue})
    t = pd.DataFrame(rows)
    print(t.to_string(index=False))
    print("share of layer comparisons with p < 0.01:", round((t.p < 0.01).mean(), 2), f"({int((t.p < 0.01).sum())} of {len(t)})")
    return t

AMP = by_layer_test("peak", ["M", "H", "NS", "E", "E40", "DS"], "peak amplitude (uV) per site, by layer")
model_ok = AMP[AMP.exp.isin(["NS", "E", "DS"])].apply(lambda r: r.p < 0.001 and r["rising (<= 1 inversion)"], axis=1)
print("models with p < 0.001 and rising medians:", dict(zip(AMP.exp[AMP.exp.isin(["NS", "E", "DS"])], model_ok)))
_ = vs_vivo("peak", [("NS", "M"), ("E", "M"), ("DS", "H")])
print("variant, H at 0.195 uV per count:")
_ = vs_vivo("peak", [("DS", "H195")])
```

**Result: reproduced** (60 s of traces). Peak amplitude differs across layers in every model
(Kruskal-Wallis p = 1.2e-25 for NS, 1.2e-23 for E, 2.2e-07 for DS) and in vivo (M p = 2.0e-32, H
p = 3.6e-25), and the layer medians rise from layer 1 to layer 6 with one inversion in each (layer 6
below layer 5, or for DS layer 4 below layer 2/3). The model medians differ from in vivo at p < 0.01 in 12 of 15 layer comparisons
with H in counts (1 of 5 for DS against H at 0.195 uV per count).

## fig3fg-wavemap-six-profiles

> WaveMap clustering of multi-electrode spike waveforms identifies six distinct profiles in the
> simulation, as in the Marques-Smith data.

Kilosort3 single units (cortical) of NS and M; templates are the mean waveforms of each unit's
spikes inside the six 10 s blocks (data check), not of all its spikes over 10 minutes (a deviation).
Per unit: the site with the largest template amplitude (largest of first peak, trough and second
peak) and the sites nearest to it, six in all ("the six nearest sites", Methods), ordered by amplitude; each site's waveform aligned at the trough of the first site and cut to +-3 ms
(6 ms of the 12 ms extracted), resampled to 50 points per site (the WaveMap paper's 50 points, a
standard choice so that 30 and 40 kHz recordings give vectors of one length); the concatenated vector
scaled to 0 to 1. UMAP (n_neighbors 15, two dimensions), then Louvain on UMAP's graph, at the
resolution in 0.1 to 6.9 that maximises modularity (claims file criterion). Variants: three sites,
the six largest sites anywhere on the probe (the previous version's choice), units with 20 spikes or
more in their template, resolution 1.5, E and NS with E pooled, and two other UMAP seeds.

```python
import networkx as nx
warnings.filterwarnings("ignore", category=UserWarning)
import community as community_louvain
import umap
from scipy.signal import resample

def wavemap_vectors(tem, n_sites=6, half_ms=3.0, n_points=50, nearest=True, min_spikes=1):
    """nearest: the largest-amplitude site and the sites nearest to it (n_sites in all), ordered by
    amplitude (Methods, "the six nearest sites"); otherwise the n_sites largest anywhere (variant)."""
    fs = tem["fs"]
    pos = tem["pos"]
    vecs, ids = [], []
    for uid, (w, n) in tem["templates"].items():
        if n < min_spikes:
            continue
        amp = np.max(np.abs(w), axis=0)
        if nearest:
            dist = np.sqrt(((pos - pos[np.argmax(amp)]) ** 2).sum(axis=1))
            near = np.argsort(dist, kind="stable")[:n_sites]
            sites = near[np.argsort(-amp[near])]
        else:
            sites = np.argsort(-amp)[:n_sites]
        tr = np.argmin(w[:, sites[0]])
        h = int(round(half_ms * fs / 1000))
        if tr - h < 0 or tr + h > w.shape[0]:
            continue
        seg = w[tr - h:tr + h, sites]
        seg = resample(seg, n_points, axis=0).T.ravel()
        seg = (seg - seg.min()) / (seg.max() - seg.min())
        vecs.append(seg); ids.append(uid)
    return np.array(vecs), ids

def wavemap(X, seed=42, resolution=None, n_neighbors=15):
    red = umap.UMAP(n_neighbors=n_neighbors, n_components=2, random_state=seed)
    red.fit(X)
    G = nx.from_scipy_sparse_array(red.graph_)
    res_grid = np.round(np.arange(0.1, 7.0, 0.1), 1)
    if resolution is None:
        mods = []
        for r in res_grid:
            part = community_louvain.best_partition(G, resolution=r, random_state=seed)
            mods.append(community_louvain.modularity(part, G))
        resolution = res_grid[int(np.argmax(mods))]
    part = community_louvain.best_partition(G, resolution=resolution, random_state=seed)
    return len(set(part.values())), resolution, collections.Counter(part.values())

rows = []
VEC = {}
for e in ["NS", "M", "E"]:
    for ns in [6, 3]:
        VEC[(e, ns)] = wavemap_vectors(TEMPL[e], n_sites=ns)
    VEC[(e, "6 largest anywhere")] = wavemap_vectors(TEMPL[e], n_sites=6, nearest=False)
    VEC[(e, "6, >= 20 spikes")] = wavemap_vectors(TEMPL[e], n_sites=6, min_spikes=20)
VEC[("NS+E", 6)] = (np.vstack([VEC[("NS", 6)][0], VEC[("E", 6)][0]]), VEC[("NS", 6)][1] + VEC[("E", 6)][1])
for key in [("NS", 6), ("M", 6), ("NS", 3), ("M", 3), ("NS", "6 largest anywhere"), ("M", "6 largest anywhere"),
            ("NS", "6, >= 20 spikes"), ("M", "6, >= 20 spikes"), ("E", 6), ("NS+E", 6)]:
    X, ids = VEC[key]
    for seed in [42, 0, 1]:
        if key[1] != 6 or key[0] in ("E", "NS+E"):
            if seed != 42:
                continue
        n, r, sizes = wavemap(X, seed=seed)
        n15, _, _ = wavemap(X, seed=seed, resolution=1.5)
        rows.append({"exp": key[0], "sites": key[1], "units": len(ids), "seed": seed,
                     "best resolution": r, "clusters": n, "clusters at 1.5": n15,
                     "cluster sizes": sorted(sizes.values(), reverse=True)})
WM = pd.DataFrame(rows)
print(WM.to_string(index=False))
print("spikes per template (median, min):", {e: (int(np.median([v[1] for v in TEMPL[e]['templates'].values()])),
                                                 int(min(v[1] for v in TEMPL[e]['templates'].values()))) for e in ["NS", "M", "E"]})
```

**Result: not reproduced.** With the six nearest sites and the resolution that maximises modularity,
WaveMap gives 6 clusters for M (107 units, with all three UMAP seeds) but 8 for NS (230 units; 9 and
7 with two other seeds), so the simulation does not give six. Of the variants, three sites give 11
(NS) and 6 (M), the six largest sites anywhere give 8 and 5, units with 20 spikes or more give 7 and
5, resolution 1.5 gives 12 and 8, E gives 8 and NS with E pooled 12. The templates rest on a median
of 43 (NS) and 54 (M) spikes each, as few as 8 and 6.


## Sorted units: shared tables

Cortical units only (layer L1 to L6; units on sites outside the cortex dropped, claims file opening
comment). Single units are KSLabel `good` for Kilosort 2 to 4 and every unit for Kilosort and
herdingspikes. Firing rate is the spike count over the first 600 s. The dense probe depths are
summed for counts. Because the deposited DS1 sortings are the DS2 sortings (data check), DS is
given two ways: as deposited (DS1 + DS2 + DS3, DS2 counted twice), and without the DS1 folder
(DS2 + DS3), which is the one used where the claims sum the depths.

```python
def cortical(exp, sorter):
    df, tr = SORT[(exp, sorter)]
    keep = df.layer.isin(sb.LAYERS).values
    return df[keep], [t for t, k in zip(tr, keep) if k]

def counts(exp, sorter="KS4"):
    parts = {"DS": ["DS2", "DS3"], "DS_dep": ["DS1", "DS2", "DS3"], "H": ["H1", "H2", "H3"]}.get(exp, [exp])
    n = s = 0
    for p in parts:
        df, _ = cortical(p, sorter)
        n += len(df); s += int(df.single.sum())
    return {"units": n, "single": s, "multi": n - s, "single fraction": s / n}

COUNTS = pd.DataFrame({e: counts(e) for e in ["M", "NS", "E", "H", "DS", "DS_dep", "H1", "H2", "H3", "DS1", "DS2", "DS3"]}).T
print("Kilosort4, cortical units, first 10 minutes")
print(COUNTS)

def rates(exp, sorter="KS4", singles=True):
    df, _ = cortical(exp, sorter)
    if singles:
        df = df[df.single]
    return df
```

## fig4bc-lower-median-narrower-distribution

> Compared with in vivo, the biophysical simulations have lower median firing rates and narrower
> firing rate distributions across layers, while the synthetic model has a lower variance in layer 5
> but a higher median.

Per layer with units in M: two-sided Mann-Whitney U test of log10 rates against M; 100 bootstrap
variances per group compared with a Mann-Whitney U test. The synthetic model has no sorting.

```python
def boot_var(x, rng, n=100):
    return np.array([np.var(rng.choice(x, size=len(x), replace=True)) for _ in range(n)])

def compare_to_m(sorter, exps, rng, singles=True):
    rows = []
    m = rates("M", sorter, singles)
    for l in sb.LAYERS:
        xm = np.log10(m.rate[m.layer == l].values)
        if len(xm) < 3:
            continue
        for e in exps:
            d = rates(e, sorter, singles)
            x = np.log10(d.rate[d.layer == l].values)
            if len(x) < 3:
                rows.append({"sorter": sorter, "exp": e, "layer": l, "n": len(x), "n M": len(xm)})
                continue
            pm = stats.mannwhitneyu(x, xm, alternative="two-sided").pvalue
            vx, vm = boot_var(x, rng), boot_var(xm, rng)
            pv = stats.mannwhitneyu(vx, vm, alternative="two-sided").pvalue
            rows.append({"sorter": sorter, "exp": e, "layer": l, "n": len(x), "n M": len(xm),
                         "median": round(10 ** np.median(x), 3), "median M": round(10 ** np.median(xm), 3),
                         "p median": pm, "var log10": round(np.var(x), 3), "var log10 M": round(np.var(xm), 3),
                         "p var": pv})
    return pd.DataFrame(rows)

rng = np.random.default_rng(4)
FIG4BC = compare_to_m("KS4", ["NS", "E"], rng)
print(FIG4BC.to_string(index=False))
t = FIG4BC.dropna(subset=["p median"])
med_ok = bool(((t["median"] < t["median M"]) & (t["p median"] < 0.001)).all())
var_ok = bool(((t["var log10"] < t["var log10 M"]) & (t["p var"] < 0.001)).all())
print("medians lower at p < 0.001 in every tested layer and simulation:", med_ok,
      f"({int(((t['median'] < t['median M']) & (t['p median'] < 0.001)).sum())} of {len(t)})")
print("variances lower at p < 0.001 in every tested layer and simulation:", var_ok,
      f"({int(((t['var log10'] < t['var log10 M']) & (t['p var'] < 0.001)).sum())} of {len(t)})")
```

**Result: partially reproduced** (assumed). In layers 5 and 6, the only layers with more than one M unit, NS and
E have lower medians (p from 4.3e-08 to 2.2e-04) and lower bootstrapped variances (p below 1e-20)
than M, 4 of 4 comparisons each. The synthetic model's part cannot be tested (no sorting of S).
The criterion defines partial as the medians or the variances part holding alone; here both hold for
NS and E and the S part cannot be tested. Partially reproduced is used, as an assumption, for "the
tested parts hold and the rest cannot be tested"; without it the claim would be blocked for its S part.
E is the first 10 minutes of the 1-hour recording (assumed, as in fig4a-lognormal-firing-rates).

```python
def gof_pct(exps):
    pct = [100 * COUNTS.loc[e, "single fraction"] for e in exps]
    return stats.chisquare(pct)

def contingency(exps):
    tab = np.array([[COUNTS.loc[e, "single"], COUNTS.loc[e, "multi"]] for e in exps], float)
    chi2, p, dof, _ = stats.chi2_contingency(tab)
    return chi2, p, dof, int(tab.sum())

print("(i) NS single-unit fraction:", round(COUNTS.loc["NS", "single fraction"], 4), "(expected 0.41 +- 0.041)")
print("M fraction", round(COUNTS.loc["M", "single fraction"], 4))
r = gof_pct(["M", "NS", "E"]); print(f"(a) on M, NS, E: chi2({2}) = {r.statistic:.2f}, p = {r.pvalue:.3g}")
c = contingency(["M", "NS", "E"]); print(f"(b) on M, NS, E: chi2({c[2]}, N = {c[3]}) = {c[0]:.1f}, p = {c[1]:.3g}")
for ds in ["DS", "DS_dep"]:
    c = contingency(["M", "H", "NS", "E", ds])
    print(f"(c) on M, H, NS, E, {ds}: chi2({c[2]}, N = {c[3]}) = {c[0]:.1f}, p = {c[1]:.3g}")# verifier round 2: readings (ii) and (iii) with NS-2X taken from the paper's own counts (193 of 508);
# shown for reference only, as NS-2X is not in the dandiset
f2x = 193 / 508
print("(ii) mean of NS and NS-2X (paper's NS-2X):", round((COUNTS.loc["NS", "single fraction"] + f2x) / 2, 4))
print("(iii) mean of M, NS and NS-2X (paper's NS-2X):",
      round((COUNTS.loc["M", "single fraction"] + COUNTS.loc["NS", "single fraction"] + f2x) / 3, 4))
```

**Result: partially reproduced** (assumed). The proportion of single units differs between experiments: the
contingency test on M, H, NS, E and DS gives chi2(4, N = 5,079) = 488.6, p = 1.9e-104 (reading c).
The 41% is not found under the only reading that can be computed: NS's fraction is 0.354 (184 of 520),
outside 0.369 to 0.451. Readings (ii) and (iii) need NS-2X, which is not in the dandiset.
The criterion names no partial outcome. The test part holds; the fraction fails under reading (i), the
only one the dandiset allows, and readings (ii) and (iii), which the paper's own counts favour, need
NS-2X. Partially reproduced is used, as an assumption, for "one part holds and the other cannot be
fully tested"; without it the claim would be blocked for the fraction. For reference only, with NS-2X
taken from the paper's counts (193 of 508), reading (ii) gives 0.367 (outside) and reading (iii) 0.405
(within); these use the paper's numbers, not the dandiset, and do not decide the outcome.

## fig5b-yield-far-below-theory

> Raw single-unit yields per site are an order of magnitude lower than the theoretical yields
> (active neurons within 50 um per site, NS 6.9, NE 8.4, S 0.65, DS 5.3) and do not differ
> significantly between probes or sorters.

Theoretical yield: ground truth neurons with at least one spike in the first 600 s, by soma layer,
divided by the sites of that layer, median over layers (code get_gt_yield_per_site). The dense probe
ground truth tables hold no soma layer or position (data check), so DS's theoretical yield cannot be
computed. S: its ground truth units over 384 sites. Sorted yield: single units per layer over the
sites of that layer, averaged over layers.

```python
def site_counts(rec):
    return ELEC[rec].layer.value_counts()

def theory_yield(e, rec=None):
    gdf, _ = GT[e]
    n = gdf[gdf.n_spikes > 0].layer_std.value_counts()
    sites = site_counts(rec or e)
    per = (n / sites).dropna()
    return per, per.median()

TY = {}
for e in ["NS", "E", "E40"]:
    per, med = theory_yield(e)
    TY[e] = med
    print(f"{e}: theoretical yield per site by layer {per.round(2).to_dict()} -> median {med:.2f}")
TY["S"] = len(GT["S"][0]) / len(ELEC["S"])
print(f"S: {len(GT['S'][0])} units / {len(ELEC['S'])} sites = {TY['S']:.3f}")
print("DS: ground truth tables of sub-003 to sub-005 have columns", list(GT["DS1"][0].columns), "- no layer")
for e, v in [("NS", 6.9), ("E", 8.4), ("E40", 8.4), ("S", 0.65)]:
    print(f"  {e}: {TY[e]:.3f} expected {v} +- {0.1 * v:.3f} within: {abs(TY[e] - v) <= 0.1 * v}")

def sorted_yield(e, sorter, rec=None):
    df = rates(e, sorter, singles=(sorter not in ("KS", "HS")))
    sites = site_counts(rec or e)
    sites = sites[sites.index.isin(sb.LAYERS)]
    per = df.layer.value_counts().reindex(sites.index, fill_value=0) / sites
    return per.mean()

rows = []
for s in SORTERS:
    row = {"sorter": s}
    for e in ["NS", "E", "M", "DS2", "DS3", "H1", "H2", "H3"]:
        row[e] = round(sorted_yield(e, s), 4)
    rows.append(row)
SY = pd.DataFrame(rows).set_index("sorter")
print("sorted single-unit yield per site, mean over layers")
print(SY)
for e in ["NS", "E"]:
    r = TY[e] / SY.loc["KS4", e]
    print(f"{e}: theoretical / Kilosort4 yield = {r:.1f}; (a) >= 10: {r >= 10}; (b) >= 3.16: {r >= 10 ** 0.5}")
print("ratio theoretical / sorted, every sorter:")
print((pd.Series({e: TY[e] for e in ["NS", "E"]}) / SY[["NS", "E"]]).round(1))
```

**Result: partially reproduced** (assumed). The theoretical yields are 6.88 (NS), 8.44 (E; 8.27 for the 10-minute
recording) and 0.651 (S), within tolerance; DS's cannot be computed, as the dense probe ground truth
has no layers. Kilosort4's yield is 8.0 times below the theoretical one for NS and 4.8 times for E, an
order of magnitude under reading (b) (at least 3.16) and not under reading (a) (at least 10).
Both parts hold for what can be computed (the theoretical yields of NS, E and S, and the sorted
yields under reading b); only DS's theoretical yield stops a full reproduction. Partially reproduced
is used, as an assumption, for "the tested parts hold and the rest cannot be tested"; without it the
claim would be blocked for DS.


## fig5cd-most-units-degraded

> Most single units isolated by Kilosort4 are less than 80% accurate, contain spikes of several true
> neurons (merged) and miss at least 20% of their matched neuron's spikes; Kilosort4 produces no
> false positives.

For each single unit its best agreement score, and the quality tags of the code (false positive,
good, overmerger, oversplitter) computed over every sorted unit of the sorting, then kept for the single units. E is the first 10 minutes of the
1-hour recording (the 10-minute recording has no sorting).

```python
def qualities(e, s):
    """Tags computed over every sorted unit of the sorting, then kept for the cortical single units."""
    gdf, _ = gt_for(e, s)
    sdf, _ = SORT[(e, s)]
    keep = (sdf.single & sdf.layer.isin(sb.LAYERS)).values
    sc = SCORES[(e, s)].loc[sdf.index, gdf.index].values
    ch = sb.chance_scores(gdf.n_spikes.values, sdf.n_spikes.values, sb.sort_duration(e, s))
    q = sb.qualify_units(sc, ch)
    n_above = (sc[keep] > ch[keep]).sum(axis=1)
    return sc[keep], [t for t, k in zip(q, keep) if k], n_above

rows = []
for e in ["NS", "E"]:
    for s in SORTERS:
        sc, q, n_above = qualities(e, s)
        best = sc.max(axis=1)
        rows.append({"exp": e, "sorter": s, "single units": len(q),
                     "best < 0.8": round((best < 0.8).mean(), 3),
                     "merged": round(np.mean(["overmerger" in t for t in q]), 3),
                     "partial": round(np.mean(["oversplitter" in t for t in q]), 3),
                     "good": round(np.mean(["good" in t for t in q]), 3),
                     "false positives": int(sum("false_positive" in t for t in q)),
                     "above-chance neurons per unit (median)": int(np.median(n_above)) if len(n_above) else 0})
FIG5C = pd.DataFrame(rows)
print(FIG5C.to_string(index=False))
for e in ["NS", "E"]:
    r = FIG5C[(FIG5C.exp == e) & (FIG5C.sorter == "KS4")].iloc[0]
    print(f"{e} Kilosort4: more than half below 0.8: {r['best < 0.8'] > 0.5}; more than half merged: {r['merged'] > 0.5}; "
          f"no false positive: {r['false positives'] == 0}")
```

**Result: reproduced** (assumed: E is the first 10 minutes of the 1-hour recording). Among Kilosort4 single units, 54.3% (NS) and 74.2% (E) have a best score below
0.8, all are tagged merged (1.000 in both), and none is a false positive. The merged tag is met
almost by chance coincidence: the median Kilosort4 single unit scores above chance with 308 (NS) and
597 (E) neurons, so this part says little. The other sorters give 57.8% to 96.4% below 0.8 and
96.6% to 100% merged, with up to 37 false positives (Kilosort on E). Tags are computed over every
sorted unit and only among above-chance matches (verifier round 1).

Accuracy of a ground truth neuron: its best score over the sorted units (single units for Kilosort

```python
gdf_full, gtr_full = GT_FULL_E
gdf_full["layer_std"] = sb.std_layer(gdf_full["layer"])
gdf_full["type"] = gdf_full["synapse_class"] + " " + gdf_full["layer_std"] + " " + gdf_full["etype"]
gdf_full["exc_layer"] = np.where(gdf_full.synapse_class == "EXC", "EXC " + gdf_full.layer_std, "INH")
print("excitatory share of the ground truth:", round((gdf_full.synapse_class == "EXC").mean(), 4), "(expected 0.91 +- 0.091)")

def matched_types(duration):
    out = {}
    for s in SORTERS:
        if duration == "10m":
            sdf, st = SORT[("E", s)]
            sc = SCORES[("E", s)]
        else:
            sdf, st = SORT_FULL_E[s]
            sc = sb.agreement_scores(gtr_full, gdf_full.index, st, sdf.index, sb.SAMPLING_HZ["E"])
        if s not in ("KS", "HS"):
            sc = sc.loc[sdf.index[sdf.single.values]]
        best = sc.values.argmax(axis=1)
        has = sc.values.max(axis=1) > 0
        out[s] = (gdf_full.iloc[best[has]], int((~has).sum()), len(has))
    return out

def bias_table(matched, key):
    gt_share = gdf_full[key].value_counts()
    rows = []
    for s, (m, dropped, total) in matched.items():
        ms = m[key].value_counts()
        for t in gt_share.index:
            a, b = int(gt_share[t]), int(ms.get(t, 0))
            A, B = len(gdf_full) - a, len(m) - b
            _, p = stats.fisher_exact([[a, b], [A, B]])
            rows.append({"sorter": s, "type": t, "gt": a, "sorted": b,
                         "ratio": (b / len(m)) / (a / len(gdf_full)) if len(m) else np.nan, "p": p})
    return pd.DataFrame(rows)

BIAS = {}
for dur in ["10m", "1h"]:
    matched = matched_types(dur)
    print(f"\n--- sortings of {dur}: units used and dropped (no match) per sorter:",
          {s: (len(v[0]), v[1]) for s, v in matched.items()})
    exc = bias_table(matched, "exc_layer")
    inh = bias_table(matched, "type")
    inh = inh[inh.type.str.startswith("INH")]
    BIAS[dur] = (exc, inh)
    piv = exc.pivot(index="type", columns="sorter", values="ratio").round(2)
    print("ratio sorted share / ground truth share, by synapse class and layer")
    print(piv)
    print("p (Fisher exact, two-sided)")
    print(exc.pivot(index="type", columns="sorter", values="p").map(lambda v: f"{v:.2g}"))
    ok, ok_strict = {}, {}
    for s in SORTERS:
        x = exc[exc.sorter == s].set_index("type")
        l5 = x.loc["EXC L5"]
        others = x.loc[[t for t in x.index if t.startswith("EXC") and t != "EXC L5"]]
        ratios_ok = bool(l5.ratio >= 1.5 and (others.ratio <= 0.67).all())
        # "with p < 0.05 for those that differ": the layers whose ratio is 1.5 or more, or 0.67 or less
        differ = [t for t in x.index if t.startswith("EXC") and (x.loc[t, "ratio"] >= 1.5 or x.loc[t, "ratio"] <= 0.67)]
        ok[s] = ratios_ok and bool((x.loc[differ, "p"] < 0.05).all())
        ok_strict[s] = ratios_ok and bool(l5.p < 0.05 and (others.p < 0.05).all())
    print("layer 5 excitatory >= 1.5 and other excitatory layers <= 0.67, p < 0.05 for those that differ:", ok,
          "-> majority:", sum(ok.values()) > len(ok) / 2)
    print("same, p < 0.05 required in every layer (previous rule):", ok_strict)
    big = inh.sort_values("ratio", ascending=False).groupby("sorter").head(1).set_index("sorter")
    print("largest interneuron-type ratio per sorter")
    print(big[["type", "gt", "sorted", "ratio", "p"]].round(3))
    print("largest ratio among KS, KS2, KS2.5 within 2.6 to 8.9:",
          {s: bool(2.6 <= big.loc[s, "ratio"] <= 8.9) for s in ["KS", "KS2", "KS2.5"]})
```

## fig6c-f-accuracy-predictors (part only)

> A fractional regression of Kilosort4 sorting accuracy on 21 neuron features fits the data (R2 =
> 0.42 spontaneous, 0.34 evoked) and finds seven significant features in both regimes; accuracy rises
> with firing rate and spike spatial extent, is higher for pyramidal cells and in layer 5, and falls
> with distance to the nearest electrode.

The claim stays blocked: two of the 21 features, snr and spatial extent, need each neuron's
waveforms over the full traces (see "Claims not tested"), so neither the 21-feature R2 and count nor
the sign of spatial extent can be computed. The other 19 are in the ground truth tables or follow
from them (verifier round 1, problem 11): firing rate (spikes in the first 600 s / 600 s), synapse
class (EXC = 1), the nine electrical types and five layers of the code's list as indicators,
distance from the soma to the nearest site, exc_mini_frequency and dynamics_holding_current, all
z-scored. Accuracy is each neuron's best agreement with any Kilosort4 unit (first 10 minutes). The
model is the criterion's: Binomial GLM, logit link, `fit_regularized` elastic net with L1_wt 1, alpha
0, maxiter 100, cnvrg_tol 1e-10, with an intercept (a standard choice); McFadden's R2 on the test
sets of 100 random 75 / 25 splits, median. The indicator coding is rank-deficient (cADpyr equals
synapse class, and the layer and type indicators sum to constants), so the unpenalised Wald refit of
the code has no unique solution; the Wald p values are taken from an identified refit that drops
cADpyr, cACint (reference interneuron type) and every layer but layer 5 (layer 5 against the rest).
This tests four of the five signs of the claim, a 19-feature R2 and a 13-feature count, none of which
can decide the outcome.


```python
import statsmodels.api as sm
from statsmodels.tools.sm_exceptions import ConvergenceWarning

ETYPES_21 = ["bAC", "bNAC", "bSTUT", "cACint", "cADpyr", "cNAC", "cSTUT", "dNAC", "dSTUT"]

def predictor_table(e):
    gdf, _ = GT[e]
    X = pd.DataFrame(index=gdf.index)
    X["firing_rate"] = gdf.n_spikes.values / sb.DURATION_S
    X["synapse_class"] = (gdf.synapse_class == "EXC").astype(float).values
    for t in ETYPES_21:
        X[t] = (gdf.etype == t).astype(float).values
    for l in sb.LAYERS:
        X["layer " + l] = (gdf.layer_std == l).astype(float).values
    X["distance"] = sb.soma_distances(gdf, ELEC[e])
    X["exc_mini_frequency"] = gdf.exc_mini_frequency.values
    X["dynamics_holding_current"] = gdf.dynamics_holding_current.values
    y = SCORES[(e, "KS4")].max(axis=0).reindex(gdf.index).values
    Z = ((X - X.mean()) / X.std(ddof=0)).fillna(0.0)  # a constant column (no neuron of a type) becomes 0
    return Z, y

def loglik(y, p):
    p = np.clip(p, 1e-12, 1 - 1e-12)
    return np.sum(y * np.log(p) + (1 - y) * np.log(1 - p))

FRACREG = {}
for e, regime in [("NS", "spontaneous"), ("E", "evoked")]:
    Z, y = predictor_table(e)
    Xc = sm.add_constant(Z, has_constant="add")
    rng = np.random.default_rng(61)
    r2, nonconv = [], 0
    for _ in range(100):
        idx = rng.permutation(len(y)); ntr = int(0.75 * len(y)); tr, te = idx[:ntr], idx[ntr:]
        with warnings.catch_warnings(record=True) as wl:
            warnings.simplefilter("always", ConvergenceWarning)
            m = sm.GLM(y[tr], Xc.values[tr], family=sm.families.Binomial()).fit_regularized(
                method="elastic_net", alpha=0.0, L1_wt=1.0, maxiter=100, cnvrg_tol=1e-10)
        nonconv += any(issubclass(w.category, ConvergenceWarning) for w in wl)
        r2.append(1 - loglik(y[te], m.predict(Xc.values[te])) / loglik(y[te], np.full(len(te), y[tr].mean())))
    with warnings.catch_warnings(record=True) as wl:
        warnings.simplefilter("always", ConvergenceWarning)
        reg = sm.GLM(y, Xc.values, family=sm.families.Binomial()).fit_regularized(
            method="elastic_net", alpha=0.0, L1_wt=1.0, maxiter=100, cnvrg_tol=1e-10)
    reg_conv = not any(issubclass(w.category, ConvergenceWarning) for w in wl)
    W = sm.add_constant(Z.drop(columns=["cADpyr", "cACint", "layer L1", "layer L2/3", "layer L4", "layer L6"]), has_constant="add")
    ident = sm.GLM(y, W, family=sm.families.Binomial()).fit()
    tab = pd.DataFrame({"weight (regularised fit)": pd.Series(reg.params, index=Xc.columns),
                        "weight (identified refit)": ident.params, "Wald p": ident.pvalues}).drop("const")
    FRACREG[e] = tab
    print(f"--- {regime} ({e}): {len(y)} neurons, rank of the 19-feature design {np.linalg.matrix_rank(Xc.values)} of {Xc.shape[1]} columns")
    print(f"test McFadden R2, median of 100 splits {np.median(r2):.3f} (range {np.min(r2):.3f} to {np.max(r2):.3f}); "
          f"in-sample, identified refit {1 - ident.llf / ident.llnull:.3f} (paper, 21 features: {0.42 if e == 'NS' else 0.34})")
    print(f"split fits that reached maxiter 100 without converging: {nonconv} of 100; the fit on all neurons converged: {reg_conv}")
    print(tab.round(4).to_string())
    print("significant (p < 0.05) of the identified refit's 13:", int((ident.pvalues.drop("const") < 0.05).sum()))
    signs = {"firing_rate +": tab.loc["firing_rate"].iloc[:2].gt(0).all(),
             "synapse_class +": tab.loc["synapse_class"].iloc[:2].gt(0).all(),
             "layer L5 +": tab.loc["layer L5"].iloc[:2].gt(0).all(),
             "distance -": tab.loc["distance"].iloc[:2].lt(0).all()}
    print("signs of the claim, in both fits:", {k: bool(v) for k, v in signs.items()}, "; spatial extent: not computable")
```

**Result: blocked (too large), part run.** Without snr and spatial extent, the 19-feature model gives
the four other signs of the claim in both regimes and both fits: firing rate positive (identified
refit 1.09 spontaneous, 0.77 evoked, p < 1e-4), synapse class (excitatory) positive (0.22 and 0.23,
p = 0.083 and 0.056), layer 5 positive (0.57 and 0.45, p < 1e-4) and distance negative (-1.23 and
-1.32, p < 1e-4). The test R2 of this smaller model is 0.307 (spontaneous) and 0.279 (evoked), below
the paper's 21-feature 0.42 and 0.34, and the identified refit has 4 and 5 significant features of
13. In the spontaneous regime every fit converged; in the evoked regime none of the 100 split fits
nor the fit on all neurons converged within the paper's 100 iterations at tolerance 1e-10; its weights
keep the signs of the identified refit for the four features above, but differ for others (synapse
class 0.40 against 0.23, bAC +0.08 against -0.03). None of this decides the claim: the sign of spatial extent and the 21-feature R2 and count need
the waveforms.

240 s, as the Methods say, and is scored over 240 s; the dense probe is summed over depths 2
