## task-performance

> Mice performed the task at 78.4% correct on average, with responses before the go cue on
> fewer than 13% of trials (n = 42 mice).

Only the 19 electrophysiology mice are in the data set, against 42 in the paper. Per cent
correct is hits over hits and errors, lick early trials excluded (Methods); no-lick trials are
left out, as the data set's `Demo_get_performance.m` does. Lick early fraction is over all
performing trials of the same set. Readings from the criterion: mean over sessions (a), over
mice (b), pooled trials (c); control trials (type 0, no StimTrials flag) or all performing trials except the NaN-type test
trials, which the data description says are not to be analysed.

```python
rows = []
for trials_name, sel in (("control", (T.ptype == 0) & ~T.StimTrials), ("all but NaN-type", T.ptype.notna())):
    d = T[(T.good == 1) & sel & T.cue.notna()]
    d = d.assign(hit=d.HitR | d.HitL, resp=d.HitR | d.HitL | d.ErrR | d.ErrL)
    ne = d[~d.LickEarly & d.resp]
    ps = ne.groupby("session").hit.mean() * 100
    le_s = d.groupby("session").LickEarly.mean() * 100
    pm = ne.groupby("animal").hit.mean() * 100
    le_m = d.groupby("animal").LickEarly.mean() * 100
    for how, pc, le in (("a sessions", ps.mean(), le_s.mean()), ("b mice", ne.assign(s=ne.session).groupby("animal").apply(lambda x: x.groupby("session").hit.mean().mean()).mean() * 100, d.groupby("animal").apply(lambda x: x.groupby("session").LickEarly.mean().mean()).mean() * 100),
                        ("c pooled", ne.hit.mean() * 100, d.LickEarly.mean() * 100)):
        rows.append({"trials": trials_name, "mean": how, "per_cent_correct": pc, "lick_early_pct": le,
                     "within_tolerance": abs(pc - 78.4) <= 7.84, "lick_early_below_14.3": le < 14.3, "n_trials": len(d)})
perf = pd.DataFrame(rows)
print(perf.round(2).to_string(index=False))
print("\nPer mouse (control trials):")
d = T[(T.good == 1) & (T.ptype == 0) & ~T.StimTrials & T.cue.notna()]
print(pd.DataFrame({"pct_correct": d[~d.LickEarly & (d.HitR | d.HitL | d.ErrR | d.ErrL)].groupby("animal").apply(lambda x: (x.HitR | x.HitL).mean() * 100),
                    "lick_early_pct": d.groupby("animal").LickEarly.mean() * 100}).round(1).T.to_string())
```

## fig2c-selectivity-time-course

> Selectivity emerges in the sample epoch, increases through the delay epoch and is largest at
> the beginning of the response epoch.

Measure 1 (ED Fig. 1f): the number of pyramidal neurons significant (two-tailed t-test on spike
counts, P < 0.05) in 200 ms windows, stepped by 10 ms. Measure 2 (Fig. 2c): the population
selectivity of the selective neurons, preferred minus non-preferred rate, with the preference
set from 10 random trials of each type (by spike count from sample onset to the end of the
response epoch, as the data set's compile script does) and the rate difference taken on the
remaining trials. Neurons need at least 11 trials of each type for the split; the data set's
script requires more than 15, a variant. The epoch boundaries relative to the cue are the
medians over trials. Seeds are fixed.

```python
rng = np.random.default_rng(20150305)
t_sample = -float(np.median(g.cue - g.pole_in))
t_delay = -float(np.median(g.cue - g.pole_out))
print(f"Median sample onset {t_sample:.3f} s and delay onset {t_delay:.3f} s relative to the cue")
cnt_sig, cnt_sig_shuf, popsel, popsel15, n_pop = None, None, [], [], 0
sel_keys = set(map(tuple, pyr[pyr.selective][["session", "unit"]].values))
pyr_keys = set(map(tuple, pyr[["session", "unit"]].values))
for s, u in units:
    key = (s["session"], u["unit"])
    if key not in pyr_keys:
        continue
    tr = s["trials"]
    r, l = alm.trial_mask(tr, u)
    if r.sum() < 2 or l.sum() < 2:
        continue
    br, bl = alm.cue_binned(u, tr, r), alm.cue_binned(u, tr, l)
    wr, centers = alm.window_sums(br)
    wl, _ = alm.window_sums(bl)
    p = stats.ttest_ind(wr, wl, axis=0).pvalue
    p = np.where(np.isnan(p), 1, p)
    allw = np.vstack([wr, wl])
    perm = rng.permutation(len(allw))
    ps = stats.ttest_ind(allw[perm[:len(wr)]], allw[perm[len(wr):]], axis=0).pvalue
    ps = np.where(np.isnan(ps), 1, ps)
    cnt_sig = (p < 0.05).astype(int) if cnt_sig is None else cnt_sig + (p < 0.05)
    cnt_sig_shuf = (ps < 0.05).astype(int) if cnt_sig_shuf is None else cnt_sig_shuf + (ps < 0.05)
    if key in sel_keys and r.sum() >= 11 and l.sum() >= 11:
        ir, il = rng.permutation(r.sum()), rng.permutation(l.sum())
        whole = lambda b: b[:, (alm.CUE_EDGES[:-1] >= t_sample) & (alm.CUE_EDGES[:-1] < alm.RESPONSE_S)].sum(axis=1)
        pref_r = whole(br)[ir[:10]].mean() >= whole(bl)[il[:10]].mean()
        diff = (wr[ir[10:]].mean(axis=0) - wl[il[10:]].mean(axis=0)) / alm.WIN_S
        popsel.append(diff if pref_r else -diff)
        if r.sum() > 15 and l.sum() > 15:
            popsel15.append(popsel[-1])
popsel, popsel15 = np.array(popsel), np.array(popsel15)
print("Neurons in measure 1:", None if cnt_sig is None else "pyramidal with >= 2 trials of each type",
      "; in measure 2:", len(popsel), "(>15 trials each:", len(popsel15), ")")

def at(x, t):
    return float(x[np.argmin(np.abs(centers - t))])

def check(x, name):
    rng_t = (centers >= -3.4) & (centers <= 1.9)
    tmax = float(centers[rng_t][np.argmax(x[rng_t])])
    v = {"measure": name, "pre_sample": at(x, t_sample - 0.1), "sample_end": at(x, t_delay - 0.1),
         "delay_start": at(x, t_delay + 0.1), "delay_end": at(x, -0.1), "max_value": float(x[rng_t].max()), "t_max_s": tmax}
    v["emerges"] = v["sample_end"] > v["pre_sample"]
    v["increases_in_delay"] = v["delay_end"] > v["delay_start"]
    v["max_within_0.5s"] = 0 < tmax <= 0.5
    v["max_within_0.2s"] = 0 < tmax <= 0.2
    return v

res = pd.DataFrame([check(cnt_sig.astype(float), "1 significant neurons"), check(popsel.mean(axis=0), "2 population selectivity"),
                    check(popsel15.mean(axis=0), "2 variant, >15 trials")])
print(res.round(3).to_string(index=False))
print("\nShuffled labels, significant neurons: median", int(np.median(cnt_sig_shuf)), "max", int(cnt_sig_shuf.max()),
      "; true labels before sample onset (window centred 0.1 s before):", int(at(cnt_sig, t_sample - 0.1)),
      "; 5% of the neurons tested:", round(0.05 * len(pyr)))
fig, ax = plt.subplots(1, 2, figsize=(11, 3.2))
ax[0].plot(centers, cnt_sig, "k"); ax[0].plot(centers, cnt_sig_shuf, color="0.6")
ax[1].plot(centers, popsel.mean(axis=0), "k")
for a in ax:
    for x in (t_sample, t_delay, 0):
        a.axvline(x, color="0.7", ls=":")
    a.set_xlabel("time from cue (s)")
ax[0].set_ylabel("significant neurons"); ax[1].set_ylabel("pref - nonpref (spikes/s)")
plt.tight_layout(); plt.show()
print("Plotted ranges: significant neurons", int(cnt_sig.min()), "to", int(cnt_sig.max()), "; population selectivity",
      round(float(popsel.mean(axis=0).min()), 2), "to", round(float(popsel.mean(axis=0).max()), 2), "spikes/s")
```

## fig2c-dynamic-single-neurons

> Individual neuron responses are dynamic while selectivity at the population level is stable.

Population response correlation (Methods): for the selective neurons, each neuron's
cue-aligned rate (200 ms window, 10 ms step, correct trials) is mean-subtracted and divided by
the variance of its response across the trial (as the Methods say; the data set's script does
the same; division by the s.d. is a variant). Pearson's r between the population vector at each
time and at the cue. Readings (a) the vector of selectivities (R minus L) and (b) the R and L
rates concatenated. The population selectivity is measure 2 of the previous claim.

```python
PR, PL = [], []
for s, u in units:
    if (s["session"], u["unit"]) not in sel_keys:
        continue
    tr = s["trials"]
    r, l = alm.trial_mask(tr, u)
    if r.sum() < 2 or l.sum() < 2:
        continue
    PR.append(alm.window_sums(alm.cue_binned(u, tr, r))[0].mean(axis=0) / alm.WIN_S)
    PL.append(alm.window_sums(alm.cue_binned(u, tr, l))[0].mean(axis=0) / alm.WIN_S)
PR, PL = np.array(PR), np.array(PL)
span = (centers >= -3.4) & (centers <= 1.9)
def norm(x, how):
    x = x[:, span]
    x = x - x.mean(axis=1, keepdims=True)
    d = x.var(axis=1, keepdims=True) if how == "var" else x.std(axis=1, keepdims=True)
    return x / np.where(d == 0, np.nan, d)
cs = centers[span]
i0 = np.argmin(np.abs(cs))
win_early = (cs >= t_sample) & (cs <= t_delay + 0.5)
rows, curves = [], {}
for how in ("var", "sd"):
    nr, nl = norm(PR, how), norm(PL, how)
    for reading, M in (("a selectivity", nr - nl), ("b concatenated", np.vstack([nr, nl]))):
        ok = ~np.isnan(M).any(axis=1)
        M = M[ok]
        rc = np.array([np.corrcoef(M[:, j], M[:, i0])[0, 1] for j in range(M.shape[1])])
        curves[(how, reading)] = rc
        rows.append({"normalised_by": how, "reading": reading, "neurons_rows": int(ok.sum()),
                     "mean_r_sample_to_delay+0.5": rc[win_early].mean(), "max_r_in_that_window": rc[win_early].max(),
                     "r_at_delay_end_-0.1s": rc[np.argmin(np.abs(cs + 0.1))]})
dyn = pd.DataFrame(rows)
dyn["below_0.5"] = dyn["mean_r_sample_to_delay+0.5"] < 0.5
print(dyn.round(3).to_string(index=False))
ps_curve = popsel.mean(axis=0)
stable = (centers >= t_delay - 0.1) & (centers <= -0.1)
print("\nPopulation selectivity from the end of the sample epoch through the delay: min", round(float(ps_curve[stable].min()), 3),
      "spikes/s; all above 0:", bool((ps_curve[stable] > 0).all()))
```

**Result: reproduced, resting on the claims file's 0.5 bound.** The mean population response correlation with the cue-time vector over the sample epoch and the first 0.5 s of the delay is -0.168 (reading a) and -0.159 (reading b), with maxima of -0.024 and -0.042 in that span; with division by the s.d. it is -0.041 and -0.084. All are below 0.5, while the population selectivity stays above 0 from the end of the sample epoch through the delay (minimum 1.19 spikes/s).

## ed2-error-trials-predict-choice

> On error trials, when mice lick opposite to the instructed direction, ALM activity predicts
> the licking direction; most selective neurons switch their trial type preference.

Error trials are labelled by the instructed side (ErrR: pole for lick right, mouse licked
left). Per selective pyramidal neuron and epoch, correct selectivity R minus L and error
selectivity ErrR minus ErrL. Readings: at least 1 or at least 5 error trials of each type; the
data set's script uses more than 10 correct and more than 5 error trials of each type (variant).

```python
rows = []
sp = pyr[pyr.selective]
for name, m in (("min 1 error each", (sp.n_errR >= 1) & (sp.n_errL >= 1)), ("min 5 error each", (sp.n_errR >= 5) & (sp.n_errL >= 5)),
                ("script: >10 correct, >5 error", (sp.n_R > 10) & (sp.n_L > 10) & (sp.n_errR > 5) & (sp.n_errL > 5))):
    d = sp[m]
    for e in ("sample", "delay", "response"):
        c = d[f"{e}_R"] - d[f"{e}_L"]
        er = d[f"{e}_errR"] - d[f"{e}_errL"]
        r, p = stats.pearsonr(c, er)
        nz = (c != 0) & (er != 0)
        rows.append({"reading": name, "epoch": e, "neurons": len(d), "r": r, "p": p,
                     "fraction_switched": float((np.sign(c[nz]) != np.sign(er[nz])).mean()), "n_nonzero": int(nz.sum())})
err = pd.DataFrame(rows)
err["holds"] = (err.r < 0) & (err.p < 0.05) & (err.fraction_switched > 0.5)
print(err.round(4).to_string(index=False))
print("\nSelective pyramidal neurons without error trials of both types:", int(((sp.n_errR < 1) | (sp.n_errL < 1)).sum()), "of", len(sp))
print("\nUnder the four readings of the shared methods (selection 'min 1 error each'):")
rows = []
for (le, resp), d in UT.items():
    sp_ = d[(d.label == "pyramidal") & d.selective & (d.n_errR >= 1) & (d.n_errL >= 1)]
    for e in ("sample", "delay", "response"):
        c, er = sp_[f"{e}_R"] - sp_[f"{e}_L"], sp_[f"{e}_errR"] - sp_[f"{e}_errL"]
        nz = (c != 0) & (er != 0)
        r, p = stats.pearsonr(c, er)
        rows.append({"lick_early": le, "response_s": resp, "epoch": e, "neurons": len(sp_), "r": r, "p": p,
                     "fraction_switched": float((np.sign(c[nz]) != np.sign(er[nz])).mean())})
er4 = pd.DataFrame(rows)
er4["holds"] = (er4.r < 0) & (er4.p < 0.05) & (er4.fraction_switched > 0.5)
print(er4.round(4).to_string(index=False))
```

The four readings of the shared methods give the same (sample r = -0.082 to -0.097, P 0.004 to 0.018; switched fractions 0.502 to 0.627). ## fig2e-equal-contra-ipsi

> Among selective putative pyramidal neurons, contra-preferring and ipsi-preferring neurons are
> present in equal proportions (427 contra-preferring and 485 ipsi-preferring in Fig. 2c).

Preference from the spike rates from sample onset to the end of the response epoch, R against

## antidromic-latency-jitter

> PT neurons have shorter antidromic latencies than IT neurons (mean 2.7 against 4.6 ms) and both
> have small jitter (mean s.d. 0.1 and 0.2 ms) ...; the mean jitter of tagged neurons is 0.15 ms.

Pulse onsets are read from the AOM input trace (10 kHz, so 0.1 ms steps); spike times are at
the 19531.25 Hz sampling of the recording. For each PT neuron, the PT axonal stimulation trials
(type 1); for each IT neuron, the IT trials (type 2, and in a variant also the undocumented
types 3 to 5). Latency is the first spike within 10 ms of a pulse onset (before the second
pulse). The paper does not say how antidromic spikes are told from spontaneous ones; here they
are the spikes within ±1 ms of the neuron's modal latency (0.2 ms bins), with ±0.5 ms and a
robust s.d. (1.4826 times the median absolute deviation within ±1 ms) as variants (an
assumption). The second cell counts, per neuron, the spikes observed between 0.5 and 1 ms from the
mode against those expected from the spontaneous background, and draws the latency histograms
of the IT neurons with both windows marked. The third gives readings that need no window: a
per-neuron fit of a Gaussian peak plus a uniform background, the same with a broad Gaussian
added for broad stimulus-locked humps, and the ±1 ms jitter without the four IT neurons whose
histograms show such humps. Readings: (a) first pulse of each
pair, (b) both pulses pooled.

```python
def lat_table(it_types, hw):
    rows = []
    for s, u in units:
        tag = "PT" if "PT" in u["cell_types"] else "IT" if "IT" in u["cell_types"] else ""
        if not tag:
            continue
        types = {1.0} if tag == "PT" else it_types
        l0 = alm.antidromic_latencies(s, u, types, order=0)
        l1 = alm.antidromic_latencies(s, u, types, order=1)
        for reading, lat in (("a first pulse", l0), ("b both pulses", np.r_[l0, l1])):
            m, sd, n, nr = alm.latency_mode_stats(lat, hw)
            rows.append({"session": s["session"], "unit": u["unit"], "animal": s["animal"], "tag": tag, "reading": reading,
                         "latency_ms": m, "jitter_ms": sd, "n_resp": n, "n_pulses": len(lat), "p_resp": n / max(len(lat), 1)})
    return pd.DataFrame(rows)

LT = {}
summ = []
for it_types, it_name in (({2.0}, "IT type 2"), ({2.0, 3.0, 4.0, 5.0}, "IT types 2-5")):
    for hw in (1.0, 0.5):
        lt = lat_table(it_types, hw)
        LT[(it_name, hw)] = lt
        for reading, d in lt.groupby("reading"):
            ok = d[d.n_resp >= 3]
            pt, it = ok[ok.tag == "PT"], ok[ok.tag == "IT"]
            summ.append({"IT trials": it_name, "window_ms": hw, "reading": reading, "PT_n": len(pt), "IT_n": len(it),
                         "PT_latency": pt.latency_ms.mean(), "IT_latency": it.latency_ms.mean(),
                         "PT_jitter": pt.jitter_ms.mean(), "IT_jitter": it.jitter_ms.mean(), "tagged_jitter": ok.jitter_ms.mean(),
                         "PT_latency_median": pt.latency_ms.median(), "IT_latency_median": it.latency_ms.median()})
lat = pd.DataFrame(summ)
lat["PT_ok"] = (abs(lat.PT_latency - 2.7) <= 0.27) & (abs(lat.PT_jitter - 0.1) <= 0.05)
lat["IT_ok"] = (abs(lat.IT_latency - 4.6) <= 0.46) & (abs(lat.IT_jitter - 0.2) <= 0.05)
lat["tagged_jitter_ok"] = abs(lat.tagged_jitter - 0.15) <= 0.015
lat["PT_shorter"] = lat.PT_latency < lat.IT_latency
print(lat.round(3).to_string(index=False))
main_lt = LT[("IT type 2", 1.0)]
print("\nPer neuron (IT type 2, ±1 ms, first pulse):")
print(main_lt[main_lt.reading == "a first pulse"].round(3).to_string(index=False))
```

```python
# Which window fits: for each tagged neuron (first pulses; IT type 2 trials), the first-spike
# latencies far from the mode (more than 2 ms) give the spontaneous background, and so the
# number of spikes expected by chance within ±0.5 and ±1 ms of the mode, against those observed.
# The spikes between 0.5 and 1 ms from the mode are antidromic when they far exceed chance.
# A robust jitter (1.4826 x median absolute deviation, within ±1 ms) is a third reading.
rows = []
for s, u in units:
    tag = "PT" if "PT" in u["cell_types"] else "IT" if "IT" in u["cell_types"] else ""
    if not tag:
        continue
    lat = alm.antidromic_latencies(s, u, {1.0} if tag == "PT" else {2.0}, order=0)
    f = lat[np.isfinite(lat)]
    if len(f) < 3:
        continue
    mode = alm.latency_mode(f)
    e05, o05 = alm.latency_background(f, mode, 0.5)
    e10, o10 = alm.latency_background(f, mode, 1.0)
    near = f[np.abs(f - mode) <= 1.0]
    rows.append({"tag": tag, "unit": f"{s['session']}/{u['unit']}", "pulses": len(lat), "mode_ms": mode,
                 "obs_0.5": o05, "chance_0.5": e05, "obs_0.5_to_1": o10 - o05, "chance_0.5_to_1": e10 - e05,
                 "jitter_0.5": f[np.abs(f - mode) <= 0.5].std(ddof=1), "jitter_1": near.std(ddof=1),
                 "jitter_mad_1": 1.4826 * np.median(np.abs(near - np.median(near)))})
bg = pd.DataFrame(rows)
print(bg[bg.tag == "IT"].round(3).to_string(index=False))
for tag, d in bg.groupby("tag"):
    print(f"\n{tag}: {len(d)} neurons; spikes 0.5 to 1 ms from the mode, observed {int(d['obs_0.5_to_1'].sum())}, expected by chance "
          f"{d['chance_0.5_to_1'].sum():.1f}; spikes within ±1 ms expected by chance as a fraction of those observed "
          f"{(d['chance_0.5_to_1'].sum() + d['chance_0.5'].sum()) / (d['obs_0.5'].sum() + d['obs_0.5_to_1'].sum()):.3f}")
    print(f"   mean jitter ±0.5 ms {d['jitter_0.5'].mean():.3f}, ±1 ms {d['jitter_1'].mean():.3f}, robust (MAD) {d['jitter_mad_1'].mean():.3f} ms")
print("\nTagged neurons, mean robust jitter:", round(bg.jitter_mad_1.mean(), 3), "ms;",
      "IT within 0.2 ± 0.05 ms:", bool(abs(bg[bg.tag == 'IT'].jitter_mad_1.mean() - 0.2) <= 0.05))
its = [(s, u) for s, u in units if "IT" in u["cell_types"]]
fig, axs = plt.subplots(4, 7, figsize=(16, 8)); axs = axs.ravel()
for ax, (s, u) in zip(axs, its):
    f = alm.antidromic_latencies(s, u, {2.0}, order=0)
    f = f[np.isfinite(f)]
    if len(f) >= 3:
        mode = alm.latency_mode(f)
        ax.hist(f, bins=np.arange(0, 10.01, 0.1), color="k")
        for hw, c in ((0.5, "tab:blue"), (1.0, "tab:red")):
            ax.axvline(mode - hw, color=c, lw=0.8); ax.axvline(mode + hw, color=c, lw=0.8)
    ax.set_title(f"{s['session'][3:]}/{u['unit']} n={len(f)}", fontsize=7); ax.tick_params(labelsize=6)
for ax in axs[len(its):]:
    ax.axis("off")
fig.suptitle("IT neurons, first-spike latency after the first pulse (ms; 0.1 ms bins), ±0.5 ms (blue) and ±1 ms (red) around the mode")
plt.tight_layout(); plt.show()
print("Histograms span 0 to 10 ms, the whole latency window; IT neurons without type 2 responses are empty panels.")
```

```python
# Readings without the window. (1) Per neuron, a maximum-likelihood fit to all first-spike
# latencies (0 to 10 ms) of a Gaussian peak plus a uniform background; (2) the same with a broad
# Gaussian (s.d. 1 to 5 ms) added to take up broad stimulus-locked humps; the jitter is the
# narrow peak's s.d. (3) The ±1 ms jitter without the four IT neurons whose histograms above show
# humps several ms wide (named in the round-2 verification).
rows = []
for s, u in units:
    tag = "PT" if "PT" in u["cell_types"] else "IT" if "IT" in u["cell_types"] else ""
    if not tag:
        continue
    lat = alm.antidromic_latencies(s, u, {1.0} if tag == "PT" else {2.0}, order=0)
    if np.isfinite(lat).sum() < 3:
        continue
    mu, sd, w, n = alm.latency_peak_fit(lat)
    mub, sdb, wb, _ = alm.latency_peak_fit(lat, broad=True)
    rows.append({"tag": tag, "unit": f"{s['session']}/{u['unit']}", "latencies": n, "peak_ms": mu, "jitter_fit": sd,
                 "peak_weight": w, "jitter_fit_broad": sdb, "peak_weight_broad": wb,
                 "jitter_1": alm.latency_mode_stats(lat, 1.0)[1], "jitter_0.5": alm.latency_mode_stats(lat, 0.5)[1]})
pf = pd.DataFrame(rows)
print(pf[pf.tag == "IT"].round(3).to_string(index=False))
hump = ["ANM219253_20140122/4", "ANM219253_20140117/1", "ANM214427_20130808/2", "ANM219248_20130924/2"]
pf["broad_hump"] = pf.unit.isin(hump)
summ = []
for name, col, d in (("±1 ms window", "jitter_1", pf), ("±0.5 ms window", "jitter_0.5", pf),
                     ("peak + background fit", "jitter_fit", pf), ("peak + broad + background fit", "jitter_fit_broad", pf),
                     ("±1 ms, without the 4 broad-hump IT neurons", "jitter_1", pf[~pf.broad_hump])):
    summ.append({"reading": name, "PT_n": int((d.tag == "PT").sum()), "IT_n": int((d.tag == "IT").sum()),
                 "PT_jitter": d[d.tag == "PT"][col].mean(), "IT_jitter": d[d.tag == "IT"][col].mean(), "tagged_jitter": d[col].mean()})
summ = pd.DataFrame(summ)
summ["PT_within_0.1"] = abs(summ.PT_jitter - 0.1) <= 0.05
summ["IT_within_0.2"] = abs(summ.IT_jitter - 0.2) <= 0.05
summ["tagged_within_0.15"] = abs(summ.tagged_jitter - 0.15) <= 0.015
print("\nFirst pulses, IT type 2 trials; neurons with at least 3 latencies:")
print(summ.round(3).to_string(index=False))
print("\nSpikes of the 4 broad-hump neurons within 0.5 to 1 ms of the mode:",
      int(bg[bg.unit.isin(hump)]["obs_0.5_to_1"].sum()), "of", int(bg[bg.tag == "IT"]["obs_0.5_to_1"].sum()), "for all IT neurons")
```

**Result: partially reproduced, resting on the assumed ±1 ms window.** With antidromic spikes taken within ±1 ms of each neuron's modal latency, first pulses, PT neurons have a mean latency of 2.672 ms and jitter 0.126 ms (45 neurons), within tolerance, and IT neurons 4.98 ms (21 neurons with type 2 trials; 4.56 ms with types 2 to 5, 27 neurons), within tolerance and longer than PT. ## fig3f-late-contra-selectivity-pt

> Robust contralateral selectivity emerges in PT neurons starting 570 ms before the response
> epoch (34% spike rate modulation, P < 0.05), but not in IT neurons (-19%, P = 0.40) nor in all
> task-selective neurons (8%, P = 0.13).

Contra-selectivity is R minus L rate per neuron (correct trials, 200 ms windows, 10 ms step).
Onset readings (a) the earliest window centre from which the PT mean stays significantly above
0 (t-test across neurons, P < 0.05) through the window centred 0.1 s before the cue, (b) the
first significant window after sample onset. Modulation windows (a) 570 ms before the cue to
the cue, (b) to the end of the response epoch; percentage of (a) the mean rate over both trial
types in the window, (b) the mean rate in the 0.5 s before sample onset. Neurons (a) all PT
(IT), (b) the selective ones; "all task-selective" is the selective pyramidal neurons. Every
combination is run under the four readings of the shared methods (lick early trials dropped or
kept, response epoch 1.3 or 1.5 s), with the selective neurons taken under the same reading.

```python
def cs_curve(keys, le="drop"):
    out = []
    for s, u in units:
        if (s["session"], u["unit"]) not in keys:
            continue
        tr = s["trials"]
        r, l = alm.trial_mask(tr, u, lick_early=le)
        if r.sum() < 2 or l.sum() < 2:
            continue
        out.append((alm.window_sums(alm.cue_binned(u, tr, r))[0].mean(axis=0) - alm.window_sums(alm.cue_binned(u, tr, l))[0].mean(axis=0)) / alm.WIN_S)
    return np.array(out)

def win_rates(keys, a_rel, b_rel, le="drop"):
    rows = []
    for s, u in units:
        if (s["session"], u["unit"]) not in keys:
            continue
        tr = s["trials"]
        r, l = alm.trial_mask(tr, u, lick_early=le)
        if r.sum() < 2 or l.sum() < 2:
            continue
        a, b = tr.cue.values + a_rel, tr.cue.values + b_rel
        pa, pb_ = tr.pole_in.values - 0.5, tr.pole_in.values
        R = alm.counts_in(u, tr, r, a, b).mean() / (b_rel - a_rel)
        L = alm.counts_in(u, tr, l, a, b).mean() / (b_rel - a_rel)
        base = (alm.counts_in(u, tr, r, pa, pb_).mean() + alm.counts_in(u, tr, l, pa, pb_).mean()) / 2 / 0.5
        rows.append((R, L, base))
    return np.array(rows)

keyset = lambda d: set(map(tuple, d[["session", "unit"]].values))

def fig3f(le, resp):
    d = UT[(le, resp)]
    groups = {"PT a all": keyset(d[d.tag == "PT"]), "PT b selective": keyset(d[(d.tag == "PT") & d.selective]),
              "IT a all": keyset(d[d.tag == "IT"]), "IT b selective": keyset(d[(d.tag == "IT") & d.selective]),
              "all task-selective": keyset(d[(d.label == "pyramidal") & d.selective])}
    onsets = []
    for name in ("PT a all", "PT b selective"):
        C = cs_curve(groups[name], le)
        p = stats.ttest_1samp(C, 0, axis=0).pvalue
        sigpos = (p < 0.05) & (C.mean(axis=0) > 0)
        idx = np.flatnonzero((centers >= t_sample) & (centers <= -0.1))
        stay = None
        for i in idx[::-1]:
            if sigpos[i]:
                stay = centers[i]
            else:
                break
        first = centers[idx][sigpos[idx]][0] if sigpos[idx].any() else np.nan
        onsets.append({"lick_early": le, "response_s": resp, "neurons": name, "n": len(C),
                       "onset_a_stays_ms_before_cue": -1e3 * stay if stay is not None else np.nan,
                       "onset_b_first_ms_before_cue": -1e3 * first})
    rows = []
    for wname, (a_rel, b_rel) in (("a -0.57 to cue", (-0.57, 0.0)), ("b -0.57 to response end", (-0.57, resp))):
        for name, keys in groups.items():
            X = win_rates(keys, a_rel, b_rel, le)
            dlt = X[:, 0] - X[:, 1]
            rows.append({"lick_early": le, "response_s": resp, "window": wname, "neurons": name, "n": len(X),
                         "mean_R_minus_L": dlt.mean(), "p": stats.ttest_1samp(dlt, 0).pvalue,
                         "mod_pct_a_window_rate": 100 * dlt.mean() / ((X[:, 0] + X[:, 1]) / 2).mean(),
                         "mod_pct_b_baseline": 100 * dlt.mean() / X[:, 2].mean()})
    return pd.DataFrame(onsets), pd.DataFrame(rows)

ons_l, mod_l = [], []
for le, resp in UT:
    o, m = fig3f(le, resp)
    ons_l.append(o); mod_l.append(m)
ons, mod = pd.concat(ons_l, ignore_index=True), pd.concat(mod_l, ignore_index=True)
ons["a_ok"] = abs(ons.onset_a_stays_ms_before_cue - 570) <= 100
ons["b_ok"] = abs(ons.onset_b_first_ms_before_cue - 570) <= 100
print("Onsets under the four readings of the shared methods (lick early trials, response epoch):")
print(ons.round(1).to_string(index=False))
print("\nModulation and t-tests, lick early dropped, 1.3 s response epoch:")
print(mod[(mod.lick_early == "drop") & (mod.response_s == 1.3)].round(4).to_string(index=False))
combos = []
for (le, resp, w), m in mod.groupby(["lick_early", "response_s", "window"]):
    m = m.set_index("neurons")
    for nr in ("a all", "b selective"):
        for pct in ("mod_pct_a_window_rate", "mod_pct_b_baseline"):
            pt_, it_, al_ = m.loc[f"PT {nr}"], m.loc[f"IT {nr}"], m.loc["all task-selective"]
            combos.append({"lick_early": le, "response_s": resp, "window": w, "neurons": nr, "percent_of": pct[8:],
                           "PT_p": pt_.p, "IT_p": it_.p, "all_p": al_.p, "PT_pct": pt_[pct], "IT_pct": it_[pct], "all_pct": al_[pct],
                           "values_ok": abs(pt_[pct] - 34) <= 3.4 and abs(it_[pct] + 19) <= 1.9 and abs(al_[pct] - 8) <= 0.8})
cmb = pd.DataFrame(combos)
cmb["pattern_ok"] = (cmb.PT_p < 0.05) & (cmb.IT_p >= 0.05) & (cmb.all_p >= 0.05)
print("\nEvery combination of readings (P values of the t-tests against 0; percentages):")
print(cmb.round(3).to_string(index=False))
print("\nCombinations with the paper's pattern of significance:", int(cmb.pattern_ok.sum()), "of", len(cmb),
      "; with the pattern and all values within tolerance:", int((cmb.pattern_ok & cmb.values_ok).sum()))
print("All task-selective neurons, window a: P under the four readings:",
      mod[(mod.neurons == "all task-selective") & (mod.window == "a -0.57 to cue")][["lick_early", "response_s", "n", "p"]].round(4).to_dict("records"))
```

5 ms, further from its pre-cue baseline than k times the baseline noise and at least a floor
(k = 3, 5, 10; floor 2 or 5 mV). The peak rule takes the first sample at which the trace has
moved by a fraction (5% to 50%) of its largest excursion in the 1 s after the cue. Trials (a)
correct trials without early licks, (b) all performing trials with a response (hit or error);
both only trials of type 0, without axonal stimulation or the NaN-type test stimulation. Per
mouse the mean, then mean and s.d. across mice; per-mouse medians are a variant. Example
traces with the onsets marked, and per-mouse histograms of the onsets, follow the table.

## contra-ipsi-selectivity-forms

> Contra-preferring neurons include most of the neurons with large spike count differences
> between trial types, while ipsi-preferring neurons more often express their selectivity
> through suppression of activity on contra trials.

Per selective pyramidal neuron, the absolute difference of R and L rates from sample onset to
the end of the response epoch. Contra-suppression: the rate on R trials (same span) below the
rate in the 0.5 s before sample onset, with the rate on L trials not below it; ipsi-suppression
is the same with the trial types swapped. The whole pre-sample period is a variant baseline.

```python
sp = pyr[pyr.selective].copy()
sp["absdiff"] = (sp.trial_R - sp.trial_L).abs()
rows = []
for q, name in ((0.75, "a top quarter"), (0.90, "b top tenth")):
    big = sp[sp.absdiff >= sp.absdiff.quantile(q)]
    rows.append({"reading": name, "neurons": len(big), "fraction_contra": big.contra.mean(), "binomial_p": stats.binomtest(int(big.contra.sum()), len(big)).pvalue})
large = pd.DataFrame(rows)
large["contra_majority"] = large.fraction_contra > 0.5
print(large.round(4).to_string(index=False))
rows = []
for bname, bcol in (("0.5 s pre-sample", "pre"), ("whole pre-sample (variant)", "pre_all")):
    base = (sp[f"{bcol}_R"] + sp[f"{bcol}_L"]) / 2
    csup = (sp.trial_R < base) & ~(sp.trial_L < base)
    isup = (sp.trial_L < base) & ~(sp.trial_R < base)
    ip, cp = sp[~sp.contra], sp[sp.contra]
    rows.append({"baseline": bname, "ipsi_pref_n": len(ip), "ipsi_pref_with_contra_suppression": float(csup[~sp.contra].mean()),
                 "contra_pref_n": len(cp), "contra_pref_with_ipsi_suppression": float(isup[sp.contra].mean()),
                 "fisher_p": stats.fisher_exact([[int(csup[~sp.contra].sum()), len(ip) - int(csup[~sp.contra].sum())],
                                                 [int(isup[sp.contra].sum()), len(cp) - int(isup[sp.contra].sum())]])[1]})
sup = pd.DataFrame(rows)
sup["ipsi_more"] = sup.ipsi_pref_with_contra_suppression > sup.contra_pref_with_ipsi_suppression
print(sup.round(4).to_string(index=False))
print("\nUnder the four readings of the shared methods (0.5 s baseline):")
rows = []
for (le, resp), d in UT.items():
    x = d[(d.label == "pyramidal") & d.selective].copy()
    x["absdiff"] = (x.trial_R - x.trial_L).abs()
    base = (x.pre_R + x.pre_L) / 2
    csup = (x.trial_R < base) & ~(x.trial_L < base)
    isup = (x.trial_L < base) & ~(x.trial_R < base)
    rows.append({"lick_early": le, "response_s": resp,
                 "contra_frac_top_quarter": x[x.absdiff >= x.absdiff.quantile(0.75)].contra.mean(),
                 "contra_frac_top_tenth": x[x.absdiff >= x.absdiff.quantile(0.9)].contra.mean(),
                 "ipsi_pref_contra_supp": float(csup[~x.contra].mean()), "contra_pref_ipsi_supp": float(isup[x.contra].mean())})
ci4 = pd.DataFrame(rows)
ci4["both_parts"] = (ci4[["contra_frac_top_quarter", "contra_frac_top_tenth"]].max(axis=1) > 0.5) & (ci4.ipsi_pref_contra_supp > ci4.contra_pref_ipsi_supp)
print(ci4.round(4).to_string(index=False))
```

**Result: partially reproduced.** Contra-preferring neurons are a slight majority of the neurons with the largest rate differences (0.535 of the top quarter, 228 neurons, binomial p = 0.32; 0.522 of the top tenth, p = 0.75). Contra-suppression in ipsi-preferring neurons (0.365) is not more common than ipsi-suppression in contra-preferring neurons (0.389, Fisher p = 0.49); the whole pre-sample baseline gives the same (0.367 against 0.391).
