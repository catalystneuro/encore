The tables are put into tidy form. Each check of the tables against what the paper and the

```python
crab_f = stns.crab_ramp_frequencies(crab_x)
crab_s = stns.crab_saline_series(crab_x)
crab_u = stns.crab_unit_crashes(crab_x)
crab_u["rhythm"] = crab_u.apply(stns.rhythm_crash, axis=1)
lob_f, lob_r = stns.lobster_ramp_frequencies(lob_x)
print("Crab ramps, PD burst frequency (Hz) per set and salinity:")
print(crab_f.groupby(["set", "salinity"]).agg(preps=("prep", "nunique"), points=("freq", "size"),
      tmin=("temp", "min"), tmax=("temp", "max"), fmin=("freq", "min"), fmax=("freq", "max")).to_string())
print("\nCrab salinity series at 10 C (Hz):")
print(crab_s.groupby(["series", "salinity"]).agg(preps=("prep", "size"), with_value=("freq", "count"),
      fmin=("freq", "min"), fmax=("freq", "max")).to_string())
print("\nCrab neuron crash temperatures (C), per set and salinity:")
print(crab_u.groupby(["set", "salinity"]).agg(ramps=("prep", "size"), LP=("LP", "count"), PY=("PY", "count"),
      PD=("PD", "count"), tmin=("LP", "min"), tmax=("LP", "max")).to_string())
print("\nLobster ramps, PD burst frequency (Hz) per salinity (Sheet4):")
print(lob_f.groupby("salinity").agg(preps=("prep", "nunique"), points=("freq", "size"), tmin=("temp", "min"),
      tmax=("temp", "max"), fmin=("freq", "min"), fmax=("freq", "max")).to_string())
print("\nLobster ramps marked 'Not used':", lob_r[lob_r.not_used][["salinity", "prep"]].values.tolist())
print("Lobster ramps with a 'Crash' step:", lob_r.crash_step.notna().groupby(lob_r.salinity).sum().to_dict(),
      "of", lob_r.groupby("salinity").size().to_dict())
```

Units are not stated in the tables; the values (crab 0.7 to 4.9, lobster 0.1 to 2.8 except one
value, see below) are burst frequencies in Hz, the paper's unit, and temperatures in C on the
paper's 2 C steps from 10 C. The crab sets hold 11 preparations (0.75x set) and 10 (1.25x set),
as the paper's 11 (Fig. 2B) and 10 (Fig. 7C).

```python
# The Prism data the paper's Q10 fits used are the Excel values: match each Prism column to a
# preparation by its values.
def match_columns(wide, tidy, sal, lo=10, hi=30):
    rows = []
    for c in wide.columns:
        v = wide[c].loc[lo:hi]
        if v.notna().sum() == 0:
            continue
        best = []
        for p, g in tidy[tidy.salinity == sal].groupby("prep"):
            s = g.drop_duplicates("temp").set_index("temp").freq.reindex(v.index)
            both = v.notna() & s.notna()
            if both.sum() and np.allclose(v[both], s[both], atol=1e-5) and both.sum() == v.notna().sum():
                best.append(p)
        rows.append({"prism_column": c, "points": int(v.notna().sum()), "matches": best})
    return pd.DataFrame(rows)

links = {}
for sal, key in ((1.0, "1x"), (0.75, "0.75x"), (1.25, "1.25x")):
    g = stns.prism_xy_groups(pr_freq[key])
    for title, w in g.items():
        tidy = crab_f if title.startswith("Canbo") else lob_f
        m = match_columns(w, tidy, sal)
        links[title] = m
        n_one = (m.matches.str.len() == 1).sum()
        print(f"{title}: {len(m)} columns with data; matched to exactly one preparation: {n_one}; "
              f"to several: {(m.matches.str.len() > 1).sum()}; to none: {(m.matches.str.len() == 0).sum()}")
        for _, r in m[m.matches.str.len() != 1].iterrows():
            print(f"    column {r.prism_column} ({r.points} points) matches {r.matches}")
print("\nLobster preparations in Sheet4 but in no Prism column:")
for sal, key in ((1.0, "Homam 1x"), (0.75, "Homam 0.75x"), (1.25, "Homam 1.25x")):
    used = {p for ps in links[key].matches for p in ps}
    print(f"  {sal}x:", sorted(set(lob_f[lob_f.salinity == sal].prep) - used))
```

Every Prism column of the frequency file is a preparation of the Excel tables, value for
value. One lobster 0.75x column matches two preparations, 111122 and 120722: their 0.75x
frequencies are identical at every step to 28 C. This is shown as a finding below. Lobster
020223 (marked "Not used") is in no fit.

```python
# Crash temperatures: the rhythm crash of the crab tables (second neuron to crash, the Methods'
# definition) against the Prism crash-temperature file the paper's species comparisons used.
pc = {k: v["table"] for k, v in pr_crash.items()}
for key, sal in (("1x", 1.0), ("0.75x", 0.75), ("1.25x", 1.25)):
    t = pc[key]
    cancer = t[0].dropna().values
    homarus = t[1].dropna().values
    mine = crab_u[crab_u.salinity == sal].rhythm.dropna().values
    same = len(mine) == len(cancer) and np.allclose(np.sort(mine), np.sort(cancer))
    print(f"{key}: Prism Cancer n={len(cancer)} (mean {cancer.mean():.2f}); crab table rhythm crash n={len(mine)} "
          f"(mean {mine.mean():.2f}); same values: {same}; Prism Homarus n={len(homarus)} (mean {homarus.mean():.2f}, "
          f"range {homarus.min()} to {homarus.max()})")
print("\nCrab ramps with fewer than two neuron crashes (no rhythm crash):")
print(crab_u[crab_u.rhythm.isna()][["set", "salinity", "prep", "LP", "PY", "PD"]].to_string(index=False))
# Lobster: the step marked 'Crash' in Sheet4 against the Prism crash temperatures (unlabelled).
for sal, key in ((1.0, "1x"), (0.75, "0.75x"), (1.25, "1.25x")):
    r = lob_r[(lob_r.salinity == sal) & ~lob_r.not_used]
    print(f"Lobster {key}: Sheet4 crash steps {sorted(r.crash_step.dropna().tolist())}; "
          f"ramps without a Crash mark {r.crash_step.isna().sum()}; Prism crash temperatures {sorted(pc[key][1].dropna().tolist())}")
```

The crab crash temperatures in the Prism file are the second-neuron crash of the crab unit table,
value for value, so the rhythm crash follows the Methods' definition. Crab 963_109 has no crash
temperature at either salinity, so 1x has 20 crab values; the paper gives n = 21 for Fig. 7E.
The lobster Prism values carry no preparation labels and are given to 0.1 C, finer than the 2 C

```python
def neuron_mixed(g):
    long = g.melt(id_vars=["prep"], value_vars=["LP", "PY", "PD"], var_name="neuron", value_name="crash").dropna()
    m = smf.mixedlm("crash ~ C(neuron)", long, groups=long["prep"]).fit(reml=True)
    names = [n for n in m.params.index if n.startswith("C(neuron)")]
    wald = m.wald_test(" , ".join(f"{n} = 0" for n in names), scalar=True)
    chi2 = float(wald.statistic)
    k = len(names)
    n_obs, n_prep = len(long), long.prep.nunique()
    df2 = n_obs - n_prep - k
    return long, chi2, k, float(stats.chi2.sf(chi2, k)), float(stats.f.sf(chi2 / k, k, df2)), df2

rows = []
for sal, g in ((1.0, one_x), (0.75, s075u), (1.25, s125u)):
    long, chi2, k, p_chi2, p_f, df2 = neuron_mixed(g)
    means = long.groupby("neuron").crash.agg(["count", "mean", "std"]).round(2)
    print(f"{sal}x: neuron means\n{means.to_string()}")
    complete = g.dropna(subset=["LP", "PY", "PD"])
    fr = stats.friedmanchisquare(complete.LP, complete.PY, complete.PD) if len(complete) >= 3 else None
    rows.append({"salinity": sal, "obs": len(long), "preps": long.prep.nunique(), "Wald_chi2": round(chi2, 3),
                 "P_chi2": round(p_chi2, 3), f"P_F(k,resid)": round(p_f, 3), "df2": df2,
                 "complete_preps": len(complete), "Friedman_P": round(fr.pvalue, 3) if fr else np.nan})
fig2i_t = pd.DataFrame(rows)
print(fig2i_t.to_string(index=False))
order = one_x.apply(lambda r: "".join(n for n, _ in sorted(((n, r[n]) for n in ("LP", "PY", "PD") if pd.notna(r[n])), key=lambda x: x[1])), axis=1)
print("1x crash order per ramp (first to last):", order.value_counts().to_dict())
```

```python
def q10_compare(groups, label):
    out = {}
    for rd in ("Q-a", "Q-b"):
        r = stns.shared_slope_test(groups, rd)
        fits = "; ".join(f"{g}: Q10 {f['q10']:.3f} [{f['q10_lo']:.2f}, {f['q10_hi']:.2f}], n = {f['n']}" for g, f in r["fits"].items())
        print(f"{label} {rd}: {fits}; F({r['df1']},{r['df2']}) = {r['F']:.3f}, P = {r['p']:.4f}, points {r['n_points']}")
        out[rd] = r
    return out

def tf(df, lo, hi):
    d = df[(df.temp >= lo) & (df.temp <= hi)]
    return d.temp.values, d.freq.values

c075 = crab_f[crab_f.set == "0.75x"]
fig2g = q10_compare({"1x": tf(c075[c075.salinity == 1.0], 10, 24), "0.75x": tf(c075[c075.salinity == 0.75], 10, 24)},
                    "Crab 0.75x set, 10-24 C")
print("Preparations:", c075.prep.nunique())
```

## Crab mixed-effects model (Fig. 2F legend)

```python
def mixed_model(d, sal_factor=False, log=False, label=""):
    d = d.copy()
    d["y"] = np.log10(d.freq) if log else d.freq
    sal = "C(salinity)" if sal_factor else "salinity"
    form = f"y ~ temp * {sal}"
    fits = {}
    for re_name, re_f in (("intercept", "1"), ("temp", "~temp"), ("salinity", "~salinity"), ("temp+salinity", "~temp+salinity")):
        try:
            m = smf.mixedlm(form, d, groups=d["prep"], re_formula=None if re_f == "1" else re_f).fit(reml=False, method=["lbfgs", "powell"])
            fits[re_name] = (m.aic if np.isfinite(m.aic) else np.inf, m.bic if np.isfinite(m.bic) else np.inf, m.converged)
        except Exception as e:
            fits[re_name] = (np.inf, np.inf, False)
    failed = [k for k, v in fits.items() if not np.isfinite(v[0])]
    best = min(fits, key=lambda k: (fits[k][0], fits[k][1]))
    re_f = {"intercept": None, "temp": "~temp", "salinity": "~salinity", "temp+salinity": "~temp+salinity"}[best]
    m = smf.mixedlm(form, d, groups=d["prep"], re_formula=re_f).fit(reml=True, method=["lbfgs", "powell"])
    fe = m.fe_params.index   # fixed effects only (m.params also holds the random-effect variances)
    terms = {"temperature": [n for n in fe if n == "temp"],
             "salinity": [n for n in fe if n.startswith(sal) and ":" not in n],
             "interaction": [n for n in fe if ":" in n]}
    res = {}
    for t, names in terms.items():
        w = m.wald_test(" , ".join(f"{n} = 0" for n in names), scalar=True)
        res[t] = float(w.pvalue)
    print(f"{label}: n = {len(d)} points, {d.prep.nunique()} preparations; random effects by AIC/BIC: {best} "
          f"(AIC/BIC {', '.join(f'{k} {v[0]:.1f}/{v[1]:.1f}' for k, v in fits.items())}); "
          f"P temperature {res['temperature']:.2g}, salinity {res['salinity']:.2g} ({len(terms['salinity'])} df), "
          f"interaction {res['interaction']:.2g}; candidates whose ML fit failed (left out): {failed or 'none'}")
    return res

crab_all = crab_f.copy()
readings = {"S-a (1.25x set: 1x, 1.25x)": crab_all[crab_all.set == "1.25x"],
            "S-b (0.75x set: 1x, 0.75x)": crab_all[crab_all.set == "0.75x"],
            "S-c (all ramps)": crab_all}
fig2f_mm = {}
for name, d in readings.items():
    for log in (False, True):
        for fac in ((False, True) if name.startswith("S-c") else (False,)):
            lab = f"{name}, {'F-b log' if log else 'F-a'}, salinity {'factor' if fac else 'number'}"
            fig2f_mm[lab] = mixed_model(d, sal_factor=fac, log=log, label=lab)
tab = pd.DataFrame(fig2f_mm).T
tab["pattern_holds"] = (tab.temperature < 0.05) & (tab.interaction < 0.05) & (tab.salinity > 0.05)
tab["two_of_three"] = ((tab.temperature < 0.05).astype(int) + (tab.interaction < 0.05).astype(int) + (tab.salinity > 0.05).astype(int)) >= 2
print(tab.to_string())
```

The paper fitted lme4 in R and tested terms with lmerTest's F tests (Satterthwaite degrees of
freedom). R is not available in this environment; the model here is statsmodels' linear mixed
model (REML, random intercept per preparation, random slopes chosen by AIC then BIC among the
same candidates), with Wald tests of each fixed-effect term. Wald tests are anti-conservative
next to Satterthwaite F tests with few denominator df: a Wald P above 0.05 would stay above it,
and a Wald P far below 0.05 would stay below. The Bayes factor comparison is not repeated. A
candidate whose maximum-likelihood fit statsmodels could not evaluate (non-finite likelihood; the
intercept-only model, mostly) is listed as failed and left out of the choice. Salinity as a
factor (S-c, three salinities) has 2 df, where the criterion asks for 1; salinity as a number
has 1. (The first run of this notebook tested salinity jointly with the variance of its random
slope where the model had one; the tests here use the fixed effects only.)

**Result: reproduced.** Under S-c with frequency as the response and salinity as a number (one
df, all 21 preparations), temperature (P = 1.5e-19) and the interaction (P = 0.0004) are
significant and salinity is not (P = 0.16; paper 0.17), the paper's pattern. The other readings
give two of the three parts, except S-a with log frequency (one): under S-b, the 0.75x set of

## In vivo PPG recordings (Fig. 8)

Each exposure is a zip of 2-min ABF files (5 kHz, one channel) inside one of four zips of record
17794764. Every exposure is read, each file is decimated to 50 Hz, the files are joined in the
order of their start times, and a spectrogram over 0 to 3 Hz is made with 10-s windows and 50%
overlap (the claims file's standard choices). The claims file's rule gives a window a rhythm
when its peak power between 0.3 and 3 Hz is above three times the window's median power
(reading B-a over the hour, B-b over the first 30 min). As a reference the same rule is applied
to the same trace with its samples shuffled (a flat spectrum), under the claim's seed and four
more. Three variants calibrated on the shuffled trace are this attempt's: the per-window peak
ratio above the 99th percentile of the shuffled trace's; and on the mean spectrum of each 1-min
block (12 windows), the peak ratio, or the prominence of the largest local peak between 0.4 and
2.8 Hz over the power 0.3 to 0.5 Hz on either side, above the shuffled trace's 99th percentile.
An exposure keeps its rhythm under a rule when at least half its windows or blocks have a band.
The reads run in eight threads; the time they took is printed below.

A failure to reproduce here may be a failure of this attempt, not of the paper: see the
reasoning and the differences from the paper's method.

**How the analysis differs from the paper's.** The isolated-STNS claims use the authors' own
per-preparation values from the deposited tables; the burst detection on the raw recordings is
not repeated. The Q10 fits reproduce the deposited Prism outputs to every printed digit under the
semi-log model (Q-a). The mixed models are statsmodels' linear mixed models with Wald tests, not
lme4 with lmerTest's Satterthwaite F tests, and Prism's REML repeated-measures model is replaced
by a mixed model with a random intercept per preparation, without the Geisser-Greenhouse
correction. The in vivo rhythm is judged by the claims file's band rule and three rules calibrated on
shuffled traces, not by eye; they disagree on the crab exposures, and the claim is left blocked.

**What was found about the datasets.** The deposit holds the authors' analysis tables and
Prism files, which reproduce exactly the paper's Q10 statistics of Figs. 2G, 2H, 3B and 7A to 7C;
those of Fig. 5D and of the PY-on phase (Fig. 3C) differ. In them, lobster 111122's 0.75x
ramp is a copy of 120722's (the 111122 recording itself is deposited and differs); one crab 1.25x
crash temperature is 36 C in the Prism file and 26 C in the unit table, the source of the paper's
two crab 1.25x means; and one lobster 1x value is 4.68 Hz. In the in vivo data, seven exposures
are the same recordings as seven others under another animal or salinity, and each "1-h"
exposure holds 32 min. The lobster salinity series, the lobster neuron crash temperatures, the
raw recording of lobster 888_156 and the 15-h survival records are not in either record, and no
analysis script, which the paper says is deposited, was found in the files read (the 21 inner
crab zips were read only at their start). The deposited Prism sheets and MATLAB files hold the
