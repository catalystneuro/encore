# Reproducing ten Brinke et al. (2017) from CRCNS cb-1

**Paper.** ten Brinke MM, Heiney SA, Wang X, Proietti-Onori M, Boele HJ, Bakermans J, Medina JF, Gao Z,
De Zeeuw CI (2017). Dynamic modulation of activity in cerebellar nuclei neurons during pavlovian eyeblink
conditioning in mice. *eLife* 6:e28132. [doi:10.7554/eLife.28132](https://doi.org/10.7554/eLife.28132).
The claims file (`claims/brinke-2017.yaml`) was written from the version of record (PMC5760204) with its
figure legends, Methods and Supplementary file 1; this attempt read the same text.

**Data set.** CRCNS cb-1, "Eyelid behavior and spike activity of cerebellar interpositus nucleus neurons
during eyeblink conditioning in awake behaving mice", [doi:10.6080/K0B85697](https://doi.org/10.6080/K0B85697),
read anonymously from the public S3 bucket `crcnsarchive` under `cb-1/live/`. CRCNS does not version its
data sets and the paper cites the data set's DOI without a version, so the version is recorded as
`unversioned`: the files as published under `cb-1/live/` (last modified 2026-02-11 in the bucket; the
description is version 0.6 of 13 Dec 2017). There is no other version to choose. The one data file is
`eLife_dataset.mat`, MD5 511cf0f257240365977d2c5960766f5d in the data set's `checksums.md5`; the notebook
streams it into memory and checks the MD5 (the description PDF is fa19108c9385b7e66d6df95e68a30c99 and the
example script `eLife_dataset.m` dce22a5c204e00f10b692f1b07a3eae0). A later change to the file would show as
a mismatch in the first cell.

**What the data set holds.** 372 recordings of interpositus nucleus neurons: 270 of the first dataset
(recorded over training, CS-US interval 250 ms) and 102 of the second (after training, 212 ms by the
description), each with per-trial eyelid traces, spike density traces, spike times, a per-trial
facilitation and CR amplitude, the mouse and the training day. It holds no recording depth, no US-only
trials, no optogenetic sessions and no Purkinje cell recordings.

Outside data. Recording depth is taken from the paper's Source data 1 (Additional files of
[doi:10.7554/eLife.28132](https://doi.org/10.7554/eLife.28132)), the file of the version of record on the eLife
site (`elife-28132-data1-v2.mat`, MD5 3ca8f91c5371e88c12f5eb32b3b209c6, checked when it is streamed), as the
claims file's opening comment names it. Only its `Depth` is used; the link claim `source-data-1-link` matches
its cells to the data set's by order and checks the values both hold. Source data 1 gives no depth for the
second dataset's 102 cells.

**Method details taken from the paper, and readings.** The helper module `ipn.py` says where each parameter
comes from. Where the claims file names readings, every one is run; the class of a cell (facilitation,
suppression, none) is run under the claims file's two baseline-correction readings, (a) and (b), and under a
variant of this attempt's own, (a+), the Methods' "mean above-baseline activity within the baseline itself"
averaged over the above-baseline bins only; no outcome rests on (a+) alone. Reading (b) on each cell's average
trace gives the authors' per-cell facilitation in Source data 1. The second dataset's classes use the claims
file's two windows, 12-212 ms (adopted) and 0-200 ms; 0-212 ms is shown as a variant where the verifier asked for
it. CR amplitude at US onset is run as the data set's own `crAmplitude` (which this notebook finds is
the eyelid 13 ms before the stated US onset) and as computed at the stated US onset. The CS pause and peak are run under the four threshold readings of
the claims file, the two windows (50-125 ms in the Methods, 50-130 ms in the Results) and two places to
look for the rebound (within the window, or up to 50 ms after the trough, marked "+50", this attempt's
variant). The spike-eyelid cross-correlation lags use an estimator the paper does not name: the adopted one is
the cross-covariance of the two mean-subtracted traces over the segment (as MATLAB's `xcov`, 'coeff'), and the
Pearson r of the spike segment with the eyelid re-cut at each lag is shown as a variant. The Methods' trial
exclusions are not applied again: Source data 1 lists exactly the deposited trials of each cell as its valid
trials. The Statistics rule
for correlations needs a normality test the paper does not name; Lilliefors is used, and a Spearman-only
count is printed beside each trial-by-trial count. Bootstrap p-values are
fractions of 500 random datasets, so 0.000 means below 1/500. Each claim's section has its own random
generator with a fixed seed.

```python
import resource
import time
import warnings

import numpy as np
import pandas as pd
from scipy import stats

import ipn

warnings.filterwarnings("ignore", category=RuntimeWarning)
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 40)
t_start = time.time()
sums = ipn.checksums()
D = ipn.load()
print("data set", ipn.DATASET, ipn.VERSION, "doi", ipn.DOI)
print("MD5 of", ipn.FILE, D["md5"], "published", sums[ipn.FILE], "match", D["md5"] == sums[ipn.FILE])
df = ipn.cell_table(D)
t = D["t"]
d1, d2 = df[df.dset == 1], df[df.dset == 2]
print("loaded in", round(time.time() - t_start, 1), "s")
```

```python
print("recordings:", len(df), "set 1:", len(d1), "set 2:", len(d2))
print("US onset (ms) by set:", sorted(set(d1.us)), sorted(set(d2.us)))
print("time axis:", t[0], "to", t[-1], "ms, step", np.unique(np.diff(t)))
print("mice: set 1", d1.mouse.nunique(), sorted(int(m) for m in d1.mouse.unique()), "| set 2", d2.mouse.nunique(), sorted(int(m) for m in d2.mouse.unique()))
print("shared mice between sets:", sorted(set(d1.mouse) & set(d2.mouse)))
print("recordings per day code, set 1:", d1.day.value_counts().sort_index().to_dict())
print("day codes, set 2:", d2.day.value_counts().to_dict())
print("days 1-3:", (d1.day <= 3).sum(), " days 1-4:", (d1.day <= 4).sum(), " later:", (d1.day > 4).sum())
print("trials per recording: set 1 min", d1.n_trials.min(), "max", d1.n_trials.max(), "total", d1.n_trials.sum(),
      "| set 2 min", d2.n_trials.min(), "max", d2.n_trials.max(), "total", d2.n_trials.sum())
print("recordings per mouse, set 1:", d1.mouse.value_counts().sort_index().to_dict())
# value ranges
ey = np.concatenate([e[:, ~np.isnan(e).all(0)].ravel() for e in D["eyelid"]])
sd_all = np.concatenate([s[:, ~np.isnan(s).all(0)].ravel() for s in D["sdf"]])
print(f"eyelid (normalized, 0 open, 1 full blink): {np.nanmin(ey):.3f} to {np.nanmax(ey):.3f}, "
      f"1st/99th percentile {np.nanpercentile(ey, 1):.3f}/{np.nanpercentile(ey, 99):.3f}")
print(f"spike density (Hz): {np.nanmin(sd_all):.1f} to {np.nanmax(sd_all):.1f}")
valid2 = ~np.isnan(D["eyelid"][300]).all(0)
print("set 2 samples present:", t[valid2][0], "to", t[valid2][-1], "ms (the rest NaN padding)")
sp = np.concatenate([np.concatenate(s) for s in D["spikes"][:270]])
print(f"spike times set 1: {sp.min():.1f} to {sp.max():.1f} ms")
print("mean baseline rate (Hz), set 1: %.1f +/- %.1f, range %.1f to %.1f" % (
    d1.rate_base.mean(), d1.rate_base.std(), d1.rate_base.min(), d1.rate_base.max()))
# eyelid baseline: trial exclusions of the Methods (eye at least half open; Grubbs on baseline SD)
half = sum(int((np.nanmean(e[:, t < 0], 1) >= 0.5).sum()) for e in D["eyelid"])
grubbs = sum(int((~ipn.grubbs_keep(np.nanstd(e[:, t < 0], 1))).sum()) for e in D["eyelid"])
print("trials with baseline eyelid at or past half closed:", half,
      "| trials a Grubbs test on baseline SD would still remove:", grubbs, "of", df.n_trials.sum())
```

The deposited spike density traces (`ephys`) are compared with the deposited spike times convolved here with
Gaussians of several widths, the stored mean traces with the trial means, and the deposited per-trial
`facilitation` and `crAmplitude` with the definitions they could follow.

```python
from scipy.ndimage import gaussian_filter1d
rs = {}
for sdv in (3, 4, 5, 6):
    vals = []
    for i in range(0, 372, 37):
        for tr in range(3):
            s = D["spikes"][i][tr]
            b = np.zeros(len(t))
            k = np.round(s).astype(int) - t[0]
            np.add.at(b, k[(k >= 0) & (k < len(t))], 1000.0)
            g = gaussian_filter1d(b, sdv, mode="constant")
            ok = ~np.isnan(D["sdf"][i][tr])
            vals.append(np.mean(np.abs(g[ok] - D["sdf"][i][tr][ok])))
    rs[sdv] = np.mean(vals)
print("mean abs difference (Hz) between deposited SDF and spikes convolved with a Gaussian of SD:",
      {k: round(v, 2) for k, v in rs.items()})
dm = max(np.nanmax(np.abs(np.nanmean(D["sdf"][i], 0) - D["sdf_mean"][i])) for i in range(372))
de = max(np.nanmax(np.abs(np.nanmean(D["eyelid"][i], 0) - D["eyelid_mean"][i])) for i in range(372))
print(f"largest difference of stored mean traces from trial means: spikes {dm:.3g} Hz, eyelid {de:.3g}")
dmed = [np.nanmedian(np.abs(np.nanmean(D["sdf"][i], 0) - D["sdf_mean"][i])) for i in range(372)]
print("stored ephysMean against the trial mean of ephys: median abs difference per recording, median over recordings"
      f" {np.median(dmed):.3g} Hz, recordings with a median difference above 1 Hz: {int(np.sum(np.array(dmed) > 1))}")
bad = [i for i in range(372) if dmed[i] > 1]
print("those recordings (cell number, set, trials deposited, median and largest abs difference in Hz):")
print([(i + 1, int(D["dset"][i]), D["sdf"][i].shape[0], round(float(dmed[i]), 1),
        round(float(np.nanmax(np.abs(np.nanmean(D["sdf"][i], 0) - D["sdf_mean"][i]))), 1)) for i in bad])
# duplicates: identical mean traces or identical first trials across recordings
keys = {}
dups = []
for i in range(372):
    k = (D["sdf"][i].shape, np.round(np.nansum(D["sdf"][i]), 3), np.round(np.nansum(D["eyelid"][i]), 5))
    if k in keys:
        dups.append((keys[k] + 1, i + 1))
    keys[k] = i
print("recordings deposited twice (same trial count and same sums of spike and eyelid traces):", dups)
```

```python
# The data set's per-trial crAmplitude ("eyelid amplitude at US onset ... from baseline (all datapoints where
# timeIndex < 0)") against the eyelid traces, at the stated US onset and at other times.
def crmatch(off):
    out = {1: [], 2: []}
    for i in range(372):
        e = D["eyelid"][i]
        v = (e[:, t == D["us"][i] + off][:, 0] - np.nanmean(e[:, t < 0], 1)) * 100
        out[int(D["dset"][i])].append(np.max(np.abs(v - D["cramp"][i])))
    return {k: (round(float(np.median(v)), 3), round(float(np.max(v)), 3)) for k, v in out.items()}
for off in (0, -12, -13, -14):
    print(f"crAmplitude vs eyelid at US onset {off:+d} ms: per-recording max abs difference (median, max) by set", crmatch(off))
# The data set's per-trial facilitation against readings (a), (a+) and (b) of the modulation measure on single trials
def facmatch(win_of):
    out = {1: [], 2: []}
    for i in range(372):
        ds, us = int(D["dset"][i]), int(D["us"][i])
        mod = ipn.modulation(D["sdf"][i], t, ipn.baseline_window(ds), win_of(ds, us))
        out[ds].append([np.max(np.abs(mod[f"fac_{rd}"] - D["fac"][i])) for rd in ("a", "ap", "b")])
    return {k: np.round(np.median(np.array(v), 0), 3).tolist() for k, v in out.items()}
print("facilitation, median per-recording max abs difference (Hz) for readings a, a+, b;",
      "window = last 200 ms:", facmatch(lambda ds, us: (us - 200, us)))
print("   window = CS onset to US onset:", facmatch(lambda ds, us: (0, us)))
```

Clock check. The traces are cut around CS onset, so there is no separate trial table; the check is that the
responses to the CS and the US occur after the stated times, by set: the eyelid's unconditioned blink and the
spike response to the US, and the same quantities in time-shifted windows as a contrast.

```python
for ds, sub in ((1, d1), (2, d2)):
    us = int(sub.us.iloc[0])
    ems = np.array([np.nanmean(D["eyelid"][i] - np.nanmean(D["eyelid"][i][:, t < 0], 1, keepdims=True), 0)
                    for i in sub.index])
    g = np.nanmean(ems, 0)
    v = np.diff(g)
    w = ipn.idx(t[1:], us - 100, us + 100)
    print(f"set {ds}: grand-mean eyelid velocity is highest {t[1:][w][np.nanargmax(v[w])] - us:+d} ms from US onset"
          f" (searched -100 to +100 ms)")
    z = np.array([(np.nanmean(D["sdf"][i], 0) - df.mu[i]) / df.sd[i] for i in sub.index])
    gz = np.nanmean(np.abs(z), 0)
    for a, b in ((-30, 0), (0, 15), (15, 30), (30, 60)):
        print(f"   mean |z| of spike traces {a:+d} to {b:+d} ms around US: {np.mean(gz[ipn.idx(t, us + a, us + b)]):.2f};"
              f" around CS: {np.mean(gz[ipn.idx(t, a, b)]):.2f}")
print("first US peak latency (ms after US), cells with a peak, set 1: median",
      np.median(d1.peak1_t[d1.peak1]), "| set 2:", np.median(d2.peak1_t[d2.peak1]))
```

## source-data-1-link

> The cells of the paper's Source data 1 are the cells of CRCNS cb-1 in the same order, so that the recording depth Source data 1 gives can be attached to the data set's cells.

Source: Additional files, Source data 1. Data: with_inputs. A link claim (Author, step 4).

Criterion (claims file): Added by the author of attempt brinke-2017-01 (Author, step 4). Cell i of the data set is matched to cell i of Source data 1, the order both use. The matched cells are compared on what both hold, the number of trials, the per-trial eyelid closure at US onset ('CR @ US ms' against the data set's crAmplitude) and the per-trial baseline rate ('Baseline Hz' against the spike count in the pre-CS baseline), and the same comparisons are shown with the cells shifted by one as a reference. Reproduced when at least 99% of the 372 cells agree on all three and the shifted match does not; the number of cells with a depth is reported.

```python
SD1, sd1_md5 = ipn.load_source_data1()
print("Source data 1:", ipn.SOURCE_DATA1)
print("  MD5", sd1_md5, "pinned", ipn.SOURCE_DATA1_MD5, "match", sd1_md5 == ipn.SOURCE_DATA1_MD5)
print("  variables:", len(SD1), "| cells per variable:", sorted({len(v) for v in SD1.values()}))


def base_hz(i):
    """Per-trial spike count in the cell's baseline window (ipn.baseline_window), in Hz."""
    lo, hi = ipn.baseline_window(int(df.dset[i]))
    return np.array([np.sum((s >= lo) & (s < hi)) for s in D["spikes"][i]]) / ((hi - lo) / 1000)


def link_checks(shift=0):
    """Match data set cell i to Source data 1 cell i + shift; per-cell agreement on what both hold."""
    out = []
    for i in range(len(df)):
        j = i + shift
        if not 0 <= j < len(df):
            continue
        n = D["sdf"][i].shape[0]
        same_n = len(SD1["# Valid Trials"][j]) == n
        cr = same_n and np.allclose(SD1["CR @ US ms"][j], D["cramp"][i], atol=1e-9, equal_nan=True)
        bh = same_n and np.allclose(SD1["Baseline Hz"][j], base_hz(i), atol=1e-6)
        out.append({"cell": i, "dset": int(df.dset[i]), "trials": same_n, "CR @ US = crAmplitude": cr, "Baseline Hz = spike count": bh})
    return pd.DataFrame(out)


lk = link_checks()
ids = np.array([np.unique(SD1["Cell"][i]).tolist() == [i + 1] for i in range(len(df))])
print("Source data 1 'Cell' is the cell's position (1-372), constant over its trials:", int(ids.sum()), "of", len(df))
print("cells matched by order that agree (of 270 first-dataset, 102 second-dataset cells):")
print(lk.groupby("dset")[["trials", "CR @ US = crAmplitude", "Baseline Hz = spike count"]].sum().to_string())
for sh in (-1, 1):
    l2 = link_checks(sh)
    print(f"  reference, cells shifted by {sh:+d}: same trial count {int(l2.trials.sum())}, same CR @ US {int(l2['CR @ US = crAmplitude'].sum())},"
          f" same baseline Hz {int(l2['Baseline Hz = spike count'].sum())} of {len(l2)}")
dep = [np.unique(SD1["Depth"][i][~np.isnan(SD1["Depth"][i])]) for i in range(len(df))]
print("cells whose Depth is one value over all trials:", sum(len(v) == 1 for v in dep), "| all NaN:", sum(len(v) == 0 for v in dep))
df["depth"] = [float(v[0]) if len(v) == 1 else np.nan for v in dep]
d1, d2 = df[df.dset == 1], df[df.dset == 2]
for ds, sub in ((1, d1), (2, d2)):
    v = sub.depth.dropna()
    print(f"set {ds}: depth given for {len(v)} of {len(sub)} cells" + (f", {v.min():.0f} to {v.max():.0f} um, mean {v.mean():.0f} +/- {v.std():.0f}" if len(v) else ""))
print("set 1 depth by mouse (mean, n):", {int(m): (round(g.depth.mean()), len(g)) for m, g in d1.groupby("mouse")})
```

**Result: reproduced.** All 372 cells matched by order agree on the number of trials, the per-trial eyelid closure at US onset and the per-trial baseline rate (270 of 270 first-dataset and 102 of 102 second-dataset cells), and Source data 1's `Cell` is each cell's position. Shifted by one cell, 9 of 371 agree on the number of trials and none on the other two. Source data 1 gives one depth per cell for the 270 first-dataset cells (1500 to 3000 um, the range the Results give) and none for the 102 second-dataset cells, whose `Depth` is NaN.

### Source data 1's per-cell values against this attempt's

The link above shows that the deposited trials of every cell are the trials Source data 1 lists, as `# Valid
Trials` (one entry per trial) has the deposited trial count in all 372 cells. The Methods' trial exclusions
(Grubbs test on baseline SDs, eye at least half open) are therefore taken as already applied, and are not applied
again here. Source data 1 also holds the authors' own per-trial CR flags and per-cell modulation and US-response
values. They are compared here with this attempt's measures, to show where the selections differ. None of these

```python
# Trial-by-trial correlations (Methods, Statistics), used by several claims below.
XREAD = ("a", "b", "c")       # per-trial firing rate readings (claims file, opening comment)
YREAD = ("data", "us", "max")  # per-trial CR amplitude: data set's crAmplitude, computed at US onset, CS-US max


def trial_test(cells, xr, yr, sign, rng, boot=True, win2=WIN2):
    """Per-cell correlations of the per-trial rate (reading xr) with CR amplitude (reading yr); counts of
    significant cells in direction `sign` and the opposite one, and the bootstrap p under both null readings."""
    pairs = [(ipn.trial_rates(D, i, None if D["dset"][i] == 1 else win2)[xr], amp_trials(i, yr)) for i in cells]
    r, p, kinds = ipn.trial_corrs(pairs)
    sig = p < 0.05
    n_dir, n_opp = int(np.sum(sig & (np.sign(r) == sign))), int(np.sum(sig & (np.sign(r) == -sign)))
    rs, ps, _ = ipn.trial_corrs(pairs, "spearman")
    out = {"x": xr, "y": yr, "cells": len(cells), "sig": n_dir, "opposite": n_opp,
           "pearson": sum(k == "pearson" for k in kinds),
           "sig, Spearman only": int(np.sum((ps < 0.05) & (np.sign(rs) == sign)))}
    if boot:
        out["p_shuffle"] = ipn.bootstrap_count(pairs, kinds, n_dir, sign, rng, reading="shuffle")[0]
        out["p_random"] = ipn.bootstrap_count(pairs, kinds, n_dir, sign, rng, reading="random")[0]
    return out, r, p, pairs


def lmm_on(cells, sig_mask, pairs, slopes=True, swap=False):
    """Mixed model of CR amplitude (y) on the per-trial rate (x) over the significant cells."""
    xs, ys, gs = [], [], []
    for c, ok, (x, y) in zip(cells, sig_mask, pairs):
        if ok:
            xs += list(x); ys += list(y); gs += [c] * len(x)
    if len(set(gs)) < 2:
        return None
    if swap:
        xs, ys = ys, xs
    return {**ipn.mixedlm(np.array(ys), np.array(xs), np.array(gs), slopes=slopes), "trials": len(xs)}


def lags_for(cells, sign):
    """Lag of the best cross-correlation per cell, for the two estimators (ipn.xcorr_lag: 'xcov' adopted, 'recut'
    variant) and segment readings (a) and (b) of the claims file; Wilcoxon signed rank against 0. Also counts the
    cells whose best lag is at the +/-100 ms edge and those whose curve is flat at its best (the best r within
    0.01 of the 21st best, so that 20 or more lags tie with it)."""
    out = {}
    for est in ("xcov", "recut"):
        for seg_name in ("a, last 200 ms", "b, whole CS-US"):
            lags, flat = [], 0
            for i in cells:
                us = int(D["us"][i])
                seg = (us - 200, us) if seg_name.startswith("a") else (0, us)
                m = np.nanmean(D["sdf"][i], 0)
                e = np.nanmean(D["eyelid"][i], 0)
                lag, r = ipn.xcorr_lag(m, e, t, seg, 100, sign, est)
                lags.append(lag)
                rr = np.sort(sign * r[~np.isnan(r)])
                flat += int(rr[-1] - rr[-21] < 0.01)
            lags = np.array(lags)
            p = stats.wilcoxon(lags).pvalue if np.any(lags != 0) else 1.0
            out[f"{est}, {seg_name}"] = (lags.mean(), lags.std(ddof=1), p, len(lags), int(np.sum(lags > 0)), int(np.sum(lags < 0)),
                                         float(np.median(lags)), int(np.sum(np.abs(lags) == 100)), flat)
    return out


def print_lags(rd, res):
    for k, (mu_, sd_, p, n, npos, nneg, med, edge, flat) in res.items():
        print(f"class {rd}, {k}: lag {mu_:.1f} +/- {sd_:.1f} ms, median {med:.0f} (n = {n}; {npos} positive, {nneg} negative,"
              f" {edge} at the +/-100 ms edge, {flat} flat at the best lag), Wilcoxon p = {p:.3g}")
```

## fig2-s1-correlation-matrix-offsets

> In spike-eyelid correlation matrices the strongest correlations of facilitation cells lie above the diagonal (spikes leading), at optimal offsets of 80 ms (position) and 20 ms (velocity) in the first dataset and 20 and 10 ms in the second, while suppression cells show offsets of 0 ms for position and -30 and -50 ms for velocity.

Source: Results, IpN facilitation but not suppression can drive conditioned eyelid behavior, paragraph 1; Figure 2C,G,K,O; Figure 2-figure supplement 1A-D. Data: dataset.

Criterion (claims file): Tolerance is one matrix step (10 ms), the resolution of the offsets. Per cell, across trials, the correlation of mean eyelid position (or velocity) in each bin with mean spike rate in each bin; r values of the sign opposite to the one of interest set to zero, then averaged over cells (Statistics); the optimal offset is the diagonal offset with the highest mean r in the last 200 ms of the CS-US interval (Figure 2-figure supplement 1 legend). The Results give 20 ms windows at 10 ms steps, the Statistics 20 ms bins (a 100 by 100 matrix); both readings are run. Reproduced when all eight offsets are within tolerance under one reading; partially when only the sign pattern holds (facilitation offsets positive, suppression offsets zero or negative). The paper gives no count of offsets that must match. The quote gives the first-dataset facilitation offsets; the others are in panels B to D of the same legend.

```python
def offsets(cells, sign, step, velocity):
    mats = []
    for i in cells:
        us = int(D["us"][i])
        lo = -200 if D["dset"][i] == 2 else -300
        c, R = ipn.corr_matrix(D["sdf"][i], D["eyelid"][i], t, lo, us + 200, 20, step, velocity)
        mats.append(R)
    us = int(D["us"][cells[0]])
    return ipn.optimal_offset(mats, c, us, sign)


expected = {("fac", 1, "pos"): 80, ("fac", 1, "vel"): 20, ("fac", 2, "pos"): 20, ("fac", 2, "vel"): 10,
            ("sup", 1, "pos"): 0, ("sup", 1, "vel"): -30, ("sup", 2, "pos"): 0, ("sup", 2, "vel"): -50}
rows = []
for rd in READINGS:
    c = cls(rd)
    for step in (20, 10):
        found = {}
        for (kind, ds, pv), exp in expected.items():
            cells = list(df.index[(c == kind) & df.behav & (df.dset == ds)])
            best, offs, vals = offsets(cells, +1 if kind == "fac" else -1, step, pv == "vel")
            found[(kind, ds, pv)] = best
        rows.append({"class": rd, "window step": step, **{f"{k[0]} set{k[1]} {k[2]}": v for k, v in found.items()},
                     "all within 10 ms": all(abs(found[k] - expected[k]) <= 10 for k in expected),
                     "sign pattern": all((found[k] > 0) if k[0] == "fac" else (found[k] <= 0) for k in expected)})
print("expected:", {f"{k[0]} set{k[1]} {k[2]}": v for k, v in expected.items()})
print(pd.DataFrame(rows).to_string(index=False))
```

```python
# CS transients: the four readings of the threshold (claims file), the two windows (Methods 50-125 ms,
# Results overview 50-130 ms), and where the rebound is looked for (within the window, or up to 50 ms after
# the trough, "+50"); see ipn.CS_WINDOWS.
CSREAD = [f"{a}/{b}" for a in ("mean", "drop") for b in ("trough", "mean")]


CSWIN = list(ipn.CS_WINDOWS)  # "125", "130", "125+50", "130+50": window end, and rebound within 50 ms after the trough


def cs_col(kind, rd, end="125"):
    return df[f"cs{kind}_{rd}_{end}"].astype(bool)


def fisher(a, b):
    """Two-sided Fisher exact test of the 2x2 table of boolean series a (rows) and b (columns)."""
    tab = [[int((a & b).sum()), int((a & ~b).sum())], [int((~a & b).sum()), int((~a & ~b).sum())]]
    return stats.fisher_exact(tab)[1], tab
```

**Result: reproduced.** US peak amplitude correlates negatively with % CRs (r = -0.381, p = 1e-10) and with CR amplitude at US onset (r = -0.271, p = 6e-06), n = 270. Both r values are weaker than the paper's -0.426 and -0.333 and just outside tolerance.

```python
def peak_trials(i, which):
    """Per-trial US response values: maximum (first or second peak) or minimum (pause) absolute rate in a 30 ms
    window centred on the cell's average peak or pause time (Methods, US-responses)."""
    us = int(D["us"][i])
    if which == "peak1":
        return ipn.trial_extreme(D["sdf"][i], t, us + df.peak1_t[i], "max")
    if which == "peak2":
        return ipn.trial_extreme(D["sdf"][i], t, us + df.peak2_t[i], "max")
    return ipn.trial_extreme(D["sdf"][i], t, us + df.pause_t[i], "min")


def us_trial_test(cells, which, yr, sign, rng):
    pairs = [(amp_trials(i, yr), peak_trials(i, which)) for i in cells]
    r, p, kinds = ipn.trial_corrs(pairs)
    sig = p < 0.05
    n_dir, n_opp = int(np.sum(sig & (np.sign(r) == sign))), int(np.sum(sig & (np.sign(r) == -sign)))
    pb_s = ipn.bootstrap_count(pairs, kinds, n_dir, sign, rng, reading="shuffle")[0]
    pb_r = ipn.bootstrap_count(pairs, kinds, n_dir, sign, rng, reading="random")[0]
    rs, ps, _ = ipn.trial_corrs(pairs, "spearman")
    return {"cells": len(cells), "sig": n_dir, "opposite": n_opp, "p_shuffle": pb_s, "p_random": pb_r,
            "sig, Spearman only": int(np.sum((ps < 0.05) & (np.sign(rs) == sign)))}, r, p, pairs
```

### The CS transient claims under one reading at a time

Each CS transient claim above is judged under any of the readings, as its criterion says. This table holds
each reading fixed (threshold reading and window) and says, for each claim, whether its criterion is met under
that reading with any class reading (and, in the second dataset, either window of the claims file). fig4b and
fig4c need the count and latency or the count of facilitation cells within tolerance and the Fisher test;
fig4s1 needs both counts and the Fisher test; fig4h needs no transient on days 1 to 3 and the Fisher test;
fig4g needs both correlations positive at p < 0.05 under a CR latency reading; fig5cd needs both tests
(either facilitation reading) and fig5hi the absence and the velocity test; cs-pause-cells-have-us-pause needs
the Fisher test, and its count of 6 +/- 1 is shown beside it. The "+50" windows are this attempt's variant,
not readings of the claims file.


```python
def met_by_reading(end, rd):
    sub1, sub2 = df.dset == 1, df.dset == 2
    cp, pk = cs_col("pause", rd, end), cs_col("peak", rd, end)
    lat = df[f"cspause_t_{rd}_{end}"]
    out = {}
    ok = False
    for crd in READINGS:
        p, tab = fisher(cp[sub1], (cls(crd) == "fac")[sub1])
        ok |= bool(6 <= cp[sub1].sum() <= 8 and abs(lat[sub1 & cp].mean() - 88) <= 8.8 and p < 0.05 and tab[0][0] / max(cp[sub1].sum(), 1) > tab[1][0] / (tab[1][0] + tab[1][1]))
    out["fig4b"] = ok
    ok = False
    for crd in READINGS:
        for w in WINS2:
            fac2 = (cls(crd, w) == "fac")[sub2]
            p, tab = fisher(cp[sub2], fac2)
            ok |= bool(36.9 <= cp[sub2].sum() <= 45.1 and 35.1 <= tab[0][0] <= 42.9 and p < 0.05)
    out["fig4c"] = ok
    ok = False
    for crl in ("crlat_a", "crlat_b", "crlat_c"):
        a, b = cp & df[crl].notna() & df.behav, cp & df[crl].notna() & df.behav & sub2
        if a.sum() > 2 and b.sum() > 2:
            r1, r2 = stats.pearsonr(lat[a], df[crl][a]), stats.pearsonr(lat[b], df[crl][b])
            ok |= bool(r1[0] > 0 and r1[1] < 0.05 and r2[0] > 0 and r2[1] < 0.05)
    out["fig4g"] = ok
    ok = False
    for crd in READINGS:
        p, tab = fisher(pk[sub1], (cls(crd) == "sup")[sub1])
        ok |= bool(3 <= pk[sub1].sum() <= 5 and pk[sub2].sum() == 0 and p < 0.05 and tab[0][0] > 0)
    out["fig4s1"] = ok
    tr, early = (cp | pk)[sub1], d1.day <= 3
    p, tab = fisher(tr, early)
    out["fig4h"] = bool(tab[0][0] == 0 and p < 0.05)
    for ds, key in ((1, "fig5cd"), (2, "fig5hi")):
        ok = False
        for crd in READINGS:
            for w in (WINS2 if ds == 2 else (WIN2,)):
                f = df[(df.dset == ds) & (cls(crd, w) == "fac")]
                a, b = f[cp[f.index]], f[~cp[f.index]]
                if len(a) == 0:
                    continue
                pv = ipn.mwu(a.vel_max, b.vel_max)
                vel = pv < 0.05 and a.vel_max.mean() > b.vel_max.mean()
                if ds == 1:
                    for col_a, col_b in ((a.max_csus, b.max_csus), (mag(crd, "fac")[a.index], mag(crd, "fac")[b.index])):
                        ok |= bool(vel and ipn.mwu(col_a, col_b) < 0.05 and col_a.mean() > col_b.mean())
                else:
                    for col_a, col_b in ((a.max_csus, b.max_csus), (mag(crd, "fac", w)[a.index], mag(crd, "fac", w)[b.index])):
                        ok |= bool(vel and ipn.mwu(col_a, col_b) >= 0.05)
        out[key] = ok
    p, tab = fisher(cp[sub1], d1.pause)
    out["cs-us pause (test)"] = bool(p < 0.05 and tab[0][0] / max(cp[sub1].sum(), 1) > tab[1][0] / (tab[1][0] + tab[1][1]))
    out["cs-us pause count"] = f"{tab[0][0]}/{int(cp[sub1].sum())}"
    return out


rows = [{"window": end, "threshold": rd, **met_by_reading(end, rd)} for end in CSWIN for rd in CSREAD]
tab = pd.DataFrame(rows)
crit = ["fig4b", "fig4c", "fig4g", "fig4s1", "fig4h", "fig5cd", "fig5hi", "cs-us pause (test)"]
tab["criteria met (of 8)"] = tab[crit].sum(axis=1)
print(tab.to_string(index=False))
print("claims whose criterion is met under at least one reading:", {k: bool(tab[k].any()) for k in crit})
```

**Result: reproduced.** Cells with only a US pause lie shallowest (1920 +/- 290 um, n = 23), cells with both in between (2061 +/- 323 um, n = 76) and cells with only a US peak deepest (2186 +/- 364 um, n = 138); all three means are within tolerance and the Kruskal-Wallis test gives p = 0.00067 (paper 0.0025). The order and significance hold with depth taken relative to each mouse's mean (p = 0.0057) and with turnaround spans of 20 and 45 ms (p = 0.00067 and 0.00052).
