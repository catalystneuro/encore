Questions asked and answered (`questions/000009.yaml`; the answers of 2026-10-20 are written
into the claims file, and this notebook follows them):

7. Asked after the first verification, not yet answered: should the whole-cell photostimulus onset

The deposited mean waveforms are short: 29 samples at 25 kHz (1.16 ms) in thalamus and SNr, and
the trough sits at sample 15 to 19, so a trough-to-peak interval longer than 0.36 to 0.52 ms cannot
be measured in many units. The peak after the trough is the last sample in 78% of ALM units
labelled 'Wide width spiking'. Widths computed from these waveforms are therefore lower bounds for
wide units, and the classes the Methods define by width (> 0.6 ms in ALM, > 0.5 ms in thalamus)
cannot be recovered from them. The files' own `cell_type` labels ('Wide width spiking', 'Narrow
width spiking', 'Cannot be determined') separate the units cleanly by computed width (narrow up
to 0.32 ms in ALM, wide from 0.44 ms) and are shown below as the second reading (P2) of every unit
class; reading P1 applies the Methods' thresholds to the computed widths.

Coordinates: `unit_x` is medio-lateral (positive), `unit_y` dorso-ventral (negative, mm), `unit_z`
antero-posterior; thalamic units lie between AP -1.2 and -2.1 mm, inside the "bregma -1 to -2 mm"
of the Methods, and ALM units at AP 2.5 to 3.0. Two inconsistencies: every electrode group, also in
thalamus and SNr, carries the ALM coordinates (AP 2.5, ML -1.5, DV 0); and the SNr units of mouse
anm00261336 are at AP -1.56 mm, where SNr is not (the other two SNr mice are at AP -3.6 to -4.2).
Photostimulus sites give ML negative for the left hemisphere, units give it positive. The thalamic
fibre site is the same nominal coordinate (AP -1.5, ML 0.9, DV -4.1) in all 29 sessions.

Spikes are stored only in part of each trial and for part of each session. Within a trial, spikes
run from the trial start (0.57 s before pole onset) to about 4.8 s after pole onset: the summed rate
5.5 to 8 s after pole onset is 1.4% of that 2 to 4.5 s after it, although trials last about 9 s.
The end is fixed to pole onset, not to the go cue (next table), so in trials with a long delay the
response epoch, and in a few the delay epoch, runs past the stored spikes. Each trial's stored span
is taken to end at the last spike of any unit of the session in it, and a window that ends later is
counted as missing for that trial. Across a session, most units hold spikes over only some of
its trials (median 70%), and the files give no observation interval per unit. A unit's trials are
therefore taken to run from the trial holding its first spike to the trial holding its last; trials
outside count as missing, not as zero spikes. This is a standard choice recorded as a deviation.

Trial types are 'lick left', 'lick right' and 'non-performing'; contra trials are taken as 'lick
right' (left hemisphere, Fig. 1a). 5375 extracellular rows have no outcome ('N/A', type
'non-performing'); 5369 of them have no pole times and are dropped, and the other 6 enter no
analysis. In the extracellular files the delay (go cue minus 1.3 s after pole onset) is 1.3 s in
81% of trials and never 1.2 s; the rest are longer (up to 4.4 s at the 99th percentile).
`pole_out_time` is 1.3 s after `pole_in_time`. Trials last about 9 s. In the whole-cell files `pole_out_time` is 1.4 s after
`pole_in_time` and the go cue 2.6 s after it. One whole-cell trial table is not in time order and
has a row whose start time (774 s) does not contain its pole onset (12 s). The whole-cell trial
tables leave every photostimulus column empty; `stim_present` marks 30 to 50% of trials, the
"25-50%" of the Methods for Fig. 3g and 4, and is read as the photostimulus flag (its effect is
shown below).

The end of the stored spikes is fixed to pole onset: its median is 4.81 to 4.84 s after pole onset
whatever the delay, so it falls 2.2 s after the go cue in trials with a 1.3 s delay and before the
go cue in trials with a delay above 3 s (table). Before this correction, such windows were counted
with the spikes they held, which biased their counts low. Every correct or incorrect trial has a
1.3 s delay; the longer delays occur only in lick-early and no-response trials, which no analysis
uses. The correction therefore touches only the kept trials whose stored spikes end early: 221 of
8991 in ALM, 173 of 14997 in thalamus and 19 of 3573 in SNr for R1.

## task-performance

> trials and 1.1% no-response trials.

Readings of the criterion: (a) all trials of all sessions pooled, (b) mean over sessions, (c)
control trials only, pooled and per session. Percent correct excludes lick-early and no-response
trials; those two are fractions of all task trials. Sessions are every behavioural recording in
the dandiset (extracellular sessions and whole-cell recordings), shown also by kind.

```python
prow = []
for s in sessions:
    t = s["trials"]
    prow.append({"kind": "extracellular", "subject": s["meta"]["subject"], "resp": t.response.values, "ctrl": (t.stim == "none").values})
for c in cells:
    t = c["trials"]
    prow.append({"kind": "whole-cell", "subject": c["meta"]["subject"], "resp": t.response.values, "ctrl": ~t.pi.values})


def perf(resp):
    n = len(resp)
    cor, inc = (resp == "correct").sum(), (resp == "incorrect").sum()
    return {"n": n, "correct": 100 * cor / (cor + inc), "early": 100 * (resp == "early lick").sum() / n,
            "noresp": 100 * (resp == "no response").sum() / n}


out = []
for kinds in (("extracellular", "whole-cell"), ("extracellular",), ("whole-cell",)):
    sel = [r for r in prow if r["kind"] in kinds]
    for reading, use_ctrl in (("a/b all trials", False), ("c control trials", True)):
        per = pd.DataFrame([perf(r["resp"][r["ctrl"]] if use_ctrl else r["resp"]) for r in sel])
        pooled = perf(np.concatenate([r["resp"][r["ctrl"]] if use_ctrl else r["resp"] for r in sel]))
        out.append({"files": "+".join(kinds), "trials": reading, "how": "pooled", "sessions": len(sel), "trials_n": pooled["n"],
                    "correct_%": pooled["correct"], "lick_early_%": pooled["early"], "no_response_%": pooled["noresp"]})
        out.append({"files": "+".join(kinds), "trials": reading, "how": "mean over sessions", "sessions": len(sel), "trials_n": int(per.n.sum()),
                    "correct_%": per.correct.mean(), "lick_early_%": per.early.mean(), "no_response_%": per.noresp.mean()})
perf_tab = pd.DataFrame(out)
perf_tab["within_tol"] = ((perf_tab["correct_%"] - 77.6).abs() <= 7.76) & ((perf_tab["lick_early_%"] - 10.3).abs() <= 1.03) & \
    ((perf_tab["no_response_%"] - 1.1).abs() <= 0.11)
print(perf_tab.round(2).to_string(index=False))
print("Paper: 77.6 +/- 7.76 % correct, 10.3 +/- 1.03 % lick early, 1.1 +/- 0.11 % no response")
by_mouse = pd.DataFrame([{"subject": r["subject"], **perf(r["resp"])} for r in prow]).groupby("subject")[["correct", "early", "noresp"]].mean()
print("\nPer mouse (mean over its sessions, all trials): correct", by_mouse.correct.round(1).min(), "to", by_mouse.correct.round(1).max(),
      "| lick early", by_mouse.early.round(1).min(), "to", by_mouse.early.round(1).max(), "| no response", by_mouse.noresp.round(2).min(), "to", by_mouse.noresp.round(2).max())
```

## fig1c-alm-photoinhibition-bias

> Unilateral photoinhibition of ALM during the delay epoch biased responses to the ipsilateral
> side, lowering performance on contra trials and raising it on ipsi trials.

The dandiset has no behaviour-only sessions; the test uses the behavioural trials of the 66
recording sessions in PV-Cre x Ai32 mice with left-ALM photoinhibition in the delay epoch (ALM, thalamic
and SNr recordings, 13 mice). Main reading: photoinhibition trials at 1.5 mW (the Methods' power);
variant: all delay-epoch powers (1.5 and 0.75/0.79 mW), outside the condition by convention P
(answer of 2026-10-20). Per session, fraction correct with
lick-early and no-response trials excluded; two-sided paired t-test over sessions.

```python
def fig1c(stims):
    rows = []
    for s in sessions:
        m = s["meta"]
        if m["genotype"] != "Ai32 x PV-Cre" or m["ogen_site"] != "ALM":
            continue
        t = s["trials"]
        k = t.response.isin(loop.KEEP).values
        r = {"subject": m["subject"], "region": m["region"]}
        for side, con in (("contra", True), ("ipsi", False)):
            for cond, msk in (("ctrl", (t.stim == "none").values), ("pi", t.stim.isin(stims).values)):
                sel = k & msk & (t.contra.values == con)
                r[f"{side}_{cond}"] = (t.response.values[sel] == "correct").mean() if sel.any() else np.nan
                r[f"n_{side}_{cond}"] = int(sel.sum())
        rows.append(r)
    return pd.DataFrame(rows)


for name, stims in (("1.5 mW", ["ALM:delay:1.5"]), ("all delay powers", ["ALM:delay:1.5", "ALM:delay:0.79", "ALM:delay:0.75"])):
    d = fig1c(stims)
    full = d.dropna(subset=["contra_ctrl", "contra_pi", "ipsi_ctrl", "ipsi_pi"])
    print(f"--- {name}: {len(full)} of {len(d)} sessions with all four cells, {full.subject.nunique()} mice;"
          f" photoinhibition trials per session median {int((full.n_contra_pi + full.n_ipsi_pi).median())}")
    for side in ("contra", "ipsi"):
        tt = stats.ttest_rel(full[f"{side}_pi"], full[f"{side}_ctrl"])
        print(f"{side}: control {full[side + '_ctrl'].mean():.3f}, photoinhibition {full[side + '_pi'].mean():.3f}, "
              f"change {(full[side + '_pi'] - full[side + '_ctrl']).mean():+.3f} (sessions with an increase "
              f"{int((full[side + '_pi'] > full[side + '_ctrl']).sum())}/{len(full)}), paired t = {tt.statistic:.2f}, p = {tt.pvalue:.2g}")
    if name == "1.5 mW":
        print("Per mouse (means over sessions):")
        print(full.groupby("subject")[["contra_ctrl", "contra_pi", "ipsi_ctrl", "ipsi_pi"]].mean().round(3).assign(
            sessions=full.groupby("subject").size()).to_string())
        print("By recorded region:")
        print(full.groupby("region")[["contra_ctrl", "contra_pi", "ipsi_ctrl", "ipsi_pi"]].mean().round(3).to_string())
```

increases) and raises ipsi performance from 0.673 to 0.744 (t = 3.99, p = 0.00017; 46 of 66

## fig6c-unmodulated-less-selective

> The few VM/VAL neurons not changed by ALM photoinhibition were significantly less selective in

(ii) change on selectivity. P1, P2 and P3.

```python
rows = []
for p in ("P1", "P2", "P3"):
    for tname in T_READ:
        for ev in (True, False):
            eff = F6[(p, tname, ev)]
            rc, ri = [], []
            for idx in eff.index:
                r = units.loc[idx]
                s = sessions[r.session]
                t = s["trials"]
                m = (t.response == "correct").values & (t.stim == "none").values
                c = s["counts"]["delay"][r.unit] / t.delay_s.values
                rc.append(np.nanmean(c[m & t.contra.values]))
                ri.append(np.nanmean(c[m & ~t.contra.values]))
            rc, ri = np.array(rc), np.array(ri)
            pref = SEL["control/R1/Student/c1"].loc[eff.index, "contra_pref"].values
            sel = {"a": np.abs(rc - ri), "b": np.where(pref, rc - ri, ri - rc)}
            for sname, sv in sel.items():
                tt = stats.ttest_ind(sv[eff.sig.values], sv[~eff.sig.values], equal_var=ev)
                for ax in ("i", "ii"):
                    x, y = (eff.delta.values, sv) if ax == "i" else (sv, eff.delta.values)
                    lr = stats.linregress(x, y)
                    rows.append({"units": p, "trials": tname, "test": "Student" if ev else "Welch", "selectivity": sname, "axes": ax,
                                 "n_sig": int(eff.sig.sum()), "n_not": int((~eff.sig).sum()),
                                 "sel_sig": round(sv[eff.sig.values].mean(), 2), "sel_not": round(sv[~eff.sig.values].mean(), 2),
                                 "t_p": tt.pvalue, "slope": round(lr.slope, 3), "r": round(lr.rvalue, 3)})
f6u = pd.DataFrame(rows)
f6u["group_ok"] = (f6u.sel_not < f6u.sel_sig) & (f6u.t_p < 0.05)
f6u["within_tol"] = f6u.group_ok & ((f6u.slope + 0.73).abs() <= 0.073) & ((f6u.r + 0.33).abs() <= 0.033)
f6u["partial"] = f6u.group_ok & (f6u.r < 0)
print(f6u.round(4).to_string(index=False))
print("Paper: P = 0.002; slope -0.73, r -0.33")
print("Readings reproduced:", int(f6u.within_tol.sum()), "| partial:", int(f6u.partial.sum()), "| group difference holds:",
      int(f6u.group_ok.sum()), "of", len(f6u))
print(f6u.groupby("units").agg(readings=("r", "size"), group_ok=("group_ok", "sum"), partial=("partial", "sum"), within_tol=("within_tol", "sum"),
                               r_min=("r", "min"), r_max=("r", "max")).to_string())
```

**How the analysis differs from the paper's.** Unit classes are run under the readings of convention C:
widths computed on the deposited waveforms (P1, the Methods' rule), which are too short for the
width thresholds, and the files' labels (P2), which match the paper's class sizes; for the thalamic
claims of Fig. 2 and 6, all units (P3) as well. Each unit's trials are taken as those
from the trial holding its first spike to the trial holding its last, as the files give no
observation interval per unit, and a window that ends after a trial's stored spikes is counted as
missing for that trial. Behaviour is measured in the recording sessions, not in behaviour
sessions. The VM/VAL centre is the claims file's proxy (reading L2). Distances from the thalamic
fibre use the files' single nominal site. The whole-cell claims are judged under the two onsets the
claims file allows (E and M) and under the onset of the deposited laser trace (L), which the claims
file did not know of (question 7); the outcomes are the same under M and L, nothing holds under E,
and the onset latency of 7.9 ms cannot be judged under E or M, the readings the claims file lists;
under L, shown pending question 7, it is 9.60 ms by V2 and 0.40 ms by V1. The (0.5, 0.5) galvo site is
taken as vM1 only to show what the data give, compared with ALM at 1.5 mW.

**What was found about the dataset.**

- The dandiset holds 966 of the paper's 1214 ALM units (47 of 57 sessions), 789 of 790 thalamic
  units and 181 of 227 SNr units (apparently only the putative GABAergic ones: 171 of the 181 by
  computed width, 146 by the files' label). Behaviour-only,
  muscimol, weak photoinhibition (Fig. 5), non-behaving and M1 whole-cell recordings and histology
  are absent.
- The mean waveforms are cut short (29 to 38 samples at 25 kHz), so trough-to-peak widths longer
  than about 0.5 ms cannot be measured; the `cell_type` labels carry the classification instead,
  without a stated rule.
- Spikes are stored only from the trial start to about 4.8 s after pole onset, although trials
  last about 9 s, so in trials with long delays the response epoch is not stored; most units hold
  spikes over only part of their session, with no per-unit observation interval.
- The whole-cell trial tables leave the photostimulus columns empty; the laser time is in an
  OptogeneticSeries described only as 'no description'. It starts between the two delay onsets
  the trial times allow, 0.10 to 0.17 s after E and 0.03 to 0.10 s before M (1.40 s after pole
  onset in one mouse, 1.47 s in the others). One whole-cell trial table has start and stop times shifted by one row. The Vm `conversion` (0.001 with unit 'mV') contradicts the values, which are in mV.
- Every electrode group, in thalamus and SNr too, carries the ALM coordinates; the SNr units of one
  mouse are at AP -1.56 mm; there is no VM/VAL location label; `unit_y` has no stated reference, so
  whether it is depth below the pia is not said. Session start times are the date of conversion
  (all on 2017-06-27), not of recording, so the missing ALM sessions cannot be identified by date.
- The Gad2 sessions (genotype given as `Ai35D x GAD2_Cre`, an Arch line, where the paper used
  Gad2-IRES-Cre with ChR2 in TRN) carry no photostimulus trial, though their 84 units match the
