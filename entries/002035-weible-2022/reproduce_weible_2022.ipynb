# Testing Weible and Wehr (2022) on dandiset 002035

**Paper.** Weible AP, Wehr M (2022). Amyloid pathology in the central auditory pathway of 5XFAD
mice appears first in auditory cortex. *J Alzheimers Dis* 89(4):1385-1402.
[doi:10.3233/JAD-220538](https://doi.org/10.3233/JAD-220538). The claims file was written from the
author manuscript in PubMed Central (NIHMS1886685, PMC10097438): main text, Methods, figure legends
and Tables 1 to 6. The paper has no supplement and links no code.

**Dandiset.** [DANDI:002035](https://dandiarchive.org/dandiset/002035/0.261022.1829), version
0.261022.1829 (doi:10.48324/dandi.002035/0.261022.1829). The paper (2022) predates the dandiset
(created 2026-10-19) and cites no version. 0.261022.1829 is the only published version, so it is
pinned (protocol, author step 1.3). The draft has the same number of assets and bytes.

**What the dandiset holds.** Raw widefield fluorescence images, one channel per OME-TIFF, of 62
hemizygous 5XFAD mice: 12.5X overviews, 100X fields, and 200X focal stacks, stained for Abeta42,
DAPI, or imaged in the red channel for autofluorescence. Its README says wild-type controls "were
not imaged", that the imaged region is given only "where recorded in the original folder name",
and that "if no images of a region are included for a mouse, no label was judged present at the
time of imaging". It holds no cell or plaque counts, no region or laminar boundaries, no atlas
alignment, and no stage position in the sampled image headers and sidecars.

**Why every claim is blocked.** Each claim of the paper is a count of Abeta42+ cells, plaques or
DAPI cells within a region (A1 and its eight laminar bins, subiculum, MGB and its subdivisions, IC
and its subdivisions, SOC, nLL, CN, an 800 um circle in PRNc), per mouse, with the region's area
(Methods, "Imaging, boundary determination, and quantification"). The notebook shows below that
the region of 93% of the images is not recorded, that no sampled image carries a position giving its
place in the section, and that the region labels that exist come from targeted
follow-up rounds in 24 of the 62 mice. Images cannot be assigned to regions, layers or subdivisions, so
no per-mouse regional value can be formed, by hand or by an automated count. The claims are
blocked as `metadata_insufficient`. The wild-type claim is blocked as `data_missing`.

No claim is tested, so no claim section follows. The data check streams a sample of images, drawn
with a fixed seed.

```python
import json
import time

import numpy as np
import pandas as pd

import amyloid as A

pd.set_option("display.width", 220)
pd.set_option("display.max_columns", 40)
pd.set_option("display.max_rows", 300)

started = time.time()
meta = A.version_metadata()
assets = A.manifest()
participants = A.table(assets, "participants.tsv")
scans = A.scans(assets).merge(assets[["path", "url", "size", "sha256"]], on="path", how="left")
print(f"dandiset {A.DANDISET} version {A.VERSION}: {len(assets)} assets, "
      f"{assets['size'].sum() / 1e9:.1f} GB; {len(scans)} images in the scans tables; "
      f"loaded in {time.time() - started:.0f} s")
print("Version status:", meta.get("version"), "| DOI:", meta.get("doi"))
```

## Data check

Each check of the metadata against the values is stated under its output.

```python
paper_n = {14: (2, 3)} | {a: (4, 4) for a in A.AGES[1:]}
by_age = pd.crosstab(participants.age, participants.sex)
by_age["paper F"] = [paper_n[a][0] for a in by_age.index]
by_age["paper M"] = [paper_n[a][1] for a in by_age.index]
print("Mice by age (days) and sex, against the paper (Methods, Mice):")
print(by_age.to_string())
print(f"Total {len(participants)} mice, {(participants.sex == 'F').sum()} F and {(participants.sex == 'M').sum()} M; "
      "the paper gives 61 5XFAD mice, '31 females and 30 males', and per age 4 F and 4 M (p21-p150), 2 F and 3 M (p14).")
print("Genotype values:", participants.genotype.value_counts().to_dict())

# Sex, age and genotype in the original folder names (scans.tsv, original_file) against participants.tsv.
folder = scans.groupby("participant_id").original_file.first().str.split("/").str[0]
parsed = folder.str.extract(r"^(?P<sexw>Male|Female) P(?P<agef>\d+) (?P<geno>\w+) (?P<label>\S+)$")
chk = participants.set_index("participant_id").join(parsed)
chk["sex_ok"] = chk.sexw.str[0] == chk.sex
chk["age_ok"] = chk.agef.astype(float) == chk.age
chk["id_ok"] = ("sub-" + chk.label) == chk.index
print("\nOriginal folder names: sex agrees", int(chk.sex_ok.sum()), "of", len(chk), "| age agrees", int(chk.age_ok.sum()),
      "| subject label agrees", int(chk.id_ok.sum()), "| genotype word:", chk.geno.value_counts().to_dict())

kind = np.select([assets.path.str.endswith(".ome.tif"), assets.path.str.endswith(".json"),
                  assets.path.str.endswith(".tsv")], ["image", "sidecar json", "tsv"], "other")
print("\nAssets by kind:", pd.Series(kind).value_counts().to_dict())
tifs = assets[assets.path.str.endswith(".ome.tif")]
print("Images in the manifest:", len(tifs), "| in the scans tables:", len(scans),
      "| in both:", len(set(tifs.path) & set(scans.path)))
print("Image files with identical content (SHA-256 shared by two or more):",
      int(tifs.sha256.duplicated(keep=False).sum()))
print("Original files listed twice:", int(scans.original_file.duplicated().sum()))
print(f"Image sizes: {tifs['size'].min() / 1e6:.2f} to {tifs['size'].max() / 1e6:.2f} MB")

print("\nImages by stain, magnification class and how the class was assigned:")
print(pd.crosstab([scans.stain, scans.magnification], [scans.magnification_confidence, scans.source_format]).to_string())
print("\nImages without an acquisition time:", int(scans.acq_time.isna().sum()))
print("\nImages per mouse: median", int(scans.groupby('participant_id').size().median()),
      "range", scans.groupby('participant_id').size().min(), "to", scans.groupby('participant_id').size().max())
per_age = scans.merge(participants[["participant_id", "age"]]).groupby("age").apply(
    lambda g: pd.Series({"mice": g.participant_id.nunique(), "images": len(g),
                         "200X Abeta42 per mouse": round(((g.magnification == "200X") & (g.stain == "Abeta42")).sum() / g.participant_id.nunique(), 1)}),
    include_groups=False)
print("\nImages by age:")
print(per_age.to_string())
```

The dandiset has 62 mice, all hemizygous 5XFAD, against the paper's 61. Ages p14, p21, p30, p60, p90
and p150 match the paper's numbers by sex; p45 has 3 females and 5 males against 4 and 4, and p120
has 5 females and 4 males, one mouse more than the paper. In all, 30 females and 32 males against
the paper's "31 females and 30 males" (which its own per-age numbers, 30 and 31, do not match
either). Sex, age and subject label in `participants.tsv` agree with the original folder names for
every mouse. No two image files are identical, and no original file is listed twice, so no
recording is deposited twice. There are no wild-type mice, as the README says.

```python
rng = np.random.default_rng(2035)
rows = []
for (mag, conf, src), g in scans[scans.stain == "DAPI"].groupby(["magnification", "magnification_confidence", "source_format"]):
    if mag == "12.5X":
        continue  # nuclei are not resolved at 5.16 um per pixel
    take = g.iloc[rng.choice(len(g), size=min(len(g), 6 if src == "czi" else 3), replace=False)]
    for _, r in take.iterrows():
        arr, ome_um = A.image(r.url)
        side = A.sidecar(assets, r.path)
        d, n = A.nucleus_diameter_um(arr, A.PIXEL_UM[mag])
        rows.append({"class": mag, "confidence": conf, "source": src, "dtype": str(arr.dtype), "shape": arr.shape,
                     "max": int(arr.max()), "saturated %": round(100 * (arr == arr.max()).mean(), 3),
                     "OME um/px": ome_um, "sidecar um/px": side["PixelSize"][0], "README um/px": A.PIXEL_UM[mag],
                     "nuclei": n, "median diameter um": round(d, 2)})
chk_img = pd.DataFrame(rows)
print(chk_img.to_string(index=False))
print("\nBy class:")
print(chk_img.groupby(["class", "source"]).agg(images=("nuclei", "size"), nuclei_median=("nuclei", "median"),
      diameter_median=("median diameter um", "median"), diameter_min=("median diameter um", "min"),
      diameter_max=("median diameter um", "max"), max_value=("max", "max")).to_string())
print("\nOME pixel size = sidecar = README for every sampled image:",
      bool(((chk_img["OME um/px"] == chk_img["sidecar um/px"]) & (chk_img["sidecar um/px"] == chk_img["README um/px"])).all()))

# Abeta42 images: bit depth by source, on a sample of each magnification class.
rows = []
for (mag, src), g in scans[scans.stain == "Abeta42"].groupby(["magnification", "source_format"]):
    take = g.iloc[rng.choice(len(g), size=min(len(g), 3), replace=False)]
    for _, r in take.iterrows():
        arr, _ = A.image(r.url)
        rows.append({"class": mag, "source": src, "dtype": str(arr.dtype), "min": int(arr.min()), "max": int(arr.max()),
                     "p50": float(np.median(arr))})
print("\nAbeta42 images, sample:")
print(pd.DataFrame(rows).to_string(index=False))
```

Checks of the image description, on a seeded sample. The pixel size in the OME header, in the
sidecar and in the README agree for each sampled image. The README assigns magnification from image
content; the DAPI nuclei measure about the same median diameter in micrometres in the 100X and the
200X classes (4.2 and 5.1 um from .czi), and a 100X field holds 3.6 times as many as a 200X field
(medians 2,536.5 and 711), near the 4 expected of a field of twice the width. This is consistent with the stated scales of the high- and low-confidence classes; it
is not a proof, and the 7,125 Abeta42 and red images assigned 200X by rule (no matching DAPI image)
were not checked. Images from the original .czi files reach 4095, the 12-bit maximum the README
gives; those from .jpg exports are 8-bit display images, as it says. The diameters here come from a
threshold of bright cores and are a check of scale only, not a cell measure.

```python
REGIONS = {"A1 (ACtx)": ["ACtx"], "subiculum (S)": ["S"], "MGB": ["MGB"], "IC": ["IC", "IC;VCN"],
           "SOC": ["SOC"], "nLL": [], "CN": ["CN", "IC;VCN"], "PRNc": ["PRNc"]}
s = scans.merge(participants[["participant_id", "age", "sex"]], on="participant_id")
recorded = s.region_abbr.notna()
print(f"Images with a recorded region: {recorded.sum()} of {len(s)} ({100 * recorded.mean():.1f}%)")
print(f"200X Abeta42 images with a recorded region: {(recorded & (s.magnification == '200X') & (s.stain == 'Abeta42')).sum()} "
      f"of {((s.magnification == '200X') & (s.stain == 'Abeta42')).sum()}")
print("Mice with no recorded region on any image:", int((~s.groupby('participant_id').region_abbr.apply(lambda x: x.notna().any())).sum()), "of 62")
print("\nOriginal subfolders of the images with a recorded region (rounds):")
print(s[recorded].groupby(["region_abbr", "original_subfolder"]).size().to_string())

cov = {}
for name, codes in REGIONS.items():
    m = s[s.region_abbr.isin(codes)]
    cov[name] = m.groupby("age").participant_id.nunique().reindex(A.AGES, fill_value=0)
cov = pd.DataFrame(cov).T
cov.columns = [f"p{a}" for a in A.AGES]
cov["mice"] = [s[s.region_abbr.isin(c)].participant_id.nunique() for c in REGIONS.values()]
cov["images"] = [int(s.region_abbr.isin(c).sum()) for c in REGIONS.values()]
mice_per_age = participants.groupby("age").size().reindex(A.AGES)
print("\nMice with any image labelled with the region, by age (mice in the dandiset per age:",
      mice_per_age.tolist(), "):")
print(cov.to_string())

# What else could place an image: positions in the OME header or the sidecar.
pos_keys = ("StagePosition", "PositionX", "PositionY", "PositionZ", "Plane ")
sample = s.iloc[rng.choice(len(s), size=20, replace=False)]
ome_pos, side_keys = 0, set()
for _, r in sample.iterrows():
    import remfile, tifffile
    with tifffile.TiffFile(remfile.File(r.url)) as t:
        ome_pos += any(k in (t.ome_metadata or "") for k in pos_keys)
    side_keys |= set(A.sidecar(assets, r.path))
print(f"\nSampled images whose OME header holds a stage or plane position: {ome_pos} of {len(sample)}")
print("Keys of the sampled sidecars:", sorted(side_keys))
print("Sidecar keys naming a position, section or coordinate:",
      sorted(k for k in side_keys if any(w in k.lower() for w in ("position", "coord", "section", "slice", "atlas", "roi"))))
print("Tables at the top level:", sorted(p for p in assets.path if "/" not in p))
```

What the claims need and what the dandiset holds. Only 864 of 12,394 images (7.0%), and 561 of
the 7,048 200X Abeta42 images, have a recorded region, all from named follow-up subfolders; 38 of
the 62 mice have none. The subiculum is labelled in 17 mice, PRNc in 10, SOC in 5, MGB and IC in 3,
A1 and CN in 2, and nLL in none; at no age is a region labelled in more than 5 of the 5 to 9 mice
of that age. No sampled image header or sidecar carries a stage position, a section
number or an atlas coordinate, and the dandiset has no table of counts, boundaries or areas. The
images of the main imaging pass cannot be assigned to a region, a cortical layer or a subdivision
from anything in the dandiset. Every claim needs, per mouse, the counts within such a region and
its area or its DAPI count, so none can be formed. The README's rule that a region without images
had no label "judged present at the time of imaging" cannot be applied either, since the region of
most images that are present is unknown.

No claim was tested. **How the analysis would differ from the paper's.** The paper counted cells and
plaques by eye in regions drawn from an affine alignment of each section to the Allen CCFv3. No
analysis was run here: the regions of the images are not recorded.

**What was found about the dandiset.**

- Wild-type controls (25 mice in the paper) were not imaged, by the README's account, so the
  control claim cannot be tested.
- The region of 93% of the images is not recorded, and no sampled image carries a stage position, section
  number or atlas coordinate; the region labels that exist come from follow-up rounds in 24 mice.
- The dandiset holds no counts, boundaries, laminar bins, areas or atlas alignments, the paper's
  derived data on which every figure rests.
- It has 62 mice against the paper's 61 (p120 has 9, p45 has 3 females and 5 males against 4 and 4).
- Magnification is inferred from image content; the sampled DAPI nucleus sizes are consistent with
  the stated scales of the 100X and 200X classes. Abeta42 stacks assigned 200X by rule were not
  checked.
