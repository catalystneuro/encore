The claims file (`claims/001260/10.1038-s41593-025-02190-z.yaml`) was written from the version of

Clock check. The spike times and the stimulus onsets must share a clock: below, the population
rate of PFC units in 25-ms bins around the 10-kHz tone onsets of each file, against the same
with the onsets shifted by 1.5 s (a time with no tone). A response locked to the true onsets, and
not to the shifted ones, shows the clocks agree.

```python
def tone_contrast(r, shift):
    """Mean rate of all units 0 to 100 ms after the onsets minus 100 to 0 ms before, in Hz per unit."""
    if r["tones"] is None:
        return np.nan
    f = P.open_asset(r["asset_id"])
    st = f["units/spike_times"][:]
    f.close()
    t = r["tones"] + shift
    st.sort()
    post = np.searchsorted(st, t + 0.1) - np.searchsorted(st, t)
    pre = np.searchsorted(st, t) - np.searchsorted(st, t - 0.1)
    return (post.sum() - pre.sum()) / len(t) / 0.1 / len(r["units"])

sample = [r for r in recs if r["tones"] is not None][::6]
cc = pd.DataFrame([{"session": r["session"], "true onsets": tone_contrast(r, 0.0),
                    "shifted 1.5 s": tone_contrast(r, 1.5)} for r in sample])
print("Post minus pre rate (Hz per unit) around 10-kHz tone onsets, every sixth file with tones:")
print(cc.round(3).to_string(index=False))
print("files with a larger change at the true onsets than at the shifted ones:",
      int((cc["true onsets"].abs() > cc["shifted 1.5 s"].abs()).sum()), "of", len(cc))
```

In every sampled file the rate change at the true onsets is larger than at onsets shifted by
1.5 s (an increase in nine, a decrease in one somatosensory recording), so spikes and onsets
share a clock there. The other files with tones are consistent with it: their units enter the
tone categories of fig5b, whose responses peak tens of ms after the onsets.

## dataset-ki-counts

> Dataset KI holds 24,248 single units from 99 Neuropixels recordings in awake head-fixed mice,
> 52% of them in the PFC and 48% in other brain regions.

Readings of the unit selection (claims file, G3): (a) the dataset's own quality metrics under
the Methods criteria, read as column 0 (ISI violation) below 0.1 or column 1 (false positive
rate) below 0.1; (b) all units. Each also needs the Methods' epoch rule. The count of 24,248 is
the one of the Methods' Unit quality control, which requires "a sufficient number of ISIs (≥4)
in at least 12 separate 3-s epochs" and does not restrict them to active epochs; the Methods'
Spontaneous firing metrics read "at least six spikes (corresponding to at least four ISIs)". The
main rule here takes both passages: at least 6 spikes in at least 12 spontaneous epochs, active or
sleep-like (epochs with licking or saturation excluded). At least 5 spikes, and only active
epochs (the claims header's rule for the metrics, which the analyses of the later claims keep),
are shown as variants, as is the paper's own quality flag from Zenodo.

```python
def n_valid_with(recs_, min_spikes, states):
    out = []
    for r in recs_:
        m = np.isin(r["state"], states)
        out.append((r["cnt"][:, m] >= min_spikes).sum(1))
    return np.concatenate(out)

SPONT = ["active", "near_sleep", "sleep"]
for D, rr in [(U, recs), (U473, recs473)]:
    D["n_qc"] = n_valid_with(rr, 6, SPONT)        # main: Unit quality control passage, 6 spikes = "4 ISIs"
    D["n_qc5"] = n_valid_with(rr, 5, SPONT)
    D["n_act5"] = n_valid_with(rr, 5, ["active"])
UK = pd.concat([U, U473], ignore_index=True)
UK = UK.merge(Z[["session", "unit_id", "z_quality", "z_wquality"]], on=["session", "unit_id"], how="left")
UK["z_pass"] = (UK.z_quality == 1) & (UK.z_wquality == 1)
rec_counts = {"001260": len(recs), "001260 + 000473": len(recs) + len(recs473)}

def count_table(D, label, col):
    rows = []
    for g3, m in [("(a) ISI violation < 0.1", D.q_isi < 0.1), ("(a) false positive rate < 0.1", D.q_fp < 0.1),
                  ("(b) all units", np.ones(len(D), bool))]:
        k = D[m & (D[col] >= P.MIN_EPOCHS)]
        rows.append({"data": label, "reading": g3, "units": len(k), "PFC units": int(k.is_pfc.sum()),
                     "PFC %": round(100 * k.is_pfc.mean(), 1)})
    return rows

print("Main rule (>= 6 spikes in >= 12 spontaneous epochs, active or sleep-like):")
print(pd.DataFrame(count_table(U, "001260", "n_qc") + count_table(UK, "001260 + 000473", "n_qc")).to_string(index=False))
print("recordings (probes):", rec_counts, "| expected 24,248 ± 2,425 units, 52 ± 5.2 % PFC, 99 recordings, 12,674 ± 1,267 PFC units")

var = []
for lab, D in [("001260", U), ("001260 + 000473", UK)]:
    for col, ms, sl in [("n_qc", 6, "all spontaneous"), ("n_qc5", 5, "all spontaneous"),
                        ("n_act5", 5, "active"), ("n_valid", 6, "active")]:
        k = D[D[col] >= P.MIN_EPOCHS]
        var.append({"data": lab, "min spikes": ms, "epochs": sl, "units": len(k),
                    "PFC %": round(100 * k.is_pfc.mean(), 1), "PFC units": int(k.is_pfc.sum())})
k = UK[UK.z_pass & (UK.n_qc >= P.MIN_EPOCHS)]
var.append({"data": "001260 + 000473, Zenodo quality flags", "min spikes": 6, "epochs": "all spontaneous",
            "units": len(k), "PFC %": round(100 * k.is_pfc.mean(), 1), "PFC units": int(k.is_pfc.sum())})
print("\nReading (b) under each epoch rule (the last row adds the paper's own quality flags from Zenodo):")
print(pd.DataFrame(var).to_string(index=False))
```

**Result: reproduced.** On 001260 + 000473 the recordings are the paper's 99, and under the main rule (reading b, at least 6 spikes in 12 spontaneous epochs) there are 24,898 units (24,248 ± 2,425), 52.4% in the PFC (52 ± 5.2) and 13,040 PFC units (12,674 ± 1,267). Every other epoch rule under reading (b) also falls within tolerance: 25,929, 23,168 and 22,091 units, 52.3%, 51.9% and 52.1% PFC, 13,559, 12,026 and 11,509 PFC units; with the paper's own quality flags from Zenodo 24,270 units remain. Reading (a) of the files' undescribed quality columns gives far fewer (6,561 or 16,545). 001260 alone holds 74 recordings and 17,205 such units, 47.3% in the PFC.

## fig1h-waveform-classes

> Peak-to-trough durations are bimodal, splitting units into 19,186 ww units (81.1%) and 4,200
> nw units (17.7%), with 862 intermediate units left unclassified.

```python
def wf_row(lab, rule, k):
    n = k.utype.value_counts()
    tot, cls = len(k), n.get("ww", 0) + n.get("nw", 0)
    return {"data": lab, "selection": rule, "ww": n.get("ww", 0), "nw": n.get("nw", 0), "unclassified": n.get("na", 0),
            "ww % of all": round(100 * n.get("ww", 0) / tot, 1), "nw % of all": round(100 * n.get("nw", 0) / tot, 1),
            "ww % of ww+nw": round(100 * n.get("ww", 0) / cls, 1), "nw % of ww+nw": round(100 * n.get("nw", 0) / cls, 1)}

wf = []
for lab, D in [("001260", U), ("001260 + 000473", UK)]:
    for rule, col in [("main: >= 6 spikes, 12 spontaneous epochs", "n_qc"), (">= 5 spikes, 12 spontaneous epochs", "n_qc5"),
                      (">= 6 spikes, 12 active epochs", "n_valid")]:
        for g3, m in [("(b) all units", np.ones(len(D), bool)), ("(a) false positive rate < 0.1", D.q_fp < 0.1)]:
            wf.append(wf_row(lab, f"{rule}, {g3}", D[m & (D[col] >= P.MIN_EPOCHS)]))
        if lab != "001260":
            wf.append(wf_row(lab, f"{rule}, Zenodo quality flags", D[D.z_pass & (D[col] >= P.MIN_EPOCHS)]))
print(pd.DataFrame(wf).to_string(index=False))
print("expected: ww 19,186 ± 1,919, nw 4,200 ± 420, unclassified 862 ± 86, ww 81.1 ± 8.1 %, nw 17.7 ± 1.8 %")
k = UK[UK.n_qc >= P.MIN_EPOCHS]
h, e = np.histogram(k.ptt, np.arange(0, 1.6, 0.02))
print("\nHistogram of peak-to-trough (ms, 0.02-ms bins), 001260 + 000473, main selection, reading (b); units beyond 1.6 ms:",
      int((k.ptt >= 1.6).sum()))
left = e[:-1]
m_lo, m_hi, m_mid = left < 0.38, left >= 0.43, (left >= 0.3) & (left < 0.5)
lo_peak, hi_peak = left[m_lo][np.argmax(h[m_lo])], left[m_hi][np.argmax(h[m_hi])]
valley = left[m_mid][np.argmin(h[m_mid])]
print(f"mode below 0.38 ms at {lo_peak:.2f} ms ({h[m_lo].max()} units), mode above 0.43 ms at {hi_peak:.2f} ms ({h[m_hi].max()}),"
      f" minimum between 0.30 and 0.50 ms at {valley:.2f} ms ({h[m_mid].min()})")
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(e[:-1], h, width=0.02, align="edge", color="0.4")
ax.axvline(0.38, color="tab:blue"); ax.axvline(0.43, color="tab:red")
ax.set_xlabel("peak-to-trough (ms)"); ax.set_ylabel("units"); ax.set_title("001260 + 000473, main selection")
plt.show()
```

**Result: reproduced** (under the main rule, an assumption). On 001260 + 000473 with at least 6 spikes in 12 spontaneous epochs, there are 19,731 ww units (19,186 ± 1,919), 4,247 nw (4,200 ± 420) and 920 unclassified (862 ± 86); ww are 79.2% and nw 17.1% of all units (82.3% and 17.7% of ww plus nw), within 81.1 ± 8.1 and 17.7 ± 1.8. The readings of the epoch rule differ in one count each: with at least 5 spikes the unclassified units are 951, just above 948; with active epochs only, the ww units are 17,239, just below 17,267. With the paper's Zenodo quality flags all three counts fall within tolerance (19,182, 4,181, 907). The histogram has a mode at 0.26 ms (475 units) and one at 0.64 ms (1,340) with a minimum at 0.36 ms (298).

## fig1i-nw-higher-metrics

> Narrow-waveform units have significantly higher firing rates, burstiness and memory than
> wide-waveform units.

Mixed model Metric ~ 1 + Unittype + (1|Mouse) per metric (Methods, equation 6), fitted with
statsmodels (REML). The paper's Kenward-Roger F test is not available there; the Wald test of the
unit-type coefficient is used (with thousands of units and 40 mice the degrees of freedom are
large either way). With one comparison per metric Tukey's correction changes nothing.

```python
cl = U[(U.n_valid >= P.MIN_EPOCHS) & U.utype.isin(["ww", "nw"])].dropna(subset=["log10FR", "B", "M"])
rows = []
for m in ["log10FR", "B", "M"]:
    f, r2 = P.mixed(cl, f'{m} ~ C(utype, Treatment("ww"))')
    rows.append({"metric": m, "nw - ww estimate": f.params.iloc[1], "p": f.pvalues.iloc[1],
                 "median ww": cl[cl.utype == "ww"][m].median(), "median nw": cl[cl.utype == "nw"][m].median(),
                 "n ww": int((cl.utype == "ww").sum()), "n nw": int((cl.utype == "nw").sum())})
print(pd.DataFrame(rows).to_string(index=False))
print("mice:", cl.subject.nunique())
print("\nPer setting (median nw minus median ww):")
print(cl.groupby(["setting", "utype"])[["log10FR", "B", "M"]].median().unstack().pipe(
    lambda d: pd.DataFrame({m: d[(m, "nw")] - d[(m, "ww")] for m in ["log10FR", "B", "M"]})).round(3).to_string())
# variant: the code drops units with log10FR < 0.1
cv = cl[cl.log10FR >= 0.1]
print("\nVariant, units with log10FR >= 0.1 only:")
for m in ["log10FR", "B", "M"]:
    f, _ = P.mixed(cv, f'{m} ~ C(utype, Treatment("ww"))')
    print(f"  {m}: nw - ww = {f.params.iloc[1]:.4f}, p = {f.pvalues.iloc[1]:.2g}, n = {len(cv)}")
```

**Result: reproduced.** nw units have a higher log10FR (+0.245), B (+0.051) and M (+0.042) than ww units, p ≤ 4.5e-84 for each, in 12,112 ww and 2,949 nw units from 41 mice; the medians differ in the same direction in every setting, and the differences hold when units with log10FR < 0.1 are dropped (p ≤ 7.2e-66).

## extfig1d-sleep-like-differ

> Firing metrics in sleep-like epochs differ from those in active spontaneous epochs, which is
> why sleep-like epochs are excluded.

ww units with at least 12 active epochs (active sample) and ww units with at least 12 sleep-like
epochs of at least 6 spikes (sleep-like sample), Metric ~ 1 + EpochType + (1|Mouse) (Methods,
equation 7). Sleep-like periods from the Methods' detector per session (main); the same detector
per probe, the code's detector, the main detector without sleep-like periods in the sessions the
paper's own detection files mark as not using it (`off_used = no`), and the paper's own
sleep-like periods (Zenodo `quiet_active_detection.zip`, an outside input) are variants.

```python
def epoch_type_models(Ux):
    ww_ = Ux[Ux.utype == "ww"]
    a = ww_[ww_.n_valid >= P.MIN_EPOCHS][["subject", "log10FR", "B", "M"]].assign(epoch="active")
    s = ww_[ww_.n_valid_sleep >= P.MIN_EPOCHS][["subject", "log10FR_sleep", "B_sleep", "M_sleep"]]
    s = s.rename(columns=lambda c: c.replace("_sleep", "")).assign(epoch="sleep")
    L = pd.concat([a, s]).dropna()
    rows = []
    for m in ["log10FR", "B", "M"]:
        f, _ = P.mixed(L, f"{m} ~ C(epoch)")
        rows.append({"metric": m, "sleep - active": f.params.iloc[1], "p": f.pvalues.iloc[1],
                     "median active": a[m].median(), "median sleep": s[m].median()})
    return pd.DataFrame(rows), len(a), len(s)

res, na_, ns_ = epoch_type_models(U)
print(f"Methods detector: {na_} ww units with >= 12 active epochs, {ns_} with >= 12 sleep-like epochs "
      "(paper, all of dataset KI: 19,186 and 11,599)")
print(res.to_string(index=False))
def with_sleep(recs_, fn):
    """copies of the records with the sleep-like periods fn(r) and the epoch states recomputed"""
    out = []
    for r in recs_:
        r2 = dict(r)
        r2["sleep"] = fn(r)
        ep = r2["epochs"]
        bad = P.overlaps(ep, r2["lick"]) | P.overlaps(ep, r2["sat"])
        sl = r2["sleep"]
        if sl is None or len(sl) == 0:
            s_, near = np.zeros(len(ep), bool), np.zeros(len(ep), bool)
        else:
            s_, near = P.overlaps(ep, sl), P.overlaps(ep, sl, pad=1.0)
        r2["state"] = np.where(bad, "excluded", np.where(s_, "sleep", np.where(near, "near_sleep", "active")))
        out.append(r2)
    return out

def paper_flag(r):
    """'yes' when a probe of the session has the paper's off_used = yes, 'no' when its files all say no"""
    day = r["session"].rsplit("-", 1)[0]
    fl = [q["used"] for k, q in PQ.items() if k.rsplit("-", 1)[0] == day]
    return "yes" if "yes" in fl else ("no" if fl else "no file")

def paper_periods(r):
    """the paper's own sleep-like periods: the probe's file, else another probe's of the session"""
    day = r["session"].rsplit("-", 1)[0]
    k = r["session"] if r["session"] in PQ else next((k for k in PQ if k.rsplit("-", 1)[0] == day), None)
    return None if k is None else PQ[k]["quiet"]

VAR = {}
for det in ["methods", "code"]:
    rr_ = P.classify_epochs([dict(r) for r in recs], detector=det)
    VAR[det] = (rr_, P.unit_table(rr_))
    res_v, na_v, ns_v = epoch_type_models(VAR[det][1])
    print(f"\nVariant, {'the Methods detector per probe (round 1)' if det == 'methods' else 'the code detector per probe'}: "
          f"{na_v} and {ns_v} units")
    print(res_v.to_string(index=False))
for det, fn in [("flags", lambda r: None if paper_flag(r) == "no" else r["sleep"]), ("paper", paper_periods)]:
    rr_ = with_sleep(recs, fn)
    VAR[det] = (rr_, P.unit_table(rr_))
    res_v, na_v, ns_v = epoch_type_models(VAR[det][1])
    print(f"\nVariant, {'the main detector with none where the paper says off_used = no' if det == 'flags' else 'the paper own sleep-like periods (Zenodo)'}: "
          f"{na_v} and {ns_v} units")
    print(res_v.to_string(index=False))
```

**Result: partially reproduced** (under the main detector, an assumption). With sleep-like periods detected per session, sleep-like epochs differ from active ones in B (+0.031, p = 5.3e-63) and M (-0.011, p = 2.3e-12) but not in log10FR (+0.008, p = 0.14), from 12,112 active and 8,405 sleep-like samples: two of three metrics. The same holds when the sessions the paper marks as not using detection are left without sleep-like periods (log10FR p = 0.41). With the detector per probe, the code's detector, or the paper's own sleep-like periods from Zenodo all three metrics differ (log10FR p = 4.4e-06, 0.0012 and 0.0027), the log10FR difference changing sign between them (-0.024, +0.018, +0.019); B and M differ in the same direction under every detector.

## extfig1-epoch-counts

> Sleep-like episodes occupy a median 6% (IQR 0 to 18%) of recording time; sessions hold a median
> 227 (IQR 172 to 284) spontaneous active epochs and 19 (IQR 1 to 84) sleep-like epochs.

Readings of "recording session": one probe recording (file), or one session of up to three
probes (subject and date; time fraction and epochs from the session's first probe, as all
probes of a session share onsets). The recording time is from the first to the last spike.
Detectors: the Methods detector per session (main), per probe (variant, round 1's main), the
code's detector (variant); and two variants that use the paper's own detection on Zenodo
(`quiet_active_detection.zip`, an outside input): the main detector, but no sleep-like period in
the sessions whose probes the paper's files mark as not using off-period detection
(`off_used = no`), and the paper's own sleep-like periods.

The second cell below prints the per-probe check of the detector asked for in review: the
number of cortical units and their mean rate, the share of time marked sleep-like by each
detector, the share it marks in a Poisson population with the same pooled rate and no off periods
(null), the median EMG RMS (1-s windows, `processing/ecephys/EMG`) in sleep-like time over that in
the rest of the recording, and the paper's own flag and share.

```python
def epoch_table(recs_):
    rows = []
    for r in recs_:
        rows.append({"session": r["session"], "day": r["session"].split("-")[0],
                     "sleep %": 100 * P.sleep_fraction(r["sleep"], r["t0"], r["t1"]),
                     "active epochs": int((r["state"] == "active").sum()), "sleep-like epochs": int((r["state"] == "sleep").sum())})
    return pd.DataFrame(rows)

def iqr_str(x):
    q = np.percentile(x, [25, 50, 75])
    return f"{q[1]:.1f} (IQR {q[0]:.1f} to {q[2]:.1f})"

DET = {"Methods detector per session (main)": recs, "Methods detector per probe (round 1)": VAR["methods"][0],
       "code's detector per probe": VAR["code"][0],
       "main, none where the paper's files say off_used = no": VAR["flags"][0],
       "the paper's own sleep-like periods (Zenodo)": VAR["paper"][0]}
UDET = {"Methods detector per session (main)": U, "Methods detector per probe (round 1)": VAR["methods"][1],
        "code's detector per probe": VAR["code"][1], "main, none where the paper's files say off_used = no": VAR["flags"][1],
        "the paper's own sleep-like periods (Zenodo)": VAR["paper"][1]}
for det, rr in DET.items():
    et = epoch_table(rr)
    per_day = et.groupby("day").first()
    print(det)
    for nm, d in [("per probe recording (n=%d)" % len(et), et), ("per session (n=%d)" % len(per_day), per_day)]:
        print(f"  {nm}: sleep-like time {iqr_str(d['sleep %'])} %, active epochs {iqr_str(d['active epochs'])}, "
              f"sleep-like epochs {iqr_str(d['sleep-like epochs'])}")
    if det in UDET:
        print("  ww units with >= 12 sleep-like epochs of >= 6 spikes:",
              int((UDET[det].query("utype == 'ww'").n_valid_sleep >= P.MIN_EPOCHS).sum()))
print("expected: 6 ± 0.6 %, 227 ± 22.7 active, 19 ± 1.9 sleep-like epochs, 11,599 ± 1,160 ww units (all of dataset KI)")
pq = pd.DataFrame([{"rec": k, "used": q["used"], "quiet %": 100 * q["quietdur"] / q["recdur"]} for k, q in PQ.items()])
print(f"\nThe paper's own detection files: {len(pq)} recordings of both dandisets, off_used = no in {int((pq.used == 'no').sum())};"
      f" quiet time over recording time (the files' own durations) {iqr_str(pq['quiet %'])} %;"
      f" in the {int(pq.rec.isin([r['session'] for r in recs]).sum())} of 001260 {iqr_str(pq[pq.rec.isin([r['session'] for r in recs])]['quiet %'])} %")
et = epoch_table(recs)
print("\nRecordings with more than half of their time sleep-like (main):", int((et["sleep %"] > 50).sum()), "of", len(et))
et473 = epoch_table(recs473)
print("000473 (25 aversion recordings), main detector:", "sleep-like time", iqr_str(et473["sleep %"]),
      "%, active epochs", iqr_str(et473["active epochs"]), ", sleep-like epochs", iqr_str(et473["sleep-like epochs"]))
```

```python
def emg_ratio(r, sl):
    """median EMG RMS in 1-s windows inside sl over that outside, within the recording"""
    e = r["emg"]
    if e is None or sl is None or len(sl) == 0:
        return np.nan
    t = e[:, 0]
    ins = P.overlaps(np.c_[t, t + 1.0], sl)
    ok = (t > r["t0"]) & (t + 1 < r["t1"])
    if ins[ok].sum() < 10 or (~ins[ok]).sum() < 10:
        return np.nan
    return float(np.median(e[ok & ins, 1]) / np.median(e[ok & ~ins, 1]))

rows = []
for r, rp, rc in zip(recs, VAR["methods"][0], VAR["code"][0]):
    q = PQ.get(r["session"])
    rows.append({"session": r["session"], "cortical units": r["n_cortical"], "rate (Hz)": round(r["cortical_rate"], 2),
                 "probe %": 100 * P.sleep_fraction(r["sleep_own"], r["t0"], r["t1"]),
                 "probe null %": 100 * P.null_sleep_fraction(r["ccounts"], r["n_cortical"], r["t0"], r["t1"], "methods"),
                 "code %": 100 * P.sleep_fraction(r["sleep_own_code"], r["t0"], r["t1"]),
                 "session units": r["sleep_units"], "session null %": 100 * r["sleep_null"],
                 "main %": 100 * P.sleep_fraction(r["sleep"], r["t0"], r["t1"]),
                 "EMG ratio probe": emg_ratio(r, r["sleep_own"]), "EMG ratio main": emg_ratio(r, r["sleep"]),
                 "paper off_used": paper_flag(r), "paper quiet %": np.nan if q is None else 100 * q["quietdur"] / q["recdur"]})
chk = pd.DataFrame(rows).sort_values("cortical units")
print(chk.round(2).to_string(index=False))
print("\nEMG ratio (below 1: lower muscle tone in sleep-like time): per-probe detection", iqr_str(chk["EMG ratio probe"].dropna()),
      "| main", iqr_str(chk["EMG ratio main"].dropna()), f"({chk['EMG ratio main'].notna().sum()} recordings with both)")
for lo, hi in [(0, 75), (75, 150), (150, 1000)]:
    s_ = chk[(chk["cortical units"] >= lo) & (chk["cortical units"] < hi)]
    print(f"probes with {lo} to {hi - 1} cortical units: {len(s_)}; per-probe sleep-like % {iqr_str(s_['probe %'])};"
          f" its null % {iqr_str(s_['probe null %'])}")
for thr in [0.001, 0.01, 0.05]:
    print(f"sessions without detection if the null threshold were {100 * thr:g} %:",
          len({r['session'].rsplit('-', 1)[0] for r in recs if r['sleep_null'] >= thr}))
yes = chk["paper off_used"] == "yes"
print("main sleep-like % where the paper used off-period detection", iqr_str(chk[yes]["main %"]),
      "| where it did not", iqr_str(chk[~yes]["main %"]))
print("Spearman r between main and the paper's quiet %, recordings with a file:",
      round(stats.spearmanr(chk["main %"], chk["paper quiet %"], nan_policy="omit")[0], 2))
```

**Result: not reproduced.** With the main detector, sleep-like periods take a median 9.9% of a probe recording (IQR 2.7 to 20.3%) against 6% (0 to 18%); recordings hold a median 215 active epochs (227 ± 22.7, within tolerance) but 73.5 sleep-like epochs (19 expected), and 8,405 ww units have at least 12 sleep-like epochs (11,599 ± 1,160). Per session the values are similar (10.5%, 207, 73.5). No detector reading reproduces the values: per probe 20.0%, 180 and 125; the code's detector 8.7%, 203 and 83.5; even the paper's own sleep-like periods from Zenodo give 1.5%, 248 and 21.5 on these recordings. The paper's own detection files give a median quiet time of 5.3% (IQR 0.7 to 17.2%) of their recording time over 73 recordings, close to the claim, and mark 19 recordings as not using off-period detection; where the paper used it, the main detector marks a median 15.9% of the time, where it did not 3.1%, and the two agree in rank (Spearman 0.65).

```python
def fig2f_parts(bc_, cat):
    M = sum(P.transitions(bc_, cat, 8).values())
    Pm = M / M.sum(1, keepdims=True)
    X = P.coincidence(M)
    low, high = [0, 1, 2, 6, 7], [3, 4, 5]
    d, dx = np.diag(Pm), np.diag(X)
    part1 = d[low].mean() < d[high].mean()
    part1x = dx[low].mean() < dx[high].mean()
    part2 = []
    for grp in ([0, 1, 2], [6, 7]):
        for i in grp:
            others = [j for j in grp if j != i]
            out = [j for j in range(8) if j not in grp]
            part2.append(Pm[i, others].sum() > Pm[i, out].max())
    return Pm, X, d, dx, part1, part1x, part2

for cat in ["cat_a", "cat_b"]:
    Pm, X, d, dx, p1, p1x, p2 = fig2f_parts(bc, cat)
    print(f"G2 reading {cat[-1]}: diagonal of P by category 1..8: {np.round(d, 3)}")
    print(f"   mean diagonal low-rate (1-3,7,8) {d[[0,1,2,6,7]].mean():.3f} vs high-rate (4-6) {d[[3,4,5]].mean():.3f} -> part 1 {p1};"
          f" coincidence diagonal {np.round(dx, 3)} -> {p1x}")
    print(f"   part 2 (within-group probability above any single outside category), categories 1,2,3,7,8: {p2}")
    print(pd.DataFrame(Pm, index=range(1, 9), columns=range(1, 9)).round(2).to_string())
```

**Result: partially reproduced,** under both G2 readings. Part 2 fails: within-group transitions exceed any single outside category for categories 1, 2 and 3 under both readings, but not for 7 and 8 (G2 a) or 7 (G2 b). ## fig4cd-dataroi-enrichment

> Low-rate, regular-firing categories 1 and 3 are significantly enriched in dataROIs in ILA and PL,
> while low-rate, bursty categories 7 and 8 are enriched in dataROIs in MOs and ORBl, the ORBl
> enrichment pinpointed to its most ventral dataROI.

Reference (a) all deep PFC ww units of the parcellation; (b) all units as for regions (the
dataROIs plus one rest population of every other unit of the region analysis). Groups by
setting (G1 a and b), both G2 readings, dataROIs from the 160-240 parcellation.

```python
def droi_escores(fm_, roicol, ref, g1, g2, seed, pool=None):
    """E-scores per ROI; reference (a) the PFC units of fm_, (b) these and the rest of pool (default sel)"""
    pool = sel if pool is None else pool
    if ref == "a":
        pops, cats = fm_[roicol].values, fm_["cat_" + g2].values
        grp = fm_.setting.replace({"Opto": "Passive"}).values if g1 == "a" else fm_.setting.values
        plist = sorted(pd.unique(pops))
    else:
        rest = pool[~pool.set_index(["session", "unit_id"]).index.isin(fm_.set_index(["session", "unit_id"]).index)]
        both = pd.concat([fm_.assign(pp=fm_[roicol]), rest.assign(pp="rest")])
        pops, cats = both.pp.values, both["cat_" + g2].values
        grp = both.setting.replace({"Opto": "Passive"}).values if g1 == "a" else both.setting.values
        plist = sorted(pd.unique(fm_[roicol])) + ["rest"]
    E, Pv, _ = P.escore(pops, cats, grp, 1000, seed=seed, pops=plist, attrs=range(1, 9))
    keep = [p for p in plist if p != "rest"]
    return E.loc[keep], Pv.loc[keep]

DROI = {}
for ref in ["a", "b"]:
    for g1 in ["a", "b"]:
        for g2 in ["a", "b"]:
            E, Pv = droi_escores(fm, "droi", ref, g1, g2, seed=21)
            DROI[(ref, g1, g2)] = (E, Pv)
            sp = (E > 0) & (Pv < 0.05)
            sub = lambda r, cs: any(sp.loc[i, cs].any() for i in E.index if i.startswith(r + "-"))
            ok = sub("ILA", [1, 3]) and sub("PL", [1, 3]) and sub("MOs", [7, 8]) and sub("ORBl", [7, 8])
            orbl = [i for i in E.index if i.startswith("ORBl-")]
            best = E.loc[orbl, [7, 8]].sum(1).idxmax() if orbl else None
            print(f"reference {ref}, G1 {g1}, G2 {g2}: ILA 1/3 {sub('ILA', [1, 3])}, PL 1/3 {sub('PL', [1, 3])}, "
                  f"MOs 7/8 {sub('MOs', [7, 8])}, ORBl 7/8 {sub('ORBl', [7, 8])} -> {ok}; ORBl dataROI with the highest 7+8 E-score: {best}")
E, Pv = DROI[("a", "a", "b")]
print("\nreference a, G1 a, G2 b (* P < 0.05):")
print(E.round(1).astype(str).where(Pv >= 0.05, E.round(1).astype(str) + "*").to_string())
```

**Result: reproduced.** On the 24 dataROIs of 001260, at least one dataROI in ILA and in PL is significantly enriched in category 1 or 3, and at least one in MOs and in ORBl in category 7 or 8, under both references, both groupings and both G2 readings. ORBl is a single dataROI here (386 units), so whether the enrichment is in its most ventral part is not tested.

## fig4e-activity-modules

> Clustering dataROIs by enrichment profile gives activity modules; ILA forms a homogeneous
> ILA-specific module bordering a similar module in adjacent PL and ACAd, while other modules
> cross cytoarchitectural boundaries and occur in several places.

Main: the 24 dataROIs of 001260 (fig4b). Two variants, asked for in review: (1) the paper's own
42 dataROI polygons from Zenodo (`flatmap_PFC_ntesselated_obeyRegions_res200.h5`), each holding
the 001260 units inside it and named by the subregion of most of its units, those with at least
20 units; (2) all of dataset KI, the 001260 units plus the 000473 deep PFC ww units projected on
the ww SOM trained here (their metrics as for 001260), parcellated as in fig4b, reference (a).

```python
from scipy.cluster.hierarchy import fcluster, linkage
for key, (E, Pv) in DROI.items():
    Zs = (E - E.mean(0)) / E.std(0)
    mod = pd.Series(fcluster(linkage(Zs.values, "ward"), 5, "maxclust"), index=E.index)
    subr = mod.index.str.split("-").str[0]
    ila = mod[subr == "ILA"]
    one = ila.nunique() == 1
    m_ila = ila.iloc[0] if one else None
    rd_a = one and (subr[mod == m_ila] == "ILA").all()
    rd_b = one and (subr[mod == m_ila] == "ILA").mean() > 0.5
    multi = sum(len(set(subr[mod == m])) > 1 for m in mod.unique() if m != m_ila)
    print(f"ref {key[0]} G1 {key[1]} G2 {key[2]}: ILA dataROIs {len(ila)} in one module {one}; ILA-only (a) {rd_a}, "
          f"mostly ILA (b) {rd_b}; other modules spanning subregions {multi}")
    if key == ("a", "a", "b"):
        print(pd.crosstab(mod.values, subr, rownames=["module"], colnames=["subregion"]).to_string())
```

```python
def module_check(E, k=5):
    Zs = (E - E.mean(0)) / E.std(0).replace(0, 1)
    mod = pd.Series(fcluster(linkage(Zs.fillna(0).values, "ward"), k, "maxclust"), index=E.index)
    subr = mod.index.str.split("-").str[0]
    ila = mod[subr == "ILA"]
    one = ila.nunique() == 1
    m_ila = ila.iloc[0] if one else None
    rd_a = bool(one and (subr[mod == m_ila] == "ILA").all())
    rd_b = bool(one and (subr[mod == m_ila] == "ILA").mean() > 0.5)
    multi = sum(len(set(subr[mod == m])) > 1 for m in mod.unique() if m != m_ila)
    comp = pd.Series(subr[mod == m_ila]).value_counts().to_dict() if one else None
    return len(ila), one, rd_a, rd_b, multi, comp, mod

# variant 1: the paper's 42 dataROI polygons
fm["droi_z"] = P.in_polygons(fm[["U", "V"]].values, paper_drois)
maj = fm[fm.droi_z.notna()].groupby("droi_z").polyreg.agg(lambda s: s.value_counts().index[0])
fm["droi_zn"] = [None if k is None else f"{maj[k]}-z{k}" for k in fm.droi_z]
zc = fm.droi_zn.value_counts()
fmz = fm[fm.droi_zn.isin(zc[zc >= 20].index)].copy()
print(f"Paper's dataROIs: {fm.droi_z.notna().sum()} of {len(fm)} units inside one; {fm.droi_z.nunique()} polygons hold units,"
      f" {len(zc[zc >= 20])} with >= 20 (units per dataROI {rng_str(zc[zc >= 20])}); by subregion",
      pd.Series(zc[zc >= 20].index.str.split("-").str[0]).value_counts().to_dict())
DROI_Z = {}
for ref in ["a", "b"]:
    for g1 in ["a", "b"]:
        for g2 in ["a", "b"]:
            E, Pv = droi_escores(fmz, "droi_zn", ref, g1, g2, seed=22)
            DROI_Z[(ref, g1, g2)] = (E, Pv)
            n_ila, one, rd_a, rd_b, multi, comp, mod = module_check(E)
            print(f"  ref {ref} G1 {g1} G2 {g2}: ILA dataROIs {n_ila} in one module {one}; ILA-only (a) {rd_a}, "
                  f"mostly ILA (b) {rd_b}; ILA module {comp}; other modules spanning subregions {multi}")

# variant 2: all of dataset KI (001260 + 000473), 000473 units projected on the ww SOM
k4 = k473[k473.polyreg.notna()].dropna(subset=["log10FR", "B", "M"]).copy()
k4["cat_a"], k4["cat_b"] = P.project_categories(fit_ww, k4)
cols = ["session", "unit_id", "U", "V", "polyreg", "setting", "cat_a", "cat_b"]
kk2 = pd.concat([fm[cols], k4[cols]], ignore_index=True)
kk2["droi"] = P.parcellate(kk2[["U", "V"]].values, kk2.polyreg.values, bounds=(160, 240))
print(f"\nAll of dataset KI: {len(kk2)} units ({len(k4)} from 000473), {kk2.droi.nunique()} dataROIs,"
      " ILA dataROIs", sorted(d for d in kk2.droi.unique() if d.startswith("ILA-")))
for g1 in ["a", "b"]:
    for g2 in ["a", "b"]:
        E, Pv = droi_escores(kk2, "droi", "a", g1, g2, seed=23)
        n_ila, one, rd_a, rd_b, multi, comp, mod = module_check(E)
        print(f"  ref a G1 {g1} G2 {g2}: ILA dataROIs {n_ila} in one module {one}; ILA-only (a) {rd_a}, "
              f"mostly ILA (b) {rd_b}; ILA module {comp}; other modules spanning subregions {multi}")
```

latency of their mean trace, the category with the lowest mean peak last (Methods: "the flat or
negative category last").

```python
from concurrent.futures import ThreadPoolExecutor
trecs = [r for r in recs if r["tones"] is not None]
with ThreadPoolExecutor(8) as ex:
    TT = list(ex.map(P.tone_traces, trecs))
T = pd.concat([pd.DataFrame({"session": r["session"], "unit_id": r["units"].unit_id.values, "rate_all": o["rate_all"],
                             "rate_pre": o["rate_pre"], "n_tone_epochs": o["n_tones"]}) for r, o in zip(trecs, TT)], ignore_index=True)
X = np.vstack([o["trace"] for o in TT])
T = T.merge(U[["session", "unit_id", "utype", "is_pfc", "deep", "area", "region", "setting", "subject", "cortical"]], on=["session", "unit_id"])
print("files with 10-kHz tones:", len(trecs), "| tone response epochs per file:", rng_str([o["n_tones"] for o in TT]),
      "(paper: 188, IQR 86 to 228)")
okT = ((T.rate_all >= 0.1) & (T.rate_pre >= 0.1) & np.isfinite(X).all(1) & T.utype.isin(["ww", "nw"])).values
print("units:", len(T), "| kept (rates >= 0.1 Hz, ww or nw):", int(okT.sum()))
T, X = T[okT].reset_index(drop=True), X[okT]
Xc = X - X.mean(0)
w, V = np.linalg.eigh(np.cov(Xc.T))
o = np.argsort(w)[::-1][:8]
pcs = V[:, o].T
print(f"variance explained by 8 components: {w[o].sum() / w.sum():.3f} (Methods: over 95%)")
Sc = X @ pcs.T
tt = np.arange(X.shape[1]) * P.PSTH_BIN
TONE = {}
for rd, m in [("a", ((T.utype == "ww") & ((~T.cortical) | T.deep)).values), ("b", (T.utype == "ww").values)]:
    Z0 = Sc[m]
    Zs = (Z0 - Z0.mean(0)) / Z0.std(0)
    som_t = P.train_som(Zs, 400)
    lab = P.ward_clusters(som_t["W"], 8)
    cl = lab[P.bmus(Zs, som_t["W"])]
    mt = np.array([X[m][cl == c].mean(0) for c in range(8)])
    peak, lat = mt.max(1), tt[mt.argmax(1)]
    flat = int(np.argmin(peak))
    order = [c for c in np.argsort(lat) if c != flat] + [flat]
    num = np.zeros(8, int)
    for i, c in enumerate(order):
        num[c] = i + 1
    D = T[m].copy()
    D["tcat"] = num[cl]
    TONE[rd] = D
    pf = D[D.is_pfc & D.deep]
    print(f"\nreading {rd}: {m.sum()} ww units trained, {len(pf)} PFC deep ww units (paper 7,184 for all of dataset KI)")
    print(pd.DataFrame({"category": num, "mean peak (z)": peak.round(2), "peak latency (ms)": (lat * 1000).round(1),
                        "units": np.bincount(cl, minlength=8), "PFC deep ww": [int((pf.tcat == num[c]).sum()) for c in range(8)]}
                       ).sort_values("category").to_string(index=False))
fig, axs = plt.subplots(1, 2, figsize=(11, 3))
for ax, rd in zip(axs, ["a", "b"]):
    D = TONE[rd]
    for c in range(1, 9):
        sel_ = (T.set_index(["session", "unit_id"]).index.isin(D[D.tcat == c].set_index(["session", "unit_id"]).index))
        ax.plot(tt * 1000, X[sel_].mean(0), label=str(c))
    ax.set_title(f"reading {rd}: mean trace per category"); ax.set_xlabel("ms after tone"); ax.set_ylabel("z")
axs[1].legend(ncol=4, fontsize=7)
plt.show()
```

## fig5b-tone-categories

> Deep-layer PFC ww tone responses cluster into eight categories ordered by peak delay; most
> units have flat or negative responses (category 8), categories 1 to 4 have sharp peaks with

ZETA test (pfcki.zeta_unit) on the PFC deep ww units of categories 1 to 7 under either reading;

```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor
# all PFC deep ww units with a tone trace (needed for fig5e as well)
need = TONE["b"][TONE["b"].is_pfc & TONE["b"].deep][["session", "unit_id"]].drop_duplicates()
by_s = {r["session"]: r for r in trecs}
args, keys = [], []
for s, g in need.groupby("session"):
    a = P.zeta_for(by_s[s], g.unit_id.values, seed=5)
    args += a
    keys += [(s, u) for u in g.unit_id.values]
t0 = time.time()
with ProcessPoolExecutor(4, mp_context=mp.get_context("fork")) as ex:
    zres = list(ex.map(P.zeta_unit, args, chunksize=8))
ZT = pd.DataFrame(keys, columns=["session", "unit_id"]).assign(zeta_p=[z[0] for z in zres], onset=[z[1] for z in zres])
print(f"ZETA on {len(ZT)} units in {time.time() - t0:.0f} s; failed or empty: {int(ZT.zeta_p.isna().sum())}")
for rd in "ab":
    D = TONE[rd][TONE[rd].is_pfc & TONE[rd].deep].merge(ZT, on=["session", "unit_id"], how="left")
    cnt = D.tcat.value_counts().sort_index()
    largest = cnt.idxmax() == 8
    rows = []
    for c in range(1, 8):
        g = D[(D.tcat == c) & (D.zeta_p < 0.01)]
        q = np.percentile(g.onset.dropna(), [25, 50, 75]) * 1000 if len(g.onset.dropna()) >= 3 else [np.nan] * 3
        rows.append({"category": c, "units": int((D.tcat == c).sum()), "ZETA P < 0.01": len(g),
                     "onset median (ms)": round(q[1], 1), "onset IQR (ms)": round(q[2] - q[0], 1)})
    R = pd.DataFrame(rows)
    iqr = R.set_index("category")["onset IQR (ms)"]
    part2 = iqr.loc[1:4].max() < iqr.loc[5:7].min()
    print(f"\nreading {rd}: category 8 is the largest: {largest} ({cnt.get(8, 0)} of {len(D)} units)")
    print(R.to_string(index=False))
    print(f"largest IQR in 1-4: {iqr.loc[1:4].max():.1f} ms, smallest IQR in 5-7: {iqr.loc[5:7].min():.1f} ms -> part 2 {part2}")
```

**How the analysis differs from the paper's.** The analyses run on 001260, which holds 74 of the
99 recordings; 000473 is added for the counts, the number of dataROIs and one variant. Units are
not filtered by the paper's quality flags, which the files lack (the flags on Zenodo pass 97% of
units). Sleep-like periods follow the Methods' detector, applied per session to the pooled
cortical units of its probes and skipped where a Poisson population of the same rate would be
marked sleep-like 1% of the time or more (a rule of this attempt; the paper describes no skipped
session); the paper also reviewed every period visually, and its own detection files mark 19
recordings as not using the detection. The main detector still marks more time (a median 9.9%)
than the paper reports (6%). With the paper's own sleep-like periods (a variant), the outcomes of
fig2g-pfc-regular-enrichment, extfig5g-nw-enrichment and fig4g-gao-category1 would differ, while
those of fig2f, extfig3f, fig3a and fig4e stay the same. The SOM, the clustering and the E-scores
are re-implemented from the Methods and the code; the SOM grid has 425 nodes, as the code's shape
rule gives. An E-score is significant at the 5th or 95th percentile of the shuffles, a 10%
two-sided test. Mixed models are fitted with statsmodels and tested by Wald tests, not
Kenward-Roger F tests. The retrained clusters cannot be tied to the paper's category numbers, and
the two readings of the claims file (G2) disagree for eight claims; projecting the units on the
paper's own trained SOM (a variant) gives categories that match the text's descriptions, and
under it four of those eight claims hold. Blocks for the stability analysis are taken from the trials
table's `Block` column or cut into 50 onsets. The ZETA onset is computed from zetapy's rate as
the first half-peak crossing.

**What was found about the dataset.** The two dandisets together hold exactly the 29,782 units
the paper had before its unit quality control, and the 66 mice; 001260 alone holds 74
recordings, 41 mice and 20,374 units. It carries no label of which units the paper kept, its
`quality` columns are undescribed, and its unit locations, flatmap positions and ww/nw labels
need the paper's Zenodo files to be matched to the paper's. The other 25 recordings are in
000473, whose metadata names another paper. Which recordings had their sleep-like detection
rejected on review is recorded only in the paper's Zenodo files.
